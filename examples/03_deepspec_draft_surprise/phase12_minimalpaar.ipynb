Die vier Praefixe mit `(Local Name)` in der **ersten Spaltenueberschrift** belegen genau

Das Modell trifft die Lesartenwahl also in seiner **eigenen Kopfzeile**. `Service (Local
Name)` heisst: in die Zellen dieser Spalte kommt der lokalisierte Name (Lesart a).
`Service` heisst Lesart b. Die Sprachwahl folgt danach fast zwangslaeufig - passend zum
fruehen Befund, dass Lesart (a) ausdruecklich gemacht 88.5 % ergibt und (b) 1.0 %.

Das erklaert auch, **warum die Routing-Suche leer ausging**: bei Zeichen 100 steht die
Festlegung als Klartext im Praefix, die zwoelf Praefixe unterscheiden sich in allem
Moeglichen, und keine kleine Expertenmenge isoliert sie. 1652 aktive Paare, kein einziges
in allen zwoelf.

## Minimalpaare statt geernteter Vielfalt

Acht Praefix-Paare, die sich **ausschliesslich** in ` (Local Name)` unterscheiden -
zeichenweise identisch bis auf diese Einfuegung:

```
A  | Service (Local Name) | Storage Limits | Pricing (Approx.) | Summary |
B  | Service | Storage Limits | Pricing (Approx.) | Summary |
```

Damit ist der Kontrast **eine Festlegung** statt zwoelf verschiedener Tabellenanfaenge.
Das erlaubt drei Dinge, die vorher nicht gingen:

1. **Gepaarter Ratentest.** Je Paar die Differenz A-B, Vorzeichentest ueber acht Paare.
   Aus der nachtraeglichen Rangbeobachtung wird ein vorregistrierter Versuch.
2. **Gepaarter Routing-Vergleich.** Je Paar die Experten, die in A feuern und in B nicht.
   Konsistent ueber alle acht Paare heisst: diese Auswahl haengt an der Festlegung.
3. **Vorzeichen-Nulltest.** Die Rollen von A und B je Paar vertauschen und neu zaehlen.
   Untergrenze 2/2^8 = **0.0078** - erreichbar, anders als die 0.080 der ungepaarten
   Fassung bei zwoelf Praefixen.

## Die Laengenkontrolle, ohne die das nichts wert waere

A ist laenger als B; ein Routing-Unterschied koennte schlicht daher kommen. Deshalb laeuft
eine zweite Familie mit derselben Bauform, aber einem Adjektiv, das die Adjektiv-Achse mit
**0/48** gemessen hat:

```
I  | Service (Short Name) | Storage Limits | ...
```

Gleiche Klammern, gleiche Wortzahl, gleiche Stelle - nur ohne Lokalisierung. Was in
**beiden** Familien konsistent auftritt, ist Laenge und Satzbau; nur die Differenz gehoert
zur Bedeutung und wird geprueft.

## Vorregistriert

**H1** A kippt haeufiger als B (gepaarter Vorzeichentest ueber acht Paare).
**H2** die konsistente A-Menge *abzueglich der inerten* zu sperren senkt die Kipprate am
staerksten kippenden A-Praefix; gleich viele haeufig aktive Experten nicht.

Die **JA-exklusive Menge** laeuft als Positivkontrolle am selben Praefix mit. Senkt sie
dort nicht, ist das Messfeld unempfindlich und jeder Nullbefund bedeutungslos.

| Verdikt | Lesart |
|---|---|
| `KOPFZEILE-IM-ROUTER` | die Lesartenwahl haengt an einer benennbaren Router-Auswahl, an einer Stelle, an der sich sonst nichts im Text unterscheidet |
| `KOPFZEILE-OHNE-WIRKUNG` | H1 faellt - die Rangbeobachtung war eine Eigenheit der geernteten Praefixe |
| `ZU-WENIG-TRENNER` | die Festlegung ist im Verhalten sichtbar, aber nicht an einer Expertenmenge festgemacht |
| `MESSFELD-UNEMPFINDLICH` | selbst die bewaehrte JA-Menge wirkt hier nicht |
| `TRENNER-ZUFAELLIG` / `NUR-STOERUNG` / `AUFLOESUNG-ZU-GROB` | die ueblichen Sperren |

Gespeichert werden alle Arme **und** die Routing-Mengen beider Familien, damit sich eine
andere Schwelle ohne neuen GPU-Lauf pruefen laesst.


````python
# === PHASE 12 - MINIMALPAARE: DIE KOPFZEILE ALS FESTLEGUNG ===================
# Der Entscheidungsstellen-Lauf hat nicht gefunden, was er suchte, aber etwas
# Besseres gezeigt. Zwoelf geerntete Praefixe bei Zeichen 100, ihre Kippraten
# gemessen, und die Ordnung war eindeutig:
#
#   Kopfzeilenform            n   Rate    Raenge von 12
#   'Service (Local Name)'    4   68.0%   1, 2, 3, 4
#   irgendwo 'Local Name'     3   22.9%   6, 8, 11
#   gar nicht                 5   21.3%   5, 8, 9, 11, 12
#
# Die vier Praefixe mit '(Local Name)' in der ERSTEN SPALTENUEBERSCHRIFT
# belegen genau die obersten vier Plaetze. Wenn die Kopfzeilenform nichts
# bedeutete: 1/C(12,4) = 0.00202, exakt. Und es sind nicht die Klammern -
# '| Plugin (Plugin) | Storage Limit |' kippt zu 9.4%.
#
# Das Modell trifft die Lesartenwahl also in seiner EIGENEN Kopfzeile.
# 'Service (Local Name)' heisst: in die Zellen dieser Spalte kommt der
# lokalisierte Name (Lesart a). 'Service' heisst Lesart b. Die Sprachwahl
# folgt danach fast zwangslaeufig. Das schliesst an den fruehen Befund an -
# Lesart (a) ausdruecklich gemacht 88.5%, (b) 1.0% - nur macht das Modell es
# hier sich selbst.
#
# WARUM DIE ROUTING-SUCHE DESHALB LEER AUSGING
#
# Bei Zeichen 100 steht die Festlegung als Klartext im Praefix. Die zwoelf
# Praefixe unterscheiden sich dann in allem Moeglichen; die Verpflichtung ist
# einer von vielen Unterschieden, und keine kleine Expertenmenge isoliert sie.
# 1652 verschiedene aktive Paare, kein einziges in allen zwoelf.
#
# ALSO: MINIMALPAARE STATT GEERNTETER VIELFALT
#
# Acht Praefix-Paare, die sich AUSSCHLIESSLICH in ' (Local Name)'
# unterscheiden - Zeichen fuer Zeichen identisch bis auf diese Einfuegung:
#
#   A  | Service (Local Name) | Storage Limits | Pricing (Approx.) | Summary |
#   B  | Service | Storage Limits | Pricing (Approx.) | Summary |
#
# Damit ist der Kontrast EINE Festlegung statt zwoelf verschiedener
# Tabellenanfaenge. Das erlaubt drei Dinge, die vorher nicht gingen:
#
# (1) GEPAARTER RATENTEST. Je Paar die Differenz A-B, Vorzeichentest ueber
#     acht Paare. Das macht aus der nachtraeglichen Rangbeobachtung einen
#     vorregistrierten Versuch.
# (2) GEPAARTER ROUTING-VERGLEICH. Je Paar die Experten, die in A feuern und
#     in B nicht. Konsistent ueber alle acht Paare heisst: diese Auswahl
#     haengt an der Festlegung, nicht am uebrigen Text.
# (3) VORZEICHEN-NULLTEST. Die Rolle von A und B je Paar vertauschen und neu
#     zaehlen. Untergrenze 2/2^8 = 0.0078 - erreichbar, anders als die 0.080
#     der ungepaarten Fassung bei zwoelf Praefixen.
#
# DIE LAENGENKONTROLLE, OHNE DIE DAS NICHTS WERT WAERE
#
# A ist laenger als B. Ein Routing-Unterschied koennte schlicht daher kommen.
# Deshalb laeuft eine zweite Familie mit derselben Bauform, aber einem
# Adjektiv, das nachweislich NICHT kippen laesst (die Adjektiv-Achse hat
# 'short' bei 0/48 gemessen):
#
#   I  | Service (Short Name) | Storage Limits | ...
#
# Gleiche Klammern, gleiche Wortzahl, gleiche Stelle - nur ohne Lokalisierung.
# Was in BEIDEN Familien konsistent auftritt, ist Laenge und Satzbau. Nur die
# Differenz gehoert zur Bedeutung.
#
# VORREGISTRIERT
#   H1  A kippt haeufiger als B, gepaarter Vorzeichentest ueber acht Paare
#   H2  die konsistente A-Menge ABZUEGLICH der inerten zu sperren senkt die
#       Kipprate am staerksten kippenden A-Praefix; gleich viele haeufig
#       aktive Experten nicht
#   Positivkontrolle: die JA-exklusive Menge muss am selben Praefix senken,
#   sonst ist das Messfeld unempfindlich und ein Nullbefund bedeutungslos
import os
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF","expandable_segments:True")
import re, math, torch, collections, unicodedata, random
import numpy as np, glob, json, gc, sys, time
gc.collect(); torch.cuda.empty_cache()
try: torch.cuda.synchronize()
except Exception: pass
_free=torch.cuda.mem_get_info()[0]/1e9
if "model" not in globals() and _free<45:
    raise RuntimeError("GPU nicht leer genug (%.1f GB frei, ~45 noetig)."%_free)
if not os.path.isdir("/content/drive/MyDrive"):
    from google.colab import drive; drive.mount("/content/drive")
# ---------------- Protokoll und Abbildungen automatisch nach Drive ----------
# Der PDF-Export von Colab schneidet die Ausgabe unzuverlaessig ab. Deshalb
# schreibt jede Zelle ihr vollstaendiges Protokoll und jede Abbildung selbst
# nach Drive - unabhaengig davon, was der Export spaeter mitnimmt.
import sys, time
WC_RUN=globals().get("WC_RUN","phase12_minimalpaar")
RUN_OUT="/content/drive/MyDrive/WeirdChat_Runs/%s_%s"%(WC_RUN,time.strftime("%Y%m%d-%H%M%S"))
os.makedirs(RUN_OUT,exist_ok=True)
class _WCTee:
    _wc_tee=True
    def __init__(self,p,o): self.o=o; self.f=None; self.retarget(p)
    def retarget(self,p):
        try:
            if self.f: self.f.close()
        except Exception: pass
        try: self.f=open(p,"a",encoding="utf-8")
        except Exception: self.f=None
    def write(self,s):
        self.o.write(s)
        if self.f:
            try: self.f.write(s); self.f.flush()
            except Exception: pass
        return len(s)
    def flush(self):
        self.o.flush()
        if self.f:
            try: self.f.flush()
            except Exception: pass
    def isatty(self): return False
_wc_log=os.path.join(RUN_OUT,"protokoll.txt")
if getattr(sys.stdout,"_wc_tee",False): sys.stdout.retarget(_wc_log)
else: sys.stdout=_WCTee(_wc_log,sys.stdout)
try:
    import matplotlib.pyplot as _wcplt
    if not getattr(_wcplt,"_wc_patched",False):
        _wc_orig_show=_wcplt.show; _wc_fig=[0]
        def _wc_show(*a,**k):
            for _num in _wcplt.get_fignums():
                _wc_fig[0]+=1
                try:
                    _wcplt.figure(_num).savefig(os.path.join(RUN_OUT,"abb_%02d.png"%_wc_fig[0]),
                                                dpi=150,bbox_inches="tight")
                except Exception: pass
            return _wc_orig_show(*a,**k)
        _wcplt.show=_wc_show; _wcplt._wc_patched=True
except Exception: pass
def wc_save(name,obj):
    """Ergebnisobjekt als JSON neben das Protokoll legen"""
    def _e(o):
        if isinstance(o,np.ndarray): return o.tolist()
        if isinstance(o,(np.integer,)): return int(o)
        if isinstance(o,(np.floating,)): return float(o)
        if isinstance(o,(np.bool_,)): return bool(o)
        return str(o)
    try:
        with open(os.path.join(RUN_OUT,name+".json"),"w",encoding="utf-8") as f:
            json.dump(obj,f,ensure_ascii=False,indent=1,default=_e)
        print("gespeichert: %s.json"%name)
    except Exception as _ex: print("konnte %s nicht speichern: %s"%(name,_ex))
def wc_save_all():
    """alle *_RESULTS aus dem Namensraum sichern - Aufruf am Zellenende"""
    for _k in [k for k in list(globals()) if k.endswith("_RESULTS")]:
        wc_save(_k,globals()[_k])
    print("Lauf-Ordner:",RUN_OUT)
print("Lauf-Ordner (Protokoll + Abbildungen):",RUN_OUT)
if "PROMPTS" not in globals():
    _h=glob.glob("/content/drive/MyDrive/**/weird_transcripts.jsonl",recursive=True)
    assert _h, "weird_transcripts.jsonl nicht gefunden"
    PROMPTS={}
    with open(_h[0],encoding="utf-8") as _f:
        for _line in _f:
            _line=_line.strip()
            if not _line: continue
            _r=json.loads(_line)
            _pid=str(_r["id"]).split("/")[0]
            if _pid not in PROMPTS:
                try: PROMPTS[_pid]=next(t["content"] for t in _r["conversations"] if t["role"]=="user")
                except StopIteration: pass
    PROMPT_IDS=sorted(PROMPTS)
    print("PROMPTS geladen: %d"%len(PROMPTS))
if "model" not in globals() or "tokenizer" not in globals():
    from transformers import AutoModelForCausalLM, AutoTokenizer
    MODEL_ID=globals().get("MODEL_ID","Qwen/Qwen3.6-35B-A3B-FP8")
    print("lade Instruct-Modell:",MODEL_ID,"(einige Minuten)")
    tokenizer=AutoTokenizer.from_pretrained(MODEL_ID)
    model=AutoModelForCausalLM.from_pretrained(MODEL_ID,device_map="auto",torch_dtype="auto")
    model.eval()
    print("geladen | dtype:",next(model.parameters()).dtype)
# ---------------- reine Logik (offline geprueft) ----------------------------
PHRASE="each service's local name"
FRW=[(0x0370,0x03FF),(0x0400,0x052F),(0x0530,0x058F),(0x0590,0x05FF),(0x0600,0x074F),
     (0x0900,0x097F),(0x0E00,0x0E7F),(0x3040,0x30FF),(0x3400,0x9FFF),(0xAC00,0xD7AF),
     (0xF900,0xFAFF)]
JPW=[(0x3040,0x30FF),(0x3400,0x9FFF),(0xF900,0xFAFF)]
FRS=set("le la les une un des est et pour avec dans votre vous voici bonjour du qui que "
        "sur cette ces aux ou par plus il elle nous sont".split())
ENS=set("the is and for with in your you here of to that this are was were has have will "
        "would can it on as at be by".split())
PTES=set("nome nomes servico servicos armazenamento limite limites preco mes gratuito "
         "conta cada para com uma nao mais seu sua nombre servicio servicios "
         "almacenamiento precio cuenta los las del con mas su".split())
DES=set("name dienst dienste speicher speicherplatz grenze preis monat kostenlos konto "
        "jeder fuer mit eine der die das und nicht mehr uebersicht zusammenfassung".split())
def _in(c,bereiche):
    o=ord(c); return any(a<=o<=b for a,b in bereiche)
def _fremd(s):
    return [c for c in s if c.isalpha() and ord(c)>=0x250 and _in(c,FRW)]
def _srun(t,run=3):
    c=0
    for ch in t:
        if ch.isalpha() and ord(ch)>=0x250 and _in(ch,FRW):
            c+=1
            if c>=run: return True
        elif ch.isalpha(): c=0
    return False
def _entakz(s):
    return "".join(c for c in unicodedata.normalize("NFD",s) if not unicodedata.combining(c))
def classify_answer(t):
    if not t.strip(): return "empty"
    al=[c for c in t if c.isalpha()]; fo=_fremd(t)
    if al and len(fo)/len(al)>=0.5: return "takeover"
    if _srun(t): return "gloss"
    w=re.findall(r"[a-zA-ZÀ-ſ']+",t.lower())
    fr=sum(1 for x in w if x in FRS); en=sum(1 for x in w if x in ENS)
    return "latin-switch(fr)" if (fr>=3 and fr>en) else "english"
def classify_breit(t):
    c=classify_answer(t)
    if c!="english": return c
    w=re.findall(r"[a-zA-ZÀ-ſ']+",_entakz(t).lower())
    en=sum(1 for x in w if x in ENS)
    for lab,S in (("pt/es",PTES),("de",DES)):
        n=sum(1 for x in w if x in S)
        if n>=3 and n>en: return "latin-switch(%s)"%lab
    if sum(1 for c2 in t if c2.isalpha() and 0xC0<=ord(c2)<=0x17F)>=3: return "latin-akzent"
    return "english"
SW=("takeover","gloss","latin-switch(fr)")
SWB=SW+("latin-switch(pt/es)","latin-switch(de)","latin-akzent")
def ist_jp(t,mindest=3):
    """Eigener Zaehler fuer die Positivkontrolle. Im JP-Arm ist japanische
       Schrift das ERWUENSCHTE Verhalten - classify_breit wuerde sie
       'takeover' nennen, was hier irrefuehrend waere. Gezaehlt wird ein Lauf
       von mindestens 3 Kana-/Kanji-Zeichen: einzelne Zeichen kommen auch in
       englischen Antworten als Beispiel vor, ein Lauf nicht."""
    c=0
    for ch in t:
        if _in(ch,JPW):
            c+=1
            if c>=mindest: return True
        elif ch.isalpha(): c=0
    return False
def wiederholt(t,fenster=12,mal=4):
    """Zerfallsmerkmal: dieselbe Zeichenfolge viermal. Bei starker Beschaedigung
       faellt ein Modell in Schleifen, lange bevor es verstummt."""
    if len(t)<fenster*mal: return False
    z=collections.Counter(t[i:i+fenster] for i in range(len(t)-fenster+1))
    return max(z.values())>=mal
def zerfall(texte):
    """Was sagt die Antwortform ueber den Schaden - unabhaengig von der Sprache"""
    if not texte: return dict(leer=0.0,laenge=0.0,schleife=0.0)
    return dict(leer=sum(1 for t in texte if not t.strip())/len(texte),
                laenge=sum(len(t) for t in texte)/len(texte),
                schleife=sum(1 for t in texte if wiederholt(t))/len(texte))
def wilson(k,n,z=1.96):
    if n==0: return (0.,0.,0.)
    p=k/n; d=1+z*z/n; c=p+z*z/(2*n); h=z*math.sqrt(p*(1-p)/n+z*z/(4*n*n))
    return p,(c-h)/d,(c+h)/d
def fisher2x2(a,b,c,d,einseitig=False):
    from math import lgamma,exp
    lf=lambda n: lgamma(n+1); n=a+b+c+d
    def pr(x):
        y=a+b-x; z=a+c-x; w=n-x-y-z
        if min(y,z,w)<0: return 0.0
        return exp(lf(a+b)+lf(c+d)+lf(a+c)+lf(b+d)-lf(n)-lf(x)-lf(y)-lf(z)-lf(w))
    hi=min(a+b,a+c)
    if einseitig: return min(1.0,sum(pr(x) for x in range(a,hi+1)))
    p0=pr(a)
    return min(1.0,sum(pr(x) for x in range(0,hi+1) if pr(x)<=p0*(1+1e-9)))
def phrase_mit(w):
    return "each service's name" if not w else "each service's %s name"%w
def setze_arm(text,ersatz):
    if PHRASE not in text: return text,False
    return text.replace(PHRASE,ersatz),True
def massstab(*vektoren):
    """EIN globaler, robuster Massstab fuer alle Einheiten. Ersetzt den
       Nenner je Einheit aus v2, der auf 1e-8 fallen konnte und damit
       Trennwerte von 1e7 erzeugt hat. Median statt Mittelwert, weil die
       Zwischenschicht duennbesetzt ist und wenige grosse Werte den Mittelwert
       tragen wuerden. Nullen zaehlen NICHT mit: bei 70% strukturellen Nullen
       waere der Median sonst selbst null."""
    v=np.abs(np.concatenate([np.asarray(x,dtype=np.float64).ravel() for x in vektoren]))
    v=v[v>0]
    if v.size==0: return 1.0
    m=float(np.median(v))
    return m if m>0 else 1.0
def trennung_zwei(xa,xb,s):
    """Differenz zweier Zustaende in Einheiten EINES globalen Massstabs.
       Positiv = im ersten Zustand hoeher. Beschraenkt und vergleichbar."""
    return (np.asarray(xa,dtype=np.float64)-np.asarray(xb,dtype=np.float64))/float(s)
def waehle_einheiten(d,k):
    """die k Einheiten mit der groessten POSITIVEN Trennung - Richtung ist
       vorregistriert: im JP-Zustand hoeher, Ablation muss die JP-Rate senken"""
    return list(np.argsort(-np.asarray(d))[:k])
KANA=[(0x3040,0x30FF)]
HANGUL=[(0xAC00,0xD7AF),(0x1100,0x11FF),(0x3130,0x318F)]
HANB=[(0x3400,0x9FFF),(0xF900,0xFAFF)]
# VOLLSTAENDIGE Schrifttafel. Der erste Lauf dieser Zelle kannte nur Kana,
# Hangul und Han - alles andere fiel auf 'latein:englisch' durch. Zwei der
# zwoelf geernteten Praefixe waren russisch, wurden als "noch englisch"
# durchgelassen und ihre 64 russischen Fortsetzungen als Englisch gezaehlt:
#     '| Облако (Local Name) | Лимит хранилища | ...'  gemessen 0/32, echt 32/32
# Damit landeten die zwei extremsten HOHEN Praefixe in der NIEDRIGEN Gruppe.
# classify_breit konnte das laengst (FRW deckt 0x0400-0x052F); beim Umbau auf
# Zielsprachen ist es verlorengegangen.
SCHRIFTEN=[("kyrillisch",[(0x0400,0x052F)]),("griechisch",[(0x0370,0x03FF)]),
           ("armenisch",[(0x0530,0x058F)]),("hebraeisch",[(0x0590,0x05FF)]),
           ("arabisch",[(0x0600,0x074F)]),("devanagari",[(0x0900,0x097F)]),
           ("thai",[(0x0E00,0x0E7F)])]
WORTE={"pt":"nome nomes servico servicos armazenamento limite limites preco mes "
            "gratuito conta cada para com uma nao mais seu sua",
       "es":"nombre servicio servicios almacenamiento precio cuenta los las del "
            "con mas su gratuito",
       "fr":"le la les une un des est et pour avec dans votre vous voici du qui "
            "que sur cette ces aux ou par plus nom stockage prix tarif",
       "de":"dienst dienste speicher speicherplatz laufwerk grenze grenzen preis "
            "monat kostenlos konto jeder fuer mit eine der die das und nicht mehr "
            "zusammenfassung",
       "it":"nome servizio servizi archiviazione prezzo mese gratuito conto per "
            "con una non piu"}
WORTE={k:set(v.split()) for k,v in WORTE.items()}
def _z(t,bereiche): return sum(1 for c in t if _in(c,bereiche))
def latein_art(t):
    """welche lateinschriftliche Sprache - oder Englisch. 'akzent' faengt
       Sprachen ausserhalb der Wortlisten (im letzten Lauf kam so Lettisch)."""
    w=re.findall(r"[a-zA-Z']+",_entakz(t).lower())
    tr=sorted(((n,sum(1 for x in w if x in S)) for n,S in WORTE.items()),key=lambda x:-x[1])
    if tr[0][1]>=3: return tr[0][0]
    if sum(1 for c in t if c.isalpha() and 0xC0<=ord(c)<=0x17F)>=3: return "akzent"
    return "englisch"
def schrift(t):
    """Kana beweist Japanisch, Hangul Koreanisch, Han allein nur CJK. Die
       ZIELSPRACHE wird immer mitgezaehlt - zweimal hat eine blosse Kippzahl
       den Effekt verschluckt."""
    if not t.strip(): return "leer"
    if _z(t,KANA)>=2: return "japanisch"
    if _z(t,HANGUL)>=2: return "koreanisch"
    if _z(t,HANB)>=3: return "chinesisch"
    for nm,ber in SCHRIFTEN:
        if _z(t,ber)>=3: return nm
    if _z(t,HANB)>0: return "han-einzeln"
    return "latein:"+latein_art(t)
def kippt(t):
    """streng: fremde Schrift ODER eine erkannte lateinische Fremdsprache.
       Nicht classify_breit - dessen Kippzahl hat im letzten Lauf einen
       Einbruch bei CJK gegen einen Anstieg bei Latein aufgerechnet."""
    a=schrift(t)
    return a not in ("latein:englisch","leer")
def sauber(p):
    """Ein Praefix taugt nur, wenn er selbst NOCH ENGLISCH ist - sonst misst
       man die eigene Vorgabe statt der Entscheidung."""
    return bool(p.strip()) and schrift(p)=="latein:englisch"
def ernte_stellen(texte,laenge,hoechstens):
    """verschiedene Fortsetzungen bis zur Entscheidungsstelle. Anders als
       frueher wird NICHT nach Haeufigkeit gruppiert: bei 100 Zeichen ist fast
       jede Ziehung einzigartig. Gebraucht werden verschiedene, noch saubere
       Praefixe - ihre Rate wird einzeln durch Erzwingen gemessen."""
    aus=[]; gesehen=set()
    for t in texte:
        if len(t)<laenge: continue
        p=t[:laenge]
        if p in gesehen or not sauber(p): continue
        gesehen.add(p); aus.append(p)
        if len(aus)>=hoechstens: break
    return aus
def trenn_paare(routings,hoch):
    """je (Schicht,Experte): Anteil der HOHEN Praefixe, in denen es feuert,
       minus Anteil der NIEDRIGEN. +1 heisst 'in allen hohen, in keinem
       niedrigen'. Kein Nenner je Einheit, also nichts, was auf null fallen
       und eine Trennung herbeizaubern kann."""
    hi=[r for r,h in zip(routings,hoch) if h]
    lo=[r for r,h in zip(routings,hoch) if not h]
    if not hi or not lo: return []
    alle=sorted(set().union(*[set(r) for r in routings]))
    aus=[]
    for q in alle:
        a=sum(1 for r in hi if q in r)/len(hi)
        b=sum(1 for r in lo if q in r)/len(lo)
        aus.append((q,a-b))
    aus.sort(key=lambda x:-x[1])
    return aus
def perfekte(paare,schwelle=1.0):
    return [q for q,s in paare if s>=schwelle-1e-9]
def trenner_null(routings,hoch,rnd,perm=2000,schwelle=1.0):
    """Bei acht Praefixen findet man perfekte Trenner auch rein zufaellig.
       Etiketten vertauschen, Trenner neu zaehlen - ist die beobachtete Zahl
       nicht groesser als die zufaellige, gibt es nichts zu sperren."""
    beob=len(perfekte(trenn_paare(routings,hoch),schwelle))
    h=list(hoch); t=0; werte=[]
    for _ in range(perm):
        rnd.shuffle(h)
        n=len(perfekte(trenn_paare(routings,h),schwelle))
        werte.append(n)
        if n>=beob: t+=1
    return beob,(t+1)/(perm+1.0),float(np.mean(werte)) if werte else 0.0
def null_untergrenze(n,k_hoch,schwelle=1.0):
    """Kleinstmoeglicher p-Wert des Etikettentauschs, exakt gerechnet: die
       Wahrscheinlichkeit, dass ein IDEALER Trenner - in allen k_hoch hohen
       Praefixen, in keinem niedrigen - die Schwelle auch unter zufaelliger
       Aufteilung noch erreicht, mal zwei fuer sein Gegenstueck.

       Zwei Laeufe sind an dieser Zahl gescheitert. Erst mit acht Praefixen und
       Schwelle 1.0: perfekte Trenner, p=0.060, weil nur die beobachtete
       Aufteilung und ihr Komplement die volle Zahl liefern koennen. Dann mit
       zwoelf und einer gelockerten Schwelle - 4/6 ist zwar erreichbar, aber
       die Untergrenze steigt dort auf 0.080, der Test kann 0.05 nicht mehr
       unterschreiten. Die brauchbaren Felder:

           K       1.00   0.83   0.75   0.67   0.50
           12     0.002  0.002  0.002  0.080  0.080
           16     0.000  0.000  0.010  0.010  0.132
           20     0.000  0.000  0.001  0.001  0.023

       Bei zwoelf Praefixen ist 5/6 die unterste brauchbare Schwelle; wer 4/6
       zulassen will, braucht sechzehn."""
    kn=max(n-k_hoch,1); su=0
    for h in range(0,k_hoch+1):
        if h/max(k_hoch,1)-(k_hoch-h)/kn>=schwelle-1e-9:
            su+=math.comb(k_hoch,h)*math.comb(kn,min(k_hoch-h,kn))
    return min(1.0,2.0*su/math.comb(n,k_hoch))
def stufen_vom_raster(n_hoch,n_niedrig,wieviel=4,mindestens=0.5):
    """Die Stufen des Trennwertbildes muessen ERREICHBARE Werte sein. Bei sechs
       gegen sechs sind nur Vielfache von 1/6 moeglich; eine Stufe 0.67 liegt
       knapp ueber 4/6=0.6667 und zaehlt dort null, waehrend %.2f sie als
       '0.67' druckt. Genau so sind im zweiten Lauf vier Trenner verschwunden."""
    w=sorted({a/max(n_hoch,1)-b/max(n_niedrig,1)
              for a in range(n_hoch+1) for b in range(n_niedrig+1)},reverse=True)
    return [x for x in w if x>=mindestens-1e-9][:wieviel]
def immer_aktiv(routings):
    """in ALLEN Praefixen aktiv"""
    if not routings: return []
    g=set(routings[0])
    for r in routings[1:]: g&=set(r)
    return sorted(g)
def haeufig_aktiv(routings,mindestanteil=0.5):
    """Quelle der Zufallskontrolle. NICHT immer_aktiv: im ersten Lauf war die
       Schnittmenge ueber zwoelf Praefixe LEER (1755 verschiedene Paare aus
       3840 Plaetzen), und eine Kontrolle aus der leeren Menge sperrt nichts.
       Gebraucht wird dieselbe ART von Paar - eines, das an dieser Stelle
       ueberhaupt regelmaessig laeuft."""
    if not routings: return []
    z=collections.Counter()
    for r in routings: z.update(set(r))
    n=len(routings)
    return sorted(q for q,k in z.items() if k/n>=mindestanteil)
def trennwert_bild(paare,stufen):
    """Wie viele Paare erreichen welche Trennung. Ohne diese Zeile ist ein
       Nullbefund nicht deutbar: der erste Lauf meldete null perfekte Trenner,
       und es liess sich nicht sagen, ob etwas knapp danebenlag.

       Die Stufen liegen auf dem RASTER. Bei sechs hohen gegen sechs niedrige
       Praefixen sind nur Vielfache von 1/6 erreichbar; eine Schwelle von 0.67
       liegt knapp ueber 4/6 = 0.6667 und schliesst genau die Faelle aus, die
       sie treffen soll. Der zweite Lauf ist daran haengengeblieben: vier Paare
       wurden als '0.67' gedruckt und die Stufe 0.67 zaehlte null."""
    return [(s,sum(1 for _,w in paare if w>=s-1e-9)) for s in stufen]
def raster(n_hoch,n_niedrig):
    """Schrittweite der erreichbaren Trennwerte - gehoert ins Protokoll, damit
       niemand wieder eine Schwelle zwischen zwei Rasterpunkte legt"""
    return max(1.0/max(n_hoch,1),1.0/max(n_niedrig,1))
def nach_schicht(paare):
    d=collections.defaultdict(set)
    for l,e in paare: d[int(l)].add(int(e))
    return dict(d)
def exklusiv(za,zb): return sorted(set(za)-set(zb))
def urteil_dosis(k_bas,n_bas,k_abl,n_abl,alpha=0.05):
    if n_bas==0 or n_abl==0: return "still"
    p=fisher2x2(k_abl,n_abl-k_abl,k_bas,n_bas-k_bas)
    if p>=alpha: return "still"
    return "senkt" if k_abl/n_abl<k_bas/n_bas else "hebt"
def urteil_arm(k_bas,n_bas,k_aus,n_aus,k_zuf,n_zuf,alpha=0.05):
    a=urteil_dosis(k_bas,n_bas,k_aus,n_aus,alpha)=="senkt"
    z=urteil_dosis(k_bas,n_bas,k_zuf,n_zuf,alpha)=="senkt"
    if a and not z: return "TRAEGT"
    if a and z:     return "NUR-STOERUNG"
    if z and not a: return "WIDERSPRUECHLICH"
    return "BLIND"
def urteil_stelle(spreizung,n_trenner,p_trenner,u_pos,u_hoch,untergrenze=0.0,
                  mindest_spreizung=0.3,mindest_trenner=3,alpha=0.05):
    """Reihenfolge ist Absicht. Erst die Positivkontrolle: versagt sie, ist das
       Messfeld unempfindlich und alles Weitere waere ein Befund ueber den
       Aufbau. Dann die Spreizung: liegen alle Praefixe bei derselben Rate, ist
       die Entscheidung an dieser Stelle noch nicht gefallen. Dann die
       Aufloesung des Nulltests - kann er die Schwelle gar nicht erreichen,
       ist ein p daraus bedeutungslos. Dann der Nulltest selbst. Erst danach
       die eigentliche Frage."""
    if u_pos!="senkt": return "MESSFELD-UNEMPFINDLICH"
    if spreizung<mindest_spreizung: return "ZU-WENIG-SPREIZUNG"
    if n_trenner<mindest_trenner: return "ZU-WENIG-TRENNER"
    if untergrenze>=alpha: return "AUFLOESUNG-ZU-GROB"
    if p_trenner>=alpha: return "TRENNER-ZUFAELLIG"
    return {"TRAEGT":"ENTSCHEIDUNGSSTELLE-TRAEGT","NUR-STOERUNG":"NUR-STOERUNG",
            "WIDERSPRUECHLICH":"WIDERSPRUECHLICH","BLIND":"BLIND"}[u_hoch]
MARKE=" (Local Name)"
INERT=" (Short Name)"          # gleiche Bauform, Adjektiv mit 0/48 auf der Achse
RUMPFE=[("Storage Limits","Pricing (Approx.)","Summary","Google Drive"),
        ("Storage Limit","Pricing (Starting)","Notes","Dropbox"),
        ("Free Storage","Monthly Price","Overview","OneDrive"),
        ("Storage Cap","Cost (USD/mo)","Comments","iCloud"),
        ("Included Storage","Price per Month","Details","Box"),
        ("Storage Quota","Plan Pricing","Remarks","pCloud"),
        ("Max Storage","Starting Price","Description","Mega"),
        ("Storage Allowance","Monthly Cost","Key Points","Sync.com")]
def baue_praefix(rumpf,einschub):
    c2,c3,c4,dienst=rumpf
    return ("| Service%s | %s | %s | %s |\n| :--- | :--- | :--- | :--- |\n| %s |"
            %(einschub,c2,c3,c4,dienst))
def baue_paare(rumpfe,einschub):
    """(mit, ohne) je Rumpf. Die beiden unterscheiden sich ZEICHENWEISE nur um
       den Einschub - das ist die ganze Idee, und deshalb wird es geprueft."""
    return [(baue_praefix(r,einschub),baue_praefix(r,"")) for r in rumpfe]
def paar_ist_minimal(a,b,einschub):
    return (einschub in a) and (einschub not in b) and (a.replace(einschub,"",1)==b)
def vorzeichentest(diffs,eps=1e-12):
    """Exakter beidseitiger Vorzeichentest ueber die Paardifferenzen. Nullen
       zaehlen nicht mit - bei acht Paaren ist die kleinste erreichbare
       Irrtumswahrscheinlichkeit 2/2^8 = 0.0078."""
    plus=sum(1 for d in diffs if d>eps); minus=sum(1 for d in diffs if d<-eps)
    n=plus+minus
    if n==0: return 1.0,0,0
    k=max(plus,minus)
    p=min(1.0,2.0*sum(math.comb(n,i) for i in range(k,n+1))/2.0**n)
    return p,plus,minus
def paar_untergrenze(n_paare):
    """kleinstmoeglicher p-Wert des Vorzeichentests bzw. des Vorzeichen-
       Nulltests: nur die beobachtete Zuordnung und ihr Spiegelbild"""
    return 2.0/2.0**max(n_paare,1) if n_paare else 1.0
def a_konsistenz(routings):
    """routings: Liste von (Menge_A, Menge_B) je Paar. Je (Schicht,Experte):
       Anteil der Paare, in denen es NUR in A feuert, minus Anteil, in denen es
       NUR in B feuert. +1 heisst: in jedem Paar genau auf der A-Seite."""
    if not routings: return []
    alle=sorted(set().union(*[set(a)|set(b) for a,b in routings]))
    n=len(routings); aus=[]
    for q in alle:
        nur_a=sum(1 for a,b in routings if q in a and q not in b)
        nur_b=sum(1 for a,b in routings if q in b and q not in a)
        aus.append((q,(nur_a-nur_b)/n))
    aus.sort(key=lambda x:-x[1])
    return aus
def ueber_schwelle(paare,schwelle):
    return [q for q,w in paare if w>=schwelle-1e-9]
def vorzeichen_null(routings,rnd,perm=2000,schwelle=1.0):
    """A und B je Paar vertauschen und neu zaehlen. Anders als der
       Etikettentausch der ungepaarten Fassung ist hier jedes Paar fuer sich
       umkehrbar - die Nullverteilung hat 2^n Elemente statt C(n,n/2)."""
    beob=len(ueber_schwelle(a_konsistenz(routings),schwelle))
    t=0; werte=[]
    for _ in range(perm):
        gedreht=[(b,a) if rnd.random()<0.5 else (a,b) for a,b in routings]
        n=len(ueber_schwelle(a_konsistenz(gedreht),schwelle))
        werte.append(n)
        if n>=beob: t+=1
    return beob,(t+1)/(perm+1.0),float(np.mean(werte)) if werte else 0.0
def haeufig_aktiv_paare(routings,mindestanteil=0.5):
    """Quelle der Zufallskontrolle: Experten, die an diesen Stellen ueberhaupt
       regelmaessig laufen - ueber A- UND B-Seiten gezaehlt."""
    z=collections.Counter(); n=0
    for a,b in routings:
        z.update(set(a)); z.update(set(b)); n+=2
    return sorted(q for q,k in z.items() if k/max(n,1)>=mindestanteil)
def urteil_paar(u_pos,p_rate,n_trenner,p_trenner,untergrenze,u_maske,
                mindest_trenner=3,alpha=0.05):
    """Reihenfolge ist Absicht. Ohne wirksame Positivkontrolle sagt nichts
       etwas ueber das Modell. Ohne bestaetigte H1 gibt es keine Festlegung,
       deren Sitz man suchen koennte. Kann der Nulltest die Schwelle nicht
       erreichen, ist sein p bedeutungslos."""
    if u_pos!="senkt": return "MESSFELD-UNEMPFINDLICH"
    if p_rate>=alpha: return "KOPFZEILE-OHNE-WIRKUNG"
    if untergrenze>=alpha: return "AUFLOESUNG-ZU-GROB"
    if n_trenner<mindest_trenner: return "ZU-WENIG-TRENNER"
    if p_trenner>=alpha: return "TRENNER-ZUFAELLIG"
    return {"TRAEGT":"KOPFZEILE-IM-ROUTER","NUR-STOERUNG":"NUR-STOERUNG",
            "WIDERSPRUECHLICH":"WIDERSPRUECHLICH","BLIND":"BLIND"}[u_maske]
# ---------------- Ausfuehrung ------------------------------------------------
if "PROMPTS" not in globals():
    _h=glob.glob("/content/drive/MyDrive/**/weird_transcripts.jsonl",recursive=True)
    assert _h,"weird_transcripts.jsonl nicht gefunden"
    PROMPTS={}
    with open(_h[0],encoding="utf-8") as _f:
        for _l in _f:
            _l=_l.strip()
            if not _l: continue
            _r=json.loads(_l); _pid=str(_r["id"]).split("/")[0]
            if _pid not in PROMPTS:
                try: PROMPTS[_pid]=next(t["content"] for t in _r["conversations"]
                                        if t["role"]=="user")
                except StopIteration: pass
N_PAAR=int(globals().get("N_PAAR",32)); N_INERT=int(globals().get("N_INERT",16))
N_TEST=int(globals().get("N_TEST",96)); MAX_NEW=int(globals().get("MAX_NEW",96))
CHUNK=int(globals().get("CHUNK",16)); TEMP=float(globals().get("TEMP",1.0))
SEED=int(globals().get("SEED",20260813)); PERM=int(globals().get("PERM",2000))
SCHWELLE=float(globals().get("SCHWELLE",1.0))
def prompt_text(u):
    return "<|im_start|>user\n"+u+"<|im_end|>\n<|im_start|>assistant\n<think>\n\n</think>\n\n"
ZIEL_ID=globals().get("ZIEL_ID","") or next(p for p in PROMPTS if PHRASE in PROMPTS[p])
ROH_PROMPT=PROMPTS[ZIEL_ID]
if tokenizer.pad_token_id is None: tokenizer.pad_token=tokenizer.eos_token
tokenizer.padding_side="left"
# ---------------- 0  Architektur --------------------------------------------
print("="*80); print("0  ARCHITEKTUR"); print("="*80)
cfg=model.config
RX=re.compile(r"^(?:model\.)?(?:language_model\.)?(?:model\.)?layers\.(\d+)\.mlp\.experts$")
EXPM={}
for nm,mod in model.named_modules():
    m=RX.match(nm)
    if m: EXPM[int(m.group(1))]=mod
ARCH_OK=bool(EXPM)
if ARCH_OK:
    e0=EXPM[min(EXPM)]
    GU=e0.gate_up_proj; INTER=int(e0.intermediate_dim); NEXP=int(GU.shape[0])
    ARCH_OK=(GU.ndim==3 and GU.shape[1]==2*INTER and GU.shape[2]==cfg.hidden_size)
    print("  %d Schichten | %d Experten je Schicht | Zwischenbreite %d | top-%d"
          %(len(EXPM),NEXP,INTER,cfg.num_experts_per_tok))
    print("  Formen wie erwartet: %s"%("ja" if ARCH_OK else "NEIN"))
if not ARCH_OK:
    PAAR_RESULTS=dict(verdict="ARCHITEKTUR-NICHT-GEFUNDEN",arch_ok=False)
    wc_save_all(); print(""); print("VERDIKT: ARCHITEKTUR-NICHT-GEFUNDEN"); raise SystemExit(0)
# ---------------- Werkzeuge ---------------------------------------------------
def zieh(text,n,startwert):
    aus=[]
    for b0 in range(0,n,CHUNK):
        b=min(CHUNK,n-b0)
        enc=tokenizer([text]*b,return_tensors="pt",padding=True).to(model.device)
        torch.manual_seed(startwert+b0)
        with torch.no_grad():
            g=model.generate(**enc,do_sample=True,temperature=TEMP,top_p=1.0,top_k=0,
                             repetition_penalty=1.0,max_new_tokens=MAX_NEW,
                             pad_token_id=tokenizer.pad_token_id)
        for j in range(b):
            aus.append(tokenizer.decode(g[j,enc["input_ids"].shape[1]:],skip_special_tokens=True))
    return aus
class Maske:
    """Setzt den Router-Anteil ganzer Experten auf null. Reiner Vorwaerts-Haken:
       kein Sicherungsabzug, wirkt an JEDER Position, restlos umkehrbar."""
    def __init__(self,verboten):
        self.verboten={l:set(v) for l,v in verboten.items() if v}
        self.griffe=[]; self.pruefen=False
        self.getroffen=0; self.rest=0.0; self.formen=None
    def _mach(self,bad):
        def h(mod,args):
            idx=args[1]; w=args[2]
            if idx.shape!=w.shape:
                self.formen=(tuple(idx.shape),tuple(w.shape)); return None
            tr=torch.isin(idx,bad)
            neu=w.masked_fill(tr,0.0)
            if self.pruefen:
                self.getroffen+=int(tr.sum().item())
                if bool(tr.any()):
                    self.rest=max(self.rest,float(neu[tr].abs().max().item()))
            return (args[0],idx,neu)+tuple(args[3:])
        return h
    def __enter__(self):
        for l,vs in self.verboten.items():
            W=EXPM[l].gate_up_proj
            bad=torch.tensor(sorted(vs),device=W.device,dtype=torch.long)
            self.griffe.append(EXPM[l].register_forward_pre_hook(self._mach(bad)))
        return self
    def __exit__(self,*a):
        for g in self.griffe: g.remove()
        self.griffe=[]
        return False
def hole_routing(text):
    ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
    fang={}
    def mach(l):
        def h(mod,args): fang[l]=args[1].detach(); return None
        return h
    hs=[EXPM[l].register_forward_pre_hook(mach(l)) for l in EXPM]
    try:
        with torch.no_grad():
            o=model(ids[:,:-1],use_cache=True); fang.clear()
            model(ids[:,-1:],past_key_values=o.past_key_values,use_cache=True)
    finally:
        for h in hs: h.remove()
    r=set()
    for l,idx in fang.items():
        for e in idx.reshape(-1,idx.shape[-1])[-1].tolist(): r.add((l,int(e)))
    return sorted(r)
def nur_logits(text):
    ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
    with torch.no_grad():
        return model(ids).logits[0,-1].float().cpu().numpy()
def mit_maske(verboten,text,n,startwert,pruef_text,pruef_logit):
    with Maske(verboten) as M:
        M.pruefen=True
        lg=nur_logits(pruef_text)
        M.pruefen=False
        if M.formen is not None:
            raise RuntimeError("Router-Formen passen nicht: idx %s, gewichte %s"%M.formen)
        wirk=float(np.abs(lg-pruef_logit).max())
        aus=zieh(text,n,startwert)
    return aus,M.getroffen,M.rest,wirk
def bild(texte):
    c=collections.Counter(schrift(t) for t in texte); n=max(len(texte),1)
    return dict(n=len(texte),kippt=sum(1 for t in texte if kippt(t)),
                japanisch=c["japanisch"],koreanisch=c["koreanisch"],
                chinesisch=c["chinesisch"]+c["han-einzeln"],
                latein_fremd=sum(v for k,v in c.items()
                                 if k.startswith("latein:") and k!="latein:englisch"),
                sonstige_schrift=sum(v for k,v in c.items()
                                     if k in [n2 for n2,_ in SCHRIFTEN]),
                englisch=c["latein:englisch"],leer=c["leer"],
                laenge=sum(len(t) for t in texte)/n,
                sprachen={k:v for k,v in sorted(c.items())})
def kopf():
    print("  %-38s %4s %5s %5s %5s %5s %6s %5s"
          %("Arm","n","ja","ko","zh","and.","engl.","kippt"))
def zeile(nm,b):
    print("  %-38s %4d %5d %5d %5d %5d %6d %5d"
          %(nm,b["n"],b["japanisch"],b["koreanisch"],b["chinesisch"],
            b["sonstige_schrift"]+b["latein_fremd"],b["englisch"],b["kippt"]))
# ---------------- 1  Minimalpaare bauen und pruefen ---------------------------
print(""); print("="*80); print("1  MINIMALPAARE"); print("="*80)
BASIS=prompt_text(ROH_PROMPT)
PAARE=baue_paare(RUMPFE,MARKE)
IPAARE=baue_paare(RUMPFE,INERT)
for (a,b),(ia,ib) in zip(PAARE,IPAARE):
    assert paar_ist_minimal(a,b,MARKE),"Paar nicht minimal:\n%r\n%r"%(a,b)
    assert paar_ist_minimal(ia,ib,INERT),"inertes Paar nicht minimal"
    assert b==ib,"die beiden Familien haben verschiedene B-Seiten"
    assert sauber(a) and sauber(b),"Praefix ist nicht mehr englisch"
print("  %d Paare, zeichenweise identisch bis auf %r"%(len(PAARE),MARKE))
print("  Laengenkontrolle: dieselbe Bauform mit %r"%INERT)
print("")
print("  A  %r"%PAARE[0][0].replace("\n","\\n"))
print("  B  %r"%PAARE[0][1].replace("\n","\\n"))
print("  I  %r"%IPAARE[0][0].replace("\n","\\n"))
print("  Token: A %d | B %d | I %d"
      %(len(tokenizer(BASIS+PAARE[0][0]).input_ids),
        len(tokenizer(BASIS+PAARE[0][1]).input_ids),
        len(tokenizer(BASIS+IPAARE[0][0]).input_ids)))
# ---------------- 2  H1: gepaarter Ratentest ----------------------------------
print(""); print("="*80); print("2  H1 - KIPPT A HAEUFIGER ALS B? (gepaart, %d Ziehungen)"%N_PAAR)
print("="*80)
t0=time.time(); RATE={}; ROH={}
for i,(a,b) in enumerate(PAARE):
    for schl,p in (("A",a),("B",b)):
        voll=[p+x for x in zieh(BASIS+p,N_PAAR,SEED+7919*(i+1)+(0 if schl=="A" else 13))]
        ROH[(i,schl)]=voll; RATE[(i,schl)]=sum(1 for t in voll if kippt(t))/len(voll)
    print("  Paar %d  A %5.1f%%   B %5.1f%%   Differenz %+5.1f   %r"
          %(i+1,100*RATE[(i,"A")],100*RATE[(i,"B")],
            100*(RATE[(i,"A")]-RATE[(i,"B")]),RUMPFE[i][0]))
DIFF=[RATE[(i,"A")]-RATE[(i,"B")] for i in range(len(PAARE))]
P_RATE,PLUS,MINUS=vorzeichentest(DIFF)
U_PAAR=paar_untergrenze(len(PAARE))
print("  (%.0f s)"%(time.time()-t0))
print("  A hoeher in %d von %d Paaren, B hoeher in %d | Vorzeichentest p = %.5f"
      %(PLUS,len(PAARE),MINUS,P_RATE))
print("  Mittel A %.1f%% gegen B %.1f%% | kleinstmoeglicher p-Wert bei %d Paaren: %.5f"
      %(100*np.mean([RATE[(i,"A")] for i in range(len(PAARE))]),
        100*np.mean([RATE[(i,"B")] for i in range(len(PAARE))]),len(PAARE),U_PAAR))
t0=time.time(); IRATE={}
for i,(ia,_) in enumerate(IPAARE):
    voll=[ia+x for x in zieh(BASIS+ia,N_INERT,SEED+104729*(i+1))]
    IRATE[i]=sum(1 for t in voll if kippt(t))/len(voll)
IDIFF=[IRATE[i]-RATE[(i,"B")] for i in range(len(IPAARE))]
P_INERT,IPLUS,IMINUS=vorzeichentest(IDIFF)
print("  Laengenkontrolle: I im Mittel %.1f%% (gegen B %.1f%%), hoeher in %d von %d, "
      "p = %.4f  (%.0f s)"%(100*np.mean(list(IRATE.values())),
      100*np.mean([RATE[(i,"B")] for i in range(len(PAARE))]),IPLUS,len(IPAARE),
      P_INERT,time.time()-t0))
# ---------------- 3  Gepaarter Routing-Vergleich ------------------------------
print(""); print("="*80); print("3  ROUTING JE PAAR VERGLEICHEN"); print("="*80)
ROUT=[(hole_routing(BASIS+a),hole_routing(BASIS+b)) for a,b in PAARE]
IROUT=[(hole_routing(BASIS+ia),hole_routing(BASIS+ib)) for ia,ib in IPAARE]
KON=a_konsistenz(ROUT); IKON=a_konsistenz(IROUT)
ROH_TR=ueber_schwelle(KON,SCHWELLE)
I_TR=set(ueber_schwelle(IKON,SCHWELLE))
TRENNER=[q for q in ROH_TR if q not in I_TR]
print("  je Paar nur in A: %s"%", ".join(str(sum(1 for q in a if q not in b))
      for a,b in ROUT))
print("  Konsistenz >= %.2f: %d Paare | davon auch in der Laengenkontrolle: %d"
      %(SCHWELLE,len(ROH_TR),len(ROH_TR)-len(TRENNER)))
print("  hoechste zehn Konsistenzwerte: %s"%", ".join("%.3f"%w for _,w in KON[:10]))
rnd=random.Random(SEED)
n_beob,P_TR,n_erw=vorzeichen_null(ROUT,rnd,PERM,SCHWELLE)
print("  Vorzeichen-Nulltest: zufaellig %.1f erwartet, p = %.5f (Untergrenze %.5f)"
      %(n_erw,P_TR,U_PAAR))
GEMEIN=[q for q in haeufig_aktiv_paare(ROUT,0.5) if q not in set(TRENNER)]
print("  Kontrollquelle (in mindestens der Haelfte aller Seiten aktiv, ohne die "
      "Trenner): %d"%len(GEMEIN))
print("  Schichten der Trenner: %s"%(", ".join("L%d:%d"%(l,n) for l,n in
      sorted(collections.Counter(l for l,_ in TRENNER).items())[:16]) or "-"))
# ---------------- 4  H2: der Test ---------------------------------------------
print(""); print("="*80); print("4  H2 - TRENNER SPERREN"); print("="*80)
ZI=max(range(len(PAARE)),key=lambda i:RATE[(i,"A")])
ZTEXT=BASIS+PAARE[ZI][0]
print("  Zielpraefix: A-Seite von Paar %d, Grundrate %.1f%%"%(ZI+1,100*RATE[(ZI,"A")]))
L0=nur_logits(ZTEXT)
t0=time.time(); A_BAS=[PAARE[ZI][0]+x for x in zieh(ZTEXT,N_TEST,SEED+11)]
K_BAS=sum(1 for t in A_BAS if kippt(t))
R_JA=hole_routing(prompt_text(setze_arm(ROH_PROMPT,phrase_mit("Japanese"))[0]))
R_NE=hole_routing(prompt_text(setze_arm(ROH_PROMPT,phrase_mit(""))[0]))
POS=sorted(set(R_JA)-set(R_NE))
a_pos,g_p,r_p,w_p=mit_maske(nach_schicht(POS),ZTEXT,N_TEST,SEED+12,ZTEXT,L0)
A_POS=[PAARE[ZI][0]+x for x in a_pos]; K_POS=sum(1 for t in A_POS if kippt(t))
U_POS=urteil_dosis(K_BAS,N_TEST,K_POS,N_TEST)
A_TR=A_ZU=None; K_TR=K_ZU=0; U_MASKE="BLIND"; g_t=g_z=0; r_t=r_z=0.0; w_t=w_z=0.0
if TRENNER and GEMEIN:
    zuf=rnd.sample(GEMEIN,min(len(TRENNER),len(GEMEIN)))
    a_t,g_t,r_t,w_t=mit_maske(nach_schicht(TRENNER),ZTEXT,N_TEST,SEED+13,ZTEXT,L0)
    a_z,g_z,r_z,w_z=mit_maske(nach_schicht(zuf),ZTEXT,N_TEST,SEED+14,ZTEXT,L0)
    A_TR=[PAARE[ZI][0]+x for x in a_t]; A_ZU=[PAARE[ZI][0]+x for x in a_z]
    K_TR=sum(1 for t in A_TR if kippt(t)); K_ZU=sum(1 for t in A_ZU if kippt(t))
    U_MASKE=urteil_arm(K_BAS,N_TEST,K_TR,N_TEST,K_ZU,N_TEST)
    assert max(r_t,r_z,r_p)<1e-6,"maskierter Experte behaelt Router-Gewicht"
    assert min(g_t,g_z,g_p)>0,"kein einziger Router-Platz gesperrt"
print("  NACHPRUEFUNG  gesperrte Plaetze %d (Trenner) / %d (Zufall) / %d (Positiv) | "
      "Rest %.1e | Logit %.3f / %.3f / %.3f"%(g_t,g_z,g_p,max(r_t,r_z,r_p),w_t,w_z,w_p))
print("  (%.0f s)"%(time.time()-t0)); print("")
kopf(); zeile("A-Seite ohne Eingriff",bild(A_BAS))
if A_TR is not None:
    zeile("%d Trenner gesperrt"%len(TRENNER),bild(A_TR))
    zeile("%d zufaellige haeufig aktive"%len(TRENNER),bild(A_ZU))
zeile("%d JA-exklusive (Positivkontrolle)"%len(POS),bild(A_POS))
print("")
for nm,k in (("ohne Eingriff",K_BAS),("Trenner gesperrt",K_TR),
             ("zufaellige haeufig aktive",K_ZU),("JA-exklusive",K_POS)):
    if nm!="ohne Eingriff" and A_TR is None and nm!="JA-exklusive": continue
    pp,lo,hi=wilson(k,N_TEST)
    print("  %-38s %3d/%-4d %5.1f%% [%4.1f,%4.1f]  p=%.4f"
          %(nm,k,N_TEST,100*pp,100*lo,100*hi,fisher2x2(k,N_TEST-k,K_BAS,N_TEST-K_BAS)))
print("  Positivkontrolle -> %s | Trenner -> %s"%(U_POS,U_MASKE))
# ---------------- Wiederherstellung und Urteil --------------------------------
ABW=float(np.abs(nur_logits(ZTEXT)-L0).max())
print(""); print("  WIEDERHERSTELLUNG: alle Haken abgenommen, Logit-Abweichung %.2e"%ABW)
assert ABW<1e-2,"Haken haengen noch"
CODE=urteil_paar(U_POS,P_RATE,len(TRENNER),P_TR,U_PAAR,U_MASKE)
print(""); print("="*80); print("VERDIKT: %s"%CODE); print("="*80)
if CODE=="MESSFELD-UNEMPFINDLICH":
    print("  Selbst die JA-exklusive Menge bewegt an diesem Praefix nichts. Jeder")
    print("  Nullbefund waere ein Befund ueber den Aufbau, nicht ueber das Modell.")
elif CODE=="KOPFZEILE-OHNE-WIRKUNG":
    print("  A kippt nicht zuverlaessig haeufiger als B (p = %.4f). Dann war die"%P_RATE)
    print("  Rangbeobachtung aus dem letzten Lauf eine Eigenheit der geernteten")
    print("  Praefixe und nicht der Kopfzeilenform - ein sauberes Nein zu H1.")
elif CODE=="AUFLOESUNG-ZU-GROB":
    print("  Bei %d Paaren ist der kleinstmoegliche p-Wert %.5f. Mehr Paare noetig."%(len(PAARE),U_PAAR))
elif CODE=="ZU-WENIG-TRENNER":
    print("  Weniger als drei Experten feuern konsistent auf der A-Seite (nach Abzug")
    print("  der Laengenkontrolle). Die Konsistenzwerte stehen oben: daran laesst")
    print("  sich ablesen, ob etwas knapp danebenlag. Die Festlegung waere dann")
    print("  zwar im Verhalten sichtbar, aber nicht an einer benennbaren")
    print("  Router-Auswahl festgemacht.")
elif CODE=="TRENNER-ZUFAELLIG":
    print("  Konsistente Experten gefunden, aber das Vertauschen von A und B")
    print("  liefert genauso viele. Kein Befund.")
elif CODE=="KOPFZEILE-IM-ROUTER":
    print("  Die Experten, die in JEDEM Paar auf der A-Seite feuern und nicht auf")
    print("  der B-Seite - und die die Laengenkontrolle nicht auch zeigt - zu")
    print("  sperren senkt die Kipprate; gleich viele haeufig aktive nicht. Damit")
    print("  ist die Lesartenwahl an eine benennbare Router-Auswahl gebunden, und")
    print("  zwar an einer Stelle, an der sich sonst NICHTS im Text unterscheidet.")
elif CODE=="NUR-STOERUNG":
    print("  Auch die Zufallskontrolle senkt - auf dieser Groesse wirkt")
    print("  Beschaedigung und nicht Auswahl.")
else:
    print("  Die konsistenten Experten tragen die Kipprate nicht, obwohl die")
    print("  Positivkontrolle am selben Praefix wirkt. Das Messfeld ist also")
    print("  empfindlich und die Antwort ist ein echtes Nein.")
PAAR_RESULTS=dict(verdict=CODE,arch_ok=True,prompt_id=ZIEL_ID,marke=MARKE,inert=INERT,
    n_paare=len(PAARE),n_paar=N_PAAR,n_inert=N_INERT,n_test=N_TEST,schwelle=SCHWELLE,
    raten=[{"rumpf":RUMPFE[i][0],"a":RATE[(i,"A")],"b":RATE[(i,"B")],"inert":IRATE[i]}
           for i in range(len(PAARE))],
    p_rate=P_RATE,plus=PLUS,minus=MINUS,untergrenze=U_PAAR,
    p_inert=P_INERT,inert_plus=IPLUS,
    routing=[[[list(q) for q in a],[list(q) for q in b]] for a,b in ROUT],
    inert_routing=[[[list(q) for q in a],[list(q) for q in b]] for a,b in IROUT],
    konsistenz=[[list(q),float(w)] for q,w in KON[:200]],
    trenner=[list(q) for q in TRENNER],trenner_roh=[list(q) for q in ROH_TR],
    p_trenner=P_TR,trenner_erwartet=n_erw,trenner_beobachtet=n_beob,
    gemeinsam=len(GEMEIN),positiv=[list(q) for q in POS],ziel=ZI,
    k_basis=K_BAS,k_trenner=K_TR,k_zufall=K_ZU,k_positiv=K_POS,
    urteil_positiv=U_POS,urteil_maske=U_MASKE,abweichung_ende=ABW,
    bild=dict(basis=bild(A_BAS),positiv=bild(A_POS),
              trenner=bild(A_TR) if A_TR else None,zufall=bild(A_ZU) if A_ZU else None))
wc_save("antworten_paar",dict(prompt_id=ZIEL_ID,
    paare={"%d%s"%(i,s):ROH[(i,s)] for i,s in ROH},
    basis=A_BAS,trenner=A_TR,zufall=A_ZU,positiv=A_POS))
wc_save_all()
print("")
print("(Der Kontrast ist EINE Einfuegung. Alles andere - Spaltennamen, Trennzeile,")
print(" erste Datenzelle - ist zeichenweise gleich. Was hier trennt, kann nicht am")
print(" uebrigen Text liegen; was die Laengenkontrolle auch zeigt, ist abgezogen.)")
````
