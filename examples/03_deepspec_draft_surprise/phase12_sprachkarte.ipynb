## Die Nachlese hat den Befund zweimal praezisiert

**Der Effekt war groesser als gemessen.** `ist_jp` zaehlte den Han-Bereich
`0x3400-0x9FFF`, und der enthaelt chinesische Zeichen und japanische Kanji gleichermassen.
Trennt man **Kana** ab - die gibt es nur im Japanischen - steht es 48/48 gegen **11/48**,
nicht 48/48 gegen 28/48. 17 der vermeintlich japanischen Antworten waren chinesisch
(`存储限制`, `价格`, `Google云盘`). Diese Zelle zaehlt Kana, Hangul und Han getrennt.

**Was das Modell stattdessen tut, ist der eigentliche Fund.** 35 von 48 Antworten
befolgen die Anweisung noch erkennbar - sie landen nur nicht mehr auf Japanisch:

```
| 서비스 | 기본 저장 공간 | 초기计划 가격 (월) |     koreanisch
| Dienst名稱(Japanisch) | Storage Limit |          deutsch + Han
| Service (Japanese Name) | Google Drive |         Kopf nennt es, Zellen englisch
```

Die 42 Experten tragen also nicht, *dass* lokalisiert werden soll, sondern **welche
Sprache dabei herauskommt**. Der Eingriff schneidet die Kette (a) Lesart "lokalisieren"
-> (b) Sprachwahl genau dazwischen durch. Strukturell sah der maskierte Arm dabei aus wie
der voellig unangetastete NEU-Arm (15 gegen 14 saubere Tabellen, 2 gegen 3 Schleifen) -
der Effekt ist kein Zerfall.

## Vier Arme aus demselben Prompt

| Arm | Phrase | erwartet |
|---|---|---|
| JA | `each service's Japanese name` | Kana |
| KO | `each service's Korean name` | Hangul |
| LOC | `each service's local name` | der urspruengliche, mehrdeutige |
| NEU | `each service's name` | Bezugsarm ohne Adjektiv |

## Fuenf Pruefungen

1. **Hauptversuch** - die LOC-exklusiven Experten sperren, **Kipprate am LOC-Prompt**
   messen, gegen gleich viele zufaellige *gemeinsame*. Das ist der direkte Schuss auf die
   Ausgangsfrage der ganzen Untersuchung.
2. **Spezifitaet** - die KO-exklusiven sperren und den JA-Arm messen.
3. **Spezifitaet** - die JA-exklusiven sperren und den KO-Arm messen.
4. **Kreuz** - die JA-exklusiven sperren und den LOC-Arm messen.
5. **Ueberlappung** - sind die exklusiven Mengen fuer Japanisch und Koreanisch dieselben?
   Nulltest: je Schicht gleich viele Experten neu ziehen, nur aus denen, die NEU dort
   nicht benutzt.

Sind die Mengen weitgehend gleich **und** wirken die Kreuzversuche, ist es ein generisches
"nicht-englische Schrift"-Buendel. Sind sie verschieden **und** wirken die Kreuzversuche
nicht, sind es Sprachidentitaeten.

**Vorregistriert:** Pruefung 1 muss die Kipprate *senken*, ihre gleich grosse
Zufallskontrolle darf es nicht. Die uebrigen sind Kontrollen und registrieren keine
Richtung.

| Verdikt | Lesart |
|---|---|
| `LOKAL-ROUTING-TRAEGT` | die Sprachwahl beim mehrdeutigen Prompt sitzt an derselben Stelle wie bei der ausdruecklichen Anweisung |
| `ZU-WENIG-EXKLUSIVE` | ` local` aendert das Routing am Prompt-Ende kaum - dann faellt die Wahl erst spaeter, passend zum Entscheidungspunkt bei Zeichen 90-130 |
| `BLIND` | die LOC-exklusiven tragen die Kipprate nicht, obwohl dasselbe Verfahren bei der ausdruecklichen Anweisung wirkte |
| `NUR-STOERUNG` | auch die Kontrolle senkt - Groesse zu hoch |

## Drei Reparaturen aus dem letzten Lauf

* Kana/Hangul/Han getrennt statt eines gemeinsamen CJK-Zaehlers
* **alle** Arme werden gespeichert - beim letzten Mal fehlten die des geteilten Experten,
  weshalb der Zerfallsverdacht dort offen bleiben musste
* je Arm werden Sprache **und** Zerfall ausgewiesen (saubere Tabelle, Schleife, Laenge),

````python
# === PHASE 12 - SPRACHKARTE IM ROUTER ========================================
# Der Maskenlauf hat gezeigt: die 42 (Schicht,Experte)-Paare, die nur im
# JP-Arm feuern, aus dem Gemisch zu nehmen senkt die Japanisch-Rate von 100%
# auf 58%, waehrend 42 zufaellige GEMEINSAME Paare bei 100% bleiben - bei
# fast gleich vielen gesperrten Router-Plaetzen (453 gegen 434).
#
# DIE OFFLINE-NACHLESE HAT DEN BEFUND ZWEIMAL PRAEZISIERT
#
# (1) Der Effekt war GROESSER als gemessen. ist_jp zaehlte den Han-Bereich
#     0x3400-0x9FFF, und der enthaelt chinesische Zeichen und japanische
#     Kanji gleichermassen. Trennt man Kana ab - die gibt es nur im
#     Japanischen - steht es 48/48 gegen 11/48, nicht 48/48 gegen 28/48.
#     17 der vermeintlich japanischen Antworten waren CHINESISCH
#     ('存储限制', '价格', 'Google云盘'). Deshalb zaehlt diese Zelle Kana,
#     Hangul und Han GETRENNT.
#
# (2) Was das Modell stattdessen tut, ist der eigentliche Fund. 35 von 48
#     Antworten befolgen die Anweisung noch erkennbar - sie landen nur nicht
#     mehr auf Japanisch:
#         | 서비스 | 기본 저장 공간 | 초기计划 가격 (월) |     koreanisch
#         | Dienst名稱(Japanisch) | Storage Limit |          deutsch + Han
#         | Service (Japanese Name) | Google Drive |         Kopf nennt es,
#                                                            Zellen englisch
#     Die 42 Experten tragen also nicht, DASS lokalisiert werden soll,
#     sondern WELCHE SPRACHE dabei herauskommt. Der Eingriff schneidet die
#     Kette (a) Lesart 'lokalisieren' -> (b) Sprachwahl genau dazwischen
#     durch. Strukturell sah der maskierte Arm dabei aus wie der voellig
#     unangetastete NEU-Arm (15 gegen 14 saubere Tabellen, 2 gegen 3
#     Schleifen) - der Effekt ist kein Zerfall.
#
# DAMIT IST DER DIREKTE SCHUSS AUF DIE AUSGANGSFRAGE MOEGLICH
#
# Die ganze Untersuchung dreht sich um ' local': das Modell liest die
# Anweisung als 'lokalisieren' und waehlt dann eine Sprache. Wenn die
# Sprachwahl im Routing sitzt, muss dasselbe Verfahren am Originalprompt
# wirken: die Experten sperren, die NUR bei ' local' feuern, und die
# Kipprate messen.
#
# VIER ARME AUS DEMSELBEN PROMPT
#   JA   "each service's Japanese name"   Kana erwartet
#   KO   "each service's Korean name"     Hangul erwartet
#   LOC  "each service's local name"      der urspruengliche, mehrdeutige
#   NEU  "each service's name"            der Bezugsarm ohne Adjektiv
#
# FUENF PRUEFUNGEN
#   I   HAUPTVERSUCH: die LOC-exklusiven Experten sperren, Kipprate am
#       LOC-Prompt messen, gegen gleich viele zufaellige GEMEINSAME
#   II  SPEZIFITAET: die KO-exklusiven sperren und den JA-Arm messen
#   III SPEZIFITAET: die JA-exklusiven sperren und den KO-Arm messen
#   IV  KREUZ: die JA-exklusiven sperren und den LOC-Arm messen
#   V   UEBERLAPPUNG: sind die exklusiven Mengen fuer Japanisch und
#       Koreanisch dieselben? Nulltest gegen zufaellige Mengen gleicher
#       Groesse aus den je Schicht nicht von NEU benutzten Experten.
#
# Sind die Mengen weitgehend gleich UND wirken die Kreuzversuche, dann ist es
# ein generisches 'nicht-englische Schrift'-Buendel. Sind sie verschieden UND
# wirken die Kreuzversuche nicht, sind es Sprachidentitaeten.
#
# VORREGISTRIERT: Pruefung I muss die Kipprate SENKEN, ihre gleich grosse
# Zufallskontrolle darf es nicht. Die uebrigen Pruefungen sind Kontrollen
# und registrieren keine Richtung.
#
# DREI REPARATUREN AUS DEM LETZTEN LAUF
#   - Kana/Hangul/Han getrennt statt eines gemeinsamen CJK-Zaehlers
#   - ALLE Arme werden gespeichert (beim letzten Mal fehlten die des
#     geteilten Experten, weshalb der Zerfallsverdacht dort offen bleibt)
#   - je Arm werden Sprache UND Zerfall ausgewiesen, damit sich Ursache und
#     Beschaedigung nicht wieder vermischen
import os
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF","expandable_segments:True")
import re, math, torch, collections, unicodedata, random
import numpy as np, glob, json, gc, sys, time
gc.collect(); torch.cuda.empty_cache()
try: torch.cuda.synchronize()
except Exception: pass
_free=torch.cuda.mem_get_info()[0]/1e9
if "model" not in globals() and _free<45:
    raise RuntimeError("GPU nicht leer genug (%.1f GB frei, ~45 noetig)."%_free)
if not os.path.isdir("/content/drive/MyDrive"):
    from google.colab import drive; drive.mount("/content/drive")
# ---------------- Protokoll und Abbildungen automatisch nach Drive ----------
# Der PDF-Export von Colab schneidet die Ausgabe unzuverlaessig ab. Deshalb
# schreibt jede Zelle ihr vollstaendiges Protokoll und jede Abbildung selbst
# nach Drive - unabhaengig davon, was der Export spaeter mitnimmt.
import sys, time
WC_RUN=globals().get("WC_RUN","phase12_sprachkarte")
RUN_OUT="/content/drive/MyDrive/WeirdChat_Runs/%s_%s"%(WC_RUN,time.strftime("%Y%m%d-%H%M%S"))
os.makedirs(RUN_OUT,exist_ok=True)
class _WCTee:
    _wc_tee=True
    def __init__(self,p,o): self.o=o; self.f=None; self.retarget(p)
    def retarget(self,p):
        try:
            if self.f: self.f.close()
        except Exception: pass
        try: self.f=open(p,"a",encoding="utf-8")
        except Exception: self.f=None
    def write(self,s):
        self.o.write(s)
        if self.f:
            try: self.f.write(s); self.f.flush()
            except Exception: pass
        return len(s)
    def flush(self):
        self.o.flush()
        if self.f:
            try: self.f.flush()
            except Exception: pass
    def isatty(self): return False
_wc_log=os.path.join(RUN_OUT,"protokoll.txt")
if getattr(sys.stdout,"_wc_tee",False): sys.stdout.retarget(_wc_log)
else: sys.stdout=_WCTee(_wc_log,sys.stdout)
try:
    import matplotlib.pyplot as _wcplt
    if not getattr(_wcplt,"_wc_patched",False):
        _wc_orig_show=_wcplt.show; _wc_fig=[0]
        def _wc_show(*a,**k):
            for _num in _wcplt.get_fignums():
                _wc_fig[0]+=1
                try:
                    _wcplt.figure(_num).savefig(os.path.join(RUN_OUT,"abb_%02d.png"%_wc_fig[0]),
                                                dpi=150,bbox_inches="tight")
                except Exception: pass
            return _wc_orig_show(*a,**k)
        _wcplt.show=_wc_show; _wcplt._wc_patched=True
except Exception: pass
def wc_save(name,obj):
    """Ergebnisobjekt als JSON neben das Protokoll legen"""
    def _e(o):
        if isinstance(o,np.ndarray): return o.tolist()
        if isinstance(o,(np.integer,)): return int(o)
        if isinstance(o,(np.floating,)): return float(o)
        if isinstance(o,(np.bool_,)): return bool(o)
        return str(o)
    try:
        with open(os.path.join(RUN_OUT,name+".json"),"w",encoding="utf-8") as f:
            json.dump(obj,f,ensure_ascii=False,indent=1,default=_e)
        print("gespeichert: %s.json"%name)
    except Exception as _ex: print("konnte %s nicht speichern: %s"%(name,_ex))
def wc_save_all():
    """alle *_RESULTS aus dem Namensraum sichern - Aufruf am Zellenende"""
    for _k in [k for k in list(globals()) if k.endswith("_RESULTS")]:
        wc_save(_k,globals()[_k])
    print("Lauf-Ordner:",RUN_OUT)
print("Lauf-Ordner (Protokoll + Abbildungen):",RUN_OUT)
if "PROMPTS" not in globals():
    _h=glob.glob("/content/drive/MyDrive/**/weird_transcripts.jsonl",recursive=True)
    assert _h, "weird_transcripts.jsonl nicht gefunden"
    PROMPTS={}
    with open(_h[0],encoding="utf-8") as _f:
        for _line in _f:
            _line=_line.strip()
            if not _line: continue
            _r=json.loads(_line)
            _pid=str(_r["id"]).split("/")[0]
            if _pid not in PROMPTS:
                try: PROMPTS[_pid]=next(t["content"] for t in _r["conversations"] if t["role"]=="user")
                except StopIteration: pass
    PROMPT_IDS=sorted(PROMPTS)
    print("PROMPTS geladen: %d"%len(PROMPTS))
if "model" not in globals() or "tokenizer" not in globals():
    from transformers import AutoModelForCausalLM, AutoTokenizer
    MODEL_ID=globals().get("MODEL_ID","Qwen/Qwen3.6-35B-A3B-FP8")
    print("lade Instruct-Modell:",MODEL_ID,"(einige Minuten)")
    tokenizer=AutoTokenizer.from_pretrained(MODEL_ID)
    model=AutoModelForCausalLM.from_pretrained(MODEL_ID,device_map="auto",torch_dtype="auto")
    model.eval()
    print("geladen | dtype:",next(model.parameters()).dtype)
# ---------------- reine Logik (offline geprueft) ----------------------------
PHRASE="each service's local name"
FRW=[(0x0370,0x03FF),(0x0400,0x052F),(0x0530,0x058F),(0x0590,0x05FF),(0x0600,0x074F),
     (0x0900,0x097F),(0x0E00,0x0E7F),(0x3040,0x30FF),(0x3400,0x9FFF),(0xAC00,0xD7AF),
     (0xF900,0xFAFF)]
JPW=[(0x3040,0x30FF),(0x3400,0x9FFF),(0xF900,0xFAFF)]
FRS=set("le la les une un des est et pour avec dans votre vous voici bonjour du qui que "
        "sur cette ces aux ou par plus il elle nous sont".split())
ENS=set("the is and for with in your you here of to that this are was were has have will "
        "would can it on as at be by".split())
PTES=set("nome nomes servico servicos armazenamento limite limites preco mes gratuito "
         "conta cada para com uma nao mais seu sua nombre servicio servicios "
         "almacenamiento precio cuenta los las del con mas su".split())
DES=set("name dienst dienste speicher speicherplatz grenze preis monat kostenlos konto "
        "jeder fuer mit eine der die das und nicht mehr uebersicht zusammenfassung".split())
def _in(c,bereiche):
    o=ord(c); return any(a<=o<=b for a,b in bereiche)
def _fremd(s):
    return [c for c in s if c.isalpha() and ord(c)>=0x250 and _in(c,FRW)]
def _srun(t,run=3):
    c=0
    for ch in t:
        if ch.isalpha() and ord(ch)>=0x250 and _in(ch,FRW):
            c+=1
            if c>=run: return True
        elif ch.isalpha(): c=0
    return False
def _entakz(s):
    return "".join(c for c in unicodedata.normalize("NFD",s) if not unicodedata.combining(c))
def classify_answer(t):
    if not t.strip(): return "empty"
    al=[c for c in t if c.isalpha()]; fo=_fremd(t)
    if al and len(fo)/len(al)>=0.5: return "takeover"
    if _srun(t): return "gloss"
    w=re.findall(r"[a-zA-ZÀ-ſ']+",t.lower())
    fr=sum(1 for x in w if x in FRS); en=sum(1 for x in w if x in ENS)
    return "latin-switch(fr)" if (fr>=3 and fr>en) else "english"
def classify_breit(t):
    c=classify_answer(t)
    if c!="english": return c
    w=re.findall(r"[a-zA-ZÀ-ſ']+",_entakz(t).lower())
    en=sum(1 for x in w if x in ENS)
    for lab,S in (("pt/es",PTES),("de",DES)):
        n=sum(1 for x in w if x in S)
        if n>=3 and n>en: return "latin-switch(%s)"%lab
    if sum(1 for c2 in t if c2.isalpha() and 0xC0<=ord(c2)<=0x17F)>=3: return "latin-akzent"
    return "english"
SW=("takeover","gloss","latin-switch(fr)")
SWB=SW+("latin-switch(pt/es)","latin-switch(de)","latin-akzent")
def ist_jp(t,mindest=3):
    """Eigener Zaehler fuer die Positivkontrolle. Im JP-Arm ist japanische
       Schrift das ERWUENSCHTE Verhalten - classify_breit wuerde sie
       'takeover' nennen, was hier irrefuehrend waere. Gezaehlt wird ein Lauf
       von mindestens 3 Kana-/Kanji-Zeichen: einzelne Zeichen kommen auch in
       englischen Antworten als Beispiel vor, ein Lauf nicht."""
    c=0
    for ch in t:
        if _in(ch,JPW):
            c+=1
            if c>=mindest: return True
        elif ch.isalpha(): c=0
    return False
def wiederholt(t,fenster=12,mal=4):
    """Zerfallsmerkmal: dieselbe Zeichenfolge viermal. Bei starker Beschaedigung
       faellt ein Modell in Schleifen, lange bevor es verstummt."""
    if len(t)<fenster*mal: return False
    z=collections.Counter(t[i:i+fenster] for i in range(len(t)-fenster+1))
    return max(z.values())>=mal
def zerfall(texte):
    """Was sagt die Antwortform ueber den Schaden - unabhaengig von der Sprache"""
    if not texte: return dict(leer=0.0,laenge=0.0,schleife=0.0)
    return dict(leer=sum(1 for t in texte if not t.strip())/len(texte),
                laenge=sum(len(t) for t in texte)/len(texte),
                schleife=sum(1 for t in texte if wiederholt(t))/len(texte))
def wilson(k,n,z=1.96):
    if n==0: return (0.,0.,0.)
    p=k/n; d=1+z*z/n; c=p+z*z/(2*n); h=z*math.sqrt(p*(1-p)/n+z*z/(4*n*n))
    return p,(c-h)/d,(c+h)/d
def fisher2x2(a,b,c,d,einseitig=False):
    from math import lgamma,exp
    lf=lambda n: lgamma(n+1); n=a+b+c+d
    def pr(x):
        y=a+b-x; z=a+c-x; w=n-x-y-z
        if min(y,z,w)<0: return 0.0
        return exp(lf(a+b)+lf(c+d)+lf(a+c)+lf(b+d)-lf(n)-lf(x)-lf(y)-lf(z)-lf(w))
    hi=min(a+b,a+c)
    if einseitig: return min(1.0,sum(pr(x) for x in range(a,hi+1)))
    p0=pr(a)
    return min(1.0,sum(pr(x) for x in range(0,hi+1) if pr(x)<=p0*(1+1e-9)))
def phrase_mit(w):
    return "each service's name" if not w else "each service's %s name"%w
def setze_arm(text,ersatz):
    if PHRASE not in text: return text,False
    return text.replace(PHRASE,ersatz),True
def massstab(*vektoren):
    """EIN globaler, robuster Massstab fuer alle Einheiten. Ersetzt den
       Nenner je Einheit aus v2, der auf 1e-8 fallen konnte und damit
       Trennwerte von 1e7 erzeugt hat. Median statt Mittelwert, weil die
       Zwischenschicht duennbesetzt ist und wenige grosse Werte den Mittelwert
       tragen wuerden. Nullen zaehlen NICHT mit: bei 70% strukturellen Nullen
       waere der Median sonst selbst null."""
    v=np.abs(np.concatenate([np.asarray(x,dtype=np.float64).ravel() for x in vektoren]))
    v=v[v>0]
    if v.size==0: return 1.0
    m=float(np.median(v))
    return m if m>0 else 1.0
def trennung_zwei(xa,xb,s):
    """Differenz zweier Zustaende in Einheiten EINES globalen Massstabs.
       Positiv = im ersten Zustand hoeher. Beschraenkt und vergleichbar."""
    return (np.asarray(xa,dtype=np.float64)-np.asarray(xb,dtype=np.float64))/float(s)
def waehle_einheiten(d,k):
    """die k Einheiten mit der groessten POSITIVEN Trennung - Richtung ist
       vorregistriert: im JP-Zustand hoeher, Ablation muss die JP-Rate senken"""
    return list(np.argsort(-np.asarray(d))[:k])
KANA=[(0x3040,0x30FF)]
HANGUL=[(0xAC00,0xD7AF),(0x1100,0x11FF),(0x3130,0x318F)]
HAN=[(0x3400,0x9FFF),(0xF900,0xFAFF)]
def _zaehl(t,bereiche):
    return sum(1 for c in t if _in(c,bereiche))
def sprache(t):
    """Kana beweist Japanisch, Hangul beweist Koreanisch, Han allein nicht:
       der Han-Bereich enthaelt chinesische Zeichen und japanische Kanji
       gleichermassen. Genau daran ist der letzte Lauf halb vorbeigemessen -
       17 als 'japanisch' gezaehlte Antworten waren chinesisch.

       Reihenfolge ist Absicht: eine Antwort mit Kana UND Han ist japanisch,
       eine mit Hangul und ein paar Han-Zeichen koreanisch."""
    if not t.strip(): return "leer"
    k=_zaehl(t,KANA); g=_zaehl(t,HANGUL); h=_zaehl(t,HAN)
    if k>=2: return "japanisch"
    if g>=2: return "koreanisch"
    if h>=3: return "han"
    if h>0:  return "han-einzeln"
    if sum(1 for c in t if c.isalpha() and 0xC0<=ord(c)<=0x17F)>=3: return "latein-akzent"
    return "lateinisch"
def anteil(texte,pruef):
    return sum(1 for t in texte if pruef(t))
def ist_ja(t): return sprache(t)=="japanisch"
def ist_ko(t): return sprache(t)=="koreanisch"
def lokalisiert(t,marke):
    """Befolgt die Antwort die Anweisung ueberhaupt noch? Entweder steht ein
       fremdes Schriftzeichen darin, oder die Kopfzeile nennt die Sprache
       beim Namen ('Service (Japanese Name)') - beides ist Befolgung, nur die
       zweite ohne Ausfuehrung."""
    return (_zaehl(t,KANA)+_zaehl(t,HANGUL)+_zaehl(t,HAN))>0 or marke.lower() in t.lower()
def tabelle_sauber(t):
    """echte Markdown-Tabelle: Kopfzeile, Trennzeile, gleiche Spaltenzahl.
       Zerfallsmass NEBEN der Sprache - beim letzten Lauf liess sich sonst
       nicht trennen, ob eine Antwort die Anweisung verloren hat oder das
       Modell beschaedigt war."""
    z=[x.strip() for x in t.splitlines() if x.strip().startswith("|")]
    if len(z)<3: return False
    if not re.match(r"^\|[\s:|-]+\|?$",z[1]): return False
    n=[x.count("|") for x in z]
    return max(n)-min(n)<=1
def zu_einheit(i,paare,breite):
    return paare[i//breite],i%breite
def exklusiv(za,zb):
    """Paare, die in za feuern und in zb nicht"""
    return sorted(set(za)-set(zb))
def gemeinsam(za,zb):
    return sorted(set(za)&set(zb))
def nach_schicht(paare):
    d=collections.defaultdict(set)
    for l,e in paare: d[int(l)].add(int(e))
    return dict(d)
def jaccard(A,B):
    A=set(A); B=set(B)
    return len(A&B)/len(A|B) if (A or B) else 0.0
def ueberlappungs_null(A,B,ref,n_experten,rnd,perm=2000):
    """Wie gross waere die Ueberlappung zweier exklusiver Mengen zufaellig?
       Je Schicht gleich viele Experten neu ziehen - aber NUR aus denen, die
       der Bezugsarm dort nicht benutzt. Sonst waere die Nullverteilung zu
       breit und jede Ueberlappung saehe bedeutsam aus."""
    RA=nach_schicht(A); RB=nach_schicht(B); RR=nach_schicht(ref)
    beob=len(set(A)&set(B))
    treffer=0; werte=[]
    for _ in range(perm):
        n=0
        for l in set(RA)|set(RB):
            frei=[e for e in range(n_experten) if e not in RR.get(l,set())]
            a=rnd.sample(frei,min(len(RA.get(l,())),len(frei)))
            b=rnd.sample(frei,min(len(RB.get(l,())),len(frei)))
            n+=len(set(a)&set(b))
        werte.append(n)
        if n>=beob: treffer+=1
    return beob,(treffer+1)/(perm+1.0),float(np.mean(werte))
def urteil_dosis(k_bas,n_bas,k_abl,n_abl,alpha=0.05):
    if n_bas==0 or n_abl==0: return "still"
    p=fisher2x2(k_abl,n_abl-k_abl,k_bas,n_bas-k_bas)
    if p>=alpha: return "still"
    return "senkt" if k_abl/n_abl<k_bas/n_bas else "hebt"
def urteil_arm(k_bas,n_bas,k_aus,n_aus,k_zuf,n_zuf,alpha=0.05):
    a=urteil_dosis(k_bas,n_bas,k_aus,n_aus,alpha)=="senkt"
    z=urteil_dosis(k_bas,n_bas,k_zuf,n_zuf,alpha)=="senkt"
    if a and not z: return "TRAEGT"
    if a and z:     return "NUR-STOERUNG"
    if z and not a: return "WIDERSPRUECHLICH"
    return "BLIND"
def urteil_haupt(n_exkl,loc_basis,u_loc,mindest_exkl=5,mindest_rate=0.15):
    """Vorregistriert: die LOC-exklusiven Experten zu sperren muss die
       Kipprate senken, die gleich grosse Zufallskontrolle nicht. Zwei
       Sanitaetschecks davor, damit ein Nein aus dem richtigen Grund kommt."""
    if n_exkl<mindest_exkl: return "ZU-WENIG-EXKLUSIVE"
    if loc_basis<mindest_rate: return "KEINE-GRUNDRATE"
    return {"TRAEGT":"LOKAL-ROUTING-TRAEGT","NUR-STOERUNG":"NUR-STOERUNG",
            "WIDERSPRUECHLICH":"WIDERSPRUECHLICH","BLIND":"BLIND"}[u_loc]
def urteil_spez(p_ueberlappung,kreuz,alpha=0.05):
    """kreuz: Liste der Urteile aus den Kreuzversuchen (KO-Menge auf den
       JA-Arm und umgekehrt). Wirken sie, ist es ein gemeinsames Buendel."""
    viel=(p_ueberlappung<alpha)
    wirkt=any(u=="senkt" for u in kreuz)
    if viel and wirkt: return "GEMEINSAMES-BUENDEL"
    if not viel and not wirkt: return "SPRACHSPEZIFISCH"
    return "UNKLAR"
# ---------------- Ausfuehrung ------------------------------------------------
if "PROMPTS" not in globals():
    _h=glob.glob("/content/drive/MyDrive/**/weird_transcripts.jsonl",recursive=True)
    assert _h,"weird_transcripts.jsonl nicht gefunden"
    PROMPTS={}
    with open(_h[0],encoding="utf-8") as _f:
        for _l in _f:
            _l=_l.strip()
            if not _l: continue
            _r=json.loads(_l); _pid=str(_r["id"]).split("/")[0]
            if _pid not in PROMPTS:
                try: PROMPTS[_pid]=next(t["content"] for t in _r["conversations"]
                                        if t["role"]=="user")
                except StopIteration: pass
N_ARM=int(globals().get("N_ARM",48)); MAX_NEW=int(globals().get("MAX_NEW",64))
CHUNK=int(globals().get("CHUNK",16)); TEMP=float(globals().get("TEMP",1.0))
SEED=int(globals().get("SEED",20260811)); PERM=int(globals().get("PERM",2000))
def prompt_text(u):
    return "<|im_start|>user\n"+u+"<|im_end|>\n<|im_start|>assistant\n<think>\n\n</think>\n\n"
ZIEL_ID=globals().get("ZIEL_ID","") or next(p for p in PROMPTS if PHRASE in PROMPTS[p])
ROH_PROMPT=PROMPTS[ZIEL_ID]
if tokenizer.pad_token_id is None: tokenizer.pad_token=tokenizer.eos_token
tokenizer.padding_side="left"
# ---------------- 0  Architektur --------------------------------------------
print("="*80); print("0  ARCHITEKTUR"); print("="*80)
cfg=model.config
RX=re.compile(r"^(?:model\.)?(?:language_model\.)?(?:model\.)?layers\.(\d+)\.mlp\.experts$")
EXPM={}
for nm,mod in model.named_modules():
    m=RX.match(nm)
    if m: EXPM[int(m.group(1))]=mod
ARCH_OK=bool(EXPM)
if ARCH_OK:
    e0=EXPM[min(EXPM)]
    GU=e0.gate_up_proj; INTER=int(e0.intermediate_dim); NEXP=int(GU.shape[0])
    ARCH_OK=(GU.ndim==3 and GU.shape[1]==2*INTER and GU.shape[2]==cfg.hidden_size)
    print("  %d Schichten | %d Experten je Schicht | Zwischenbreite %d | top-%d"
          %(len(EXPM),NEXP,INTER,cfg.num_experts_per_tok))
    print("  Formen wie erwartet: %s"%("ja" if ARCH_OK else "NEIN"))
if not ARCH_OK:
    KARTE_RESULTS=dict(verdict="ARCHITEKTUR-NICHT-GEFUNDEN",arch_ok=False)
    wc_save_all(); print(""); print("VERDIKT: ARCHITEKTUR-NICHT-GEFUNDEN"); raise SystemExit(0)
# ---------------- Werkzeuge ---------------------------------------------------
def zieh(text,n,startwert):
    aus=[]
    for b0 in range(0,n,CHUNK):
        b=min(CHUNK,n-b0)
        enc=tokenizer([text]*b,return_tensors="pt",padding=True).to(model.device)
        torch.manual_seed(startwert+b0)
        with torch.no_grad():
            g=model.generate(**enc,do_sample=True,temperature=TEMP,top_p=1.0,top_k=0,
                             repetition_penalty=1.0,max_new_tokens=MAX_NEW,
                             pad_token_id=tokenizer.pad_token_id)
        for j in range(b):
            aus.append(tokenizer.decode(g[j,enc["input_ids"].shape[1]:],skip_special_tokens=True))
    return aus
class Maske:
    """Setzt den Router-Anteil ganzer Experten auf null. Der Block rechnet
       out = summe_e down_proj(h_e)*gewicht_e, also entfernt gewicht_e = 0 den
       Experten vollstaendig. Reiner Vorwaerts-Haken: kein Sicherungsabzug,
       kein Speicher, wirkt an JEDER Position, restlos umkehrbar.

       Die Buchfuehrung laeuft nur bei 'pruefen' - waehrend der Erzeugung
       wuerde ein .item() je Schicht und Token die Zelle unnoetig bremsen."""
    def __init__(self,verboten):
        self.verboten={l:set(v) for l,v in verboten.items() if v}
        self.griffe=[]; self.pruefen=False
        self.getroffen=0; self.rest=0.0; self.formen=None
    def _mach(self,bad):
        def h(mod,args):
            idx=args[1]; w=args[2]
            if idx.shape!=w.shape:
                self.formen=(tuple(idx.shape),tuple(w.shape)); return None
            tr=torch.isin(idx,bad)
            neu=w.masked_fill(tr,0.0)
            if self.pruefen:
                self.getroffen+=int(tr.sum().item())
                if bool(tr.any()):
                    self.rest=max(self.rest,float(neu[tr].abs().max().item()))
            return (args[0],idx,neu)+tuple(args[3:])
        return h
    def __enter__(self):
        for l,vs in self.verboten.items():
            W=EXPM[l].gate_up_proj
            bad=torch.tensor(sorted(vs),device=W.device,dtype=torch.long)
            self.griffe.append(EXPM[l].register_forward_pre_hook(self._mach(bad)))
        return self
    def __exit__(self,*a):
        for g in self.griffe: g.remove()
        self.griffe=[]
        return False
def hole_routing(text):
    """welche (Schicht,Experte) feuern am LETZTEN Token. Der letzte Token ist
       in allen Armen derselbe - nur der Zusammenhang davor unterscheidet sich."""
    ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
    fang={}
    def mach(l):
        def h(mod,args): fang[l]=args[1].detach(); return None
        return h
    hs=[EXPM[l].register_forward_pre_hook(mach(l)) for l in EXPM]
    try:
        with torch.no_grad():
            o=model(ids[:,:-1],use_cache=True); fang.clear()
            model(ids[:,-1:],past_key_values=o.past_key_values,use_cache=True)
    finally:
        for h in hs: h.remove()
    r=set()
    for l,idx in fang.items():
        for e in idx.reshape(-1,idx.shape[-1])[-1].tolist(): r.add((l,int(e)))
    return sorted(r)
def nur_logits(text):
    ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
    with torch.no_grad():
        return model(ids).logits[0,-1].float().cpu().numpy()
def mit_maske(verboten,text,n,startwert,pruef_text,pruef_logit):
    with Maske(verboten) as M:
        M.pruefen=True
        lg=nur_logits(pruef_text)
        M.pruefen=False
        if M.formen is not None:
            raise RuntimeError("Router-Formen passen nicht zueinander: idx %s, "
                               "gewichte %s - die Maske greift nicht"%M.formen)
        wirk=float(np.abs(lg-pruef_logit).max())
        aus=zieh(text,n,startwert)
    return aus,M.getroffen,M.rest,wirk
def bild(texte):
    """Sprache UND Zerfall - getrennt, damit sich Ursache und Beschaedigung
       nicht wieder vermischen"""
    c=collections.Counter(sprache(t) for t in texte); n=max(len(texte),1)
    return dict(n=len(texte),japanisch=c["japanisch"],koreanisch=c["koreanisch"],
                han=c["han"],han_einzeln=c["han-einzeln"],
                latein_akzent=c["latein-akzent"],lateinisch=c["lateinisch"],
                leer=c["leer"],kippt=sum(1 for t in texte if classify_breit(t) in SWB),
                tabelle=sum(1 for t in texte if tabelle_sauber(t)),
                schleife=sum(1 for t in texte if wiederholt(t)),
                laenge=sum(len(t) for t in texte)/n)
def kopf():
    print("  %-30s %5s %5s %5s %5s %5s %5s | %5s %5s %5s"
          %("Arm","n","ja","ko","han","lat","kippt","Tab.","Schl.","Zei."))
def zeile(nm,b):
    print("  %-30s %5d %5d %5d %5d %5d %5d | %5d %5d %5.0f"
          %(nm,b["n"],b["japanisch"],b["koreanisch"],b["han"]+b["han_einzeln"],
            b["lateinisch"]+b["latein_akzent"],b["kippt"],b["tabelle"],
            b["schleife"],b["laenge"]))
def vergleich(nm,k,n,kb,nb):
    pp,lo,hi=wilson(k,n)
    p=fisher2x2(k,n-k,kb,nb-kb)
    print("  %-30s %3d/%-4d %5.1f%% [%4.1f,%4.1f]  p=%.4f"%(nm,k,n,100*pp,100*lo,100*hi,p))
    return p
# ---------------- 1  Vier Arme ------------------------------------------------
print(""); print("="*80); print("1  VIER ARME AUS DEMSELBEN PROMPT"); print("="*80)
ARME={}
for schl,wort in (("JA","Japanese"),("KO","Korean"),("LOC","local"),("NEU","")):
    txt,ok=setze_arm(ROH_PROMPT,phrase_mit(wort))
    assert ok,"Phrase %r nicht im Prompt gefunden"%PHRASE
    ARME[schl]=prompt_text(txt)
    print("  %-4s ...%s..."%(schl,phrase_mit(wort)))
t0=time.time(); TEXTE={}
for i,schl in enumerate(("JA","KO","LOC","NEU")):
    TEXTE[schl]=zieh(ARME[schl],N_ARM,SEED+1+i)
print("  (%.0f s)"%(time.time()-t0)); print("")
BILD={s:bild(TEXTE[s]) for s in TEXTE}
kopf()
for s in ("JA","KO","LOC","NEU"): zeile("%s ohne Eingriff"%s,BILD[s])
print("  (ja = Kana, ko = Hangul, han = Han ohne Kana/Hangul - meist Chinesisch)")
K_LOC=BILD["LOC"]["kippt"]
# ---------------- 2  Routing und exklusive Mengen -----------------------------
print(""); print("="*80); print("2  ROUTING UND EXKLUSIVE MENGEN"); print("="*80)
ROUT={s:hole_routing(ARME[s]) for s in ARME}
EXK={s:exklusiv(ROUT[s],ROUT["NEU"]) for s in ("JA","KO","LOC")}
GEM={s:gemeinsam(ROUT[s],ROUT["NEU"]) for s in ("JA","KO","LOC")}
print("  %-6s %10s %12s %12s"%("Arm","aktiv","nur dieser","gemeinsam"))
for s in ("JA","KO","LOC","NEU"):
    print("  %-6s %10d %12s %12s"%(s,len(ROUT[s]),
          len(EXK[s]) if s in EXK else "-",len(GEM[s]) if s in GEM else "-"))
rnd=random.Random(SEED)
# Bezugspunkt VOR jedem Eingriff. Ein Vergleich zweier frisch gerechneter
# Logits waere immer null und wuerde nichts pruefen.
L_REF=nur_logits(ARME["JA"])
print("")
print("  UEBERLAPPUNG der exklusiven Mengen (Nulltest: gleich viele Experten je")
print("  Schicht neu ziehen, nur aus denen, die NEU dort nicht benutzt)")
UEB={}
for a,b in (("JA","KO"),("JA","LOC"),("KO","LOC")):
    n,p,erw=ueberlappungs_null(EXK[a],EXK[b],ROUT["NEU"],NEXP,rnd,PERM)
    UEB["%s|%s"%(a,b)]=dict(gemeinsam=n,p=p,erwartet=erw,jaccard=jaccard(EXK[a],EXK[b]))
    print("    %-9s %2d gemeinsam (zufaellig %.2f erwartet) | Jaccard %.3f | p %.4f"
          %("%s & %s"%(a,b),n,erw,jaccard(EXK[a],EXK[b]),p))
P_JAKO=UEB["JA|KO"]["p"]
# ---------------- 3  Hauptversuch ---------------------------------------------
print(""); print("="*80); print("3  HAUPTVERSUCH - LOC-EXKLUSIVE EXPERTEN SPERREN")
print("  vorregistriert: muss die KIPPRATE am LOC-Prompt senken,")
print("  die gleich grosse Zufallskontrolle aus den GEMEINSAMEN nicht")
CODE=None; U_LOC=None; A_LOC=A_ZUF=None
if len(EXK["LOC"])<5:
    print("")
    print("  nur %d LOC-exklusive Paare - zu wenige fuer einen Versuch."%len(EXK["LOC"]))
    print("  Das ist selbst ein Befund: ' local' aendert die Router-Auswahl am")
    print("  letzten Prompt-Token kaum, obwohl es das Verhalten stark aendert.")
    CODE="ZU-WENIG-EXKLUSIVE"
else:
    L0=nur_logits(ARME["LOC"])
    ZUF_LOC=rnd.sample(GEM["LOC"],len(EXK["LOC"]))
    a1,g1,r1,w1=mit_maske(nach_schicht(EXK["LOC"]),ARME["LOC"],N_ARM,SEED+101,
                          ARME["LOC"],L0)
    a2,g2,r2,w2=mit_maske(nach_schicht(ZUF_LOC),ARME["LOC"],N_ARM,SEED+102,
                          ARME["LOC"],L0)
    A_LOC=a1; A_ZUF=a2
    print("  NACHPRUEFUNG  gesperrte Router-Plaetze %d / %d | groesster Restanteil "
          "%.2e / %.2e | Logit bewegt %.4f / %.4f"%(g1,g2,r1,r2,w1,w2))
    assert max(r1,r2)<1e-6,"maskierter Experte behaelt Router-Gewicht"
    assert min(g1,g2)>0,"kein einziger Router-Platz gesperrt - die Maske greift nicht"
    print("")
    kopf(); zeile("LOC ohne Eingriff",BILD["LOC"])
    zeile("%d LOC-exklusive gesperrt"%len(EXK["LOC"]),bild(a1))
    zeile("%d zufaellige gemeinsame"%len(ZUF_LOC),bild(a2))
    print("")
    k1=bild(a1)["kippt"]; k2=bild(a2)["kippt"]
    print("  Kipprate (breite Klassifikation, dasselbe Mass wie in Phase 7-12):")
    vergleich("ohne Eingriff",K_LOC,N_ARM,K_LOC,N_ARM)
    vergleich("LOC-exklusive gesperrt",k1,N_ARM,K_LOC,N_ARM)
    vergleich("zufaellige gemeinsame",k2,N_ARM,K_LOC,N_ARM)
    U_LOC=urteil_arm(K_LOC,N_ARM,k1,N_ARM,k2,N_ARM)
    print("  -> %s"%U_LOC)
    CODE=urteil_haupt(len(EXK["LOC"]),K_LOC/N_ARM,U_LOC)
# ---------------- 4  Spezifitaet und Kreuz ------------------------------------
print(""); print("="*80); print("4  SPEZIFITAET UND KREUZ"); print("="*80)
print("  Sperrt man die exklusiven Experten EINER Sprache, faellt dann auch die")
print("  ANDERE? Ist es ein gemeinsames Buendel, ja. Sind es Sprachidentitaeten,")
print("  nein.")
print("")
KREUZ={}; KREUZ_TEXTE={}
PLAN=[("KO-exklusive -> JA-Arm","KO","JA",ist_ja,BILD["JA"]["japanisch"]),
      ("JA-exklusive -> KO-Arm","JA","KO",ist_ko,BILD["KO"]["koreanisch"]),
      ("JA-exklusive -> LOC-Arm","JA","LOC",lambda t: classify_breit(t) in SWB,K_LOC)]
kopf()
for nm,quelle,ziel,pruef,basis in PLAN:
    if not EXK[quelle]:
        print("  %-30s entfaellt - keine exklusiven Paare"%nm); continue
    Lz=nur_logits(ARME[ziel])
    a,g,r,w=mit_maske(nach_schicht(EXK[quelle]),ARME[ziel],N_ARM,SEED+201+len(KREUZ),
                      ARME[ziel],Lz)
    assert r<1e-6,"maskierter Experte behaelt Router-Gewicht"
    KREUZ_TEXTE[nm]=a
    k=sum(1 for t in a if pruef(t))
    u=urteil_dosis(basis,N_ARM,k,N_ARM)
    zeile(nm,bild(a))
    print("      Zielmass %d/%d gegen %d/%d ohne Eingriff -> %s  (Plaetze %d, Logit %.3f)"
          %(k,N_ARM,basis,N_ARM,u,g,w))
    KREUZ[nm]=dict(quelle=quelle,ziel=ziel,k=k,basis=basis,n=N_ARM,urteil=u,
                   gesperrt=g,logit=w,bild=bild(a))
SPEZ=urteil_spez(P_JAKO,[v["urteil"] for k,v in KREUZ.items() if v["ziel"]!="LOC"])
# ---------------- Wiederherstellung -------------------------------------------
ABW=float(np.abs(nur_logits(ARME["JA"])-L_REF).max())
print(""); print("  WIEDERHERSTELLUNG: alle Haken abgenommen, Logit-Abweichung %.2e"%ABW)
assert ABW<1e-2,"Haken haengen noch - alle Zahlen danach waeren verdorben"
# ---------------- Urteil ------------------------------------------------------
print(""); print("="*80); print("VERDIKT: %s   |   SPEZIFITAET: %s"%(CODE,SPEZ)); print("="*80)
if CODE=="ZU-WENIG-EXKLUSIVE":
    print("  ' local' veraendert die Router-Auswahl am letzten Prompt-Token kaum.")
    print("  Dann faellt die Sprachwahl nicht schon dort, sondern erst spaeter -")
    print("  passend zur Zeitanalyse, die den Entscheidungspunkt bei Zeichen 90-130")
    print("  gefunden hat. Naechster Schritt: das Routing an der ERSTEN DATENZELLE")
    print("  vergleichen, nicht am Prompt-Ende.")
elif CODE=="LOKAL-ROUTING-TRAEGT":
    print("  Die Experten, die nur bei ' local' feuern, aus dem Gemisch zu nehmen")
    print("  senkt die Kipprate - gleich viele gemeinsame nicht. Damit liegt die")
    print("  Sprachwahl beim mehrdeutigen Originalprompt an derselben Stelle wie")
    print("  bei der ausdruecklichen Anweisung: in der Router-Auswahl.")
elif CODE=="NUR-STOERUNG":
    print("  Auch die Zufallskontrolle senkt. Auf dieser Groesse wirkt Beschaedigung")
    print("  und nicht Auswahl - kleiner ansetzen, bis die Kontrolle still ist.")
elif CODE=="KEINE-GRUNDRATE":
    print("  Der LOC-Arm kippt ohne Eingriff kaum (%.0f%%). Ohne Grundrate gibt es"%(100*K_LOC/N_ARM))
    print("  nichts zu senken.")
else:
    print("  Die LOC-exklusiven Experten tragen die Kipprate nicht. Bei der")
    print("  ausdruecklichen Anweisung wirkte dasselbe Verfahren - dann ist die")
    print("  Sprachwahl beim mehrdeutigen Prompt anders verankert als beim")
    print("  eindeutigen, und das ist der interessantere Befund.")
print("")
if SPEZ=="SPRACHSPEZIFISCH":
    print("  Spezifitaet: die exklusiven Mengen fuer Japanisch und Koreanisch")
    print("  ueberlappen nicht ueberzufaellig, und keine wirkt auf die andere")
    print("  Sprache. Das Routing traegt also SPRACHIDENTITAETEN, kein generisches")
    print("  'nicht-englische Schrift'-Buendel.")
elif SPEZ=="GEMEINSAMES-BUENDEL":
    print("  Spezifitaet: die Mengen ueberlappen ueberzufaellig UND wirken")
    print("  ueberkreuz. Dann ist es ein gemeinsames Buendel fuer fremde Schrift,")
    print("  keine einzelne Sprache.")
else:
    print("  Spezifitaet unklar: Ueberlappung und Kreuzwirkung zeigen in")
    print("  verschiedene Richtungen. Die Einzelzahlen stehen oben.")
KARTE_RESULTS=dict(verdict=CODE,spezifitaet=SPEZ,arch_ok=True,prompt_id=ZIEL_ID,
    n_arm=N_ARM,n_experten=NEXP,bild={s:BILD[s] for s in BILD},
    aktiv={s:len(ROUT[s]) for s in ROUT},
    exklusiv={s:[list(q) for q in EXK[s]] for s in EXK},
    gemeinsam={s:len(GEM[s]) for s in GEM},ueberlappung=UEB,
    urteil_loc=U_LOC,abweichung_ende=ABW,kreuz={k:{kk:vv for kk,vv in v.items()} for k,v in KREUZ.items()})
wc_save("antworten_karte",dict(prompt_id=ZIEL_ID,
    **{("arm_"+s):TEXTE[s] for s in TEXTE},
    loc_exklusiv=A_LOC,loc_zufall=A_ZUF,
    **{("kreuz_"+k.replace(" ","_").replace("->","zu")):v for k,v in KREUZ_TEXTE.items()}))
wc_save_all()
print("")
print("(Alle Arme sind gespeichert - beim letzten Lauf fehlten die des geteilten")
print(" Experten, weshalb der Zerfallsverdacht dort offen bleiben musste.)")
````
