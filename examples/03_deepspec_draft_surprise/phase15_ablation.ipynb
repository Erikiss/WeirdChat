# Phase 15 - Ablation der Screen-Menge

**Braucht eine A100**, ~45 min.

Phase 14 hat eine zweite Kandidatenmenge geliefert, und sie teilt mit der ersten **keinen
einzigen Experten**:

| | Herkunft | Status |
|---|---|---|
| **die 42** | Differenz an *einer* Position, Japanisch gegen Englisch | kausal verifiziert: 99→14 %, 68→14 %, 90→29 %, dosisgleich kontrolliert, zwei Durchgänge, gepoolt p < 10⁻⁶ |
| **die 8** | anhaltende Routing-Selektivität über die ganze Antwort, von JA, BR1 und MORSE geteilt und von SR/RU nicht | **nicht geprüft** |

Überlappung: **null**. Diese Zelle prüft die zweite mit dem Apparat, der die erste geprüft
hat.

## Was am ersten Anlauf falsch war

Der erste Durchgang hat auf **Router-Plätze** angeglichen: aus den 42 wurde eine Teilmenge
gezogen, die genauso viele Plätze sperrt wie die 8. Das klingt fair und ist es nicht. Am
Japanisch-Arm sperrte diese Teilmenge **1266 von 17583 Plätzen — 7,2 %** des Eingriffs,
der in Phase 12 gewirkt hat. Alle Bedingungen blieben still, und es gab **keine
Positivkontrolle**. Ohne einen Eingriff, von dem man *weiß*, dass er wirkt, lässt sich
„keine Menge ist ursächlich" nicht von „die Dosis war zu klein" oder „die Messkette klemmt"
unterscheiden. Das Ergebnis war unlesbar.

## Was jetzt anders ist

Angeglichen wird die **Zahl der Experten**. Die sinnvolle Frage lautet nicht „wie viele
Plätze", sondern *trägt ein Satz von acht Experten mehr als ein anderer Satz von acht?*
Und die volle Menge läuft als **Positivkontrolle** mit:

```
ohne | SCREEN (8) | P12-TEIL (8) | ZUFALL (8) | P12-VOLL (42, Positivkontrolle)
```

Die **Platzzahl fällt dabei auseinander**. Sie wird ausgewiesen, und die Wirkung wird
zusätzlich je 100 Plätzen berichtet — sonst wäre ein Unterschied zwischen gleich großen
Mengen heimlich ein Dosisunterschied.

Die Positivkontrolle ist ihrerseits *nicht* zahlangeglichen, 42 gegen 8. Das muss sie auch
nicht: dass die 42 gegen eine dosisgleiche Zufallsmenge tragen, ist in Phase 12 zweimal
gezeigt worden. Hier hat sie nur eine Aufgabe — zu zeigen, dass die Messkette in *diesem*
Lauf überhaupt etwas registriert.

Die Zelle leitet **beide Mengen selbst her**, statt sie eingetragen zu bekommen: den Screen
aus den Abschnitten 1–5 (das ist Phase 14, unverändert), die 42 aus dem Routing an der
Entscheidungsstelle. So bleibt ein Nachlauf in sich stimmig und keine Zahl hängt an einer
Übertragung.

| Verdikt | Lesart |
|---|---|
| `POSITIVKONTROLLE-FEHLT` | die vollen 42 wirken in diesem Lauf nicht. Dann misst nicht die Frage, sondern die Messkette — alles darunter ist unlesbar. Steht deshalb **vorn** |
| `BEIDE-TRAGEN` | zwei unabhängige Wege, obwohl die Mengen disjunkt sind — der Befund aus Phase 12 hängt dann nicht an seiner Menge |
| `NUR-P12-TRAEGT` | der Screen findet **Korrelate** und keine Ursachen — genau die Warnung aus §7 des Quellpapiers |
| `NUR-SCREEN-TRAEGT` | acht von unten gefundene Experten tragen, acht aus Phase 12 gezogene nicht |
| `WIRKUNG-IST-VERTEILT` | die 42 tragen, kein Achtel von ihnen und auch der Screen nicht. Eigener Befund, kein Fehlschlag: die Wirkung sitzt dann nicht in wenigen Knoten |
| `SCREEN-MENGE-ZU-KLEIN` | der Screen liefert unter drei Paaren — dann teilen die konstruierenden Arme keine eigene Signatur |

**Alle Ausgänge sind verwertbar** — bis auf den einen, der die Messkette selbst betrifft.


````python
# Welcher Durchgang? 0 ist der erste Lauf, jede andere Zahl ein eigenstaendiger
# Nachlauf mit anderen Ziehungen, anderer Nullkalibrierung und neu gezogenen
# Kontrollmengen. Die GEPRUEFTEN Mengen werden beide neu hergeleitet.
WIEDERHOLUNG = 0
# === PHASE 15 - ABLATION DER SCREEN-MENGE ===================================
# Phase 14 hat eine zweite Kandidatenmenge geliefert, und sie teilt mit der
# ersten KEINEN EINZIGEN EXPERTEN:
#
#   die 42   Ein-Stichproben-Differenz an der Entscheidungsstelle. Kausal
#            verifiziert: sperrt man sie, faellt Japanisch von 99 auf 14 %,
#            Braille von 68 auf 14, Morse von 90 auf 29 - dosisgleich
#            kontrolliert, in zwei unabhaengigen Durchgaengen, gepoolt
#            p < 1e-6.
#   die 8    anhaltende Routing-Selektivitaet ueber die ganze Antwort, von
#            Japanisch, Braille und Morse geteilt und von Serbisch und
#            Russisch nicht. NICHT kausal geprueft.
#
#   Ueberlappung der beiden: NULL.
#
# WAS AM ERSTEN ANLAUF FALSCH WAR
# Der erste Durchgang hat auf ROUTER-PLAETZE angeglichen: aus den 42 wurde
# eine Teilmenge gezogen, die genauso viele Plaetze sperrt wie die 8. Das
# klingt fair und ist es nicht. Am Japanisch-Arm sperrte diese Teilmenge
# 1266 von 17583 Plaetzen - 7,2 % des Eingriffs, der in Phase 12 gewirkt
# hat. Alle Bedingungen blieben still, und es gab keine Positivkontrolle:
# ohne einen Eingriff, von dem man WEISS, dass er wirkt, laesst sich
# "keine Menge ist ursaechlich" nicht von "die Dosis war zu klein" oder
# "die Messkette klemmt" unterscheiden. Das Ergebnis war unlesbar.
#
# WAS JETZT ANDERS IST
# Angeglichen wird die ZAHL DER EXPERTEN. Die sinnvolle Frage lautet nicht
# "wie viele Plaetze", sondern "traegt ein Satz von acht Experten mehr als
# ein anderer Satz von acht?". Und die volle Menge laeuft als
# POSITIVKONTROLLE mit:
#
#   ohne | SCREEN (8) | P12-TEIL (8) | ZUFALL (8) | P12-VOLL (42, Kontrolle)
#
# Die Platzzahl faellt dabei auseinander. Sie wird ausgewiesen, und die
# Wirkung wird zusaetzlich JE 100 PLAETZEN berichtet - sonst waere ein
# Unterschied zwischen gleich grossen Mengen heimlich ein Dosisunterschied.
#
# Die Zelle leitet BEIDE Mengen selbst her statt sie eingetragen zu bekommen:
# den Screen aus Abschnitt 1 bis 5 (das ist Phase 14, unveraendert), die 42
# aus dem Routing an der Entscheidungsstelle. So bleibt ein Nachlauf in sich
# stimmig, und keine Zahl haengt an einer Uebertragung.
#
#   POSITIVKONTROLLE-FEHLT  die vollen 42 wirken in DIESEM Lauf nicht.
#                       Dann misst nicht die Frage, sondern die Messkette,
#                       und alles darunter ist unlesbar. Steht deshalb VORN
#   BEIDE-TRAGEN        zwei unabhaengige Wege, obwohl die Mengen disjunkt
#                       sind - der Befund aus Phase 12 haengt dann nicht an
#                       seiner Menge
#   NUR-P12-TRAEGT      der Screen findet KORRELATE und keine Ursachen. Genau
#                       die Warnung aus Paragraph 7 des Quellpapiers:
#                       Musterselektivitaet und aufgabenspezifische
#                       Kausalstruktur sind zwei verschiedene Dinge
#   NUR-SCREEN-TRAEGT   acht von unten gefundene Experten tragen, acht aus
#                       Phase 12 gezogene nicht
#   WIRKUNG-IST-VERTEILT die 42 tragen, kein Achtel von ihnen und auch der
#                       Screen nicht. Eigener Befund, kein Fehlschlag: die
#                       Wirkung sitzt dann nicht in wenigen Knoten
#
# Alle Ausgaenge sind verwertbar - deshalb gibt es keinen Misserfolg-Code
# ausser dem einen, der die Messkette selbst betrifft.
import os
os.environ.setdefault("PYTORCH_CUDA_ALLOC_CONF","expandable_segments:True")
import re, math, torch, collections, unicodedata, random
import numpy as np, glob, json, gc, sys, time
gc.collect(); torch.cuda.empty_cache()
try: torch.cuda.synchronize()
except Exception: pass
_free=torch.cuda.mem_get_info()[0]/1e9
if "model" not in globals() and _free<45:
    raise RuntimeError("GPU nicht leer genug (%.1f GB frei, ~45 noetig)."%_free)
if not os.path.isdir("/content/drive/MyDrive"):
    from google.colab import drive; drive.mount("/content/drive")
# ---------------- Protokoll und Abbildungen automatisch nach Drive ----------
# Der PDF-Export von Colab schneidet die Ausgabe unzuverlaessig ab. Deshalb
# schreibt jede Zelle ihr vollstaendiges Protokoll und jede Abbildung selbst
# nach Drive - unabhaengig davon, was der Export spaeter mitnimmt.
import sys, time
WC_RUN=globals().get("WC_RUN","phase15_ablation")
RUN_OUT="/content/drive/MyDrive/WeirdChat_Runs/%s_%s"%(WC_RUN,time.strftime("%Y%m%d-%H%M%S"))
os.makedirs(RUN_OUT,exist_ok=True)
class _WCTee:
    """Schreibt alles doppelt: in die Zelle und nach Drive. Die Datei bleibt
       dabei die ganze Sitzung offen - und genau das kostete einmal ein
       Protokoll. Der Drive-FUSE-Einhang macht eine noch OFFENE Datei nicht
       unbedingt sichtbar; wird die Laufzeit weiterverwendet statt neu
       gestartet, wird das Handle nie geschlossen und protokoll.txt taucht in
       Drive gar nicht auf, waehrend die JSON-Dateien (auf, schreiben, zu) alle
       da sind. Deshalb haelt der Tee zusaetzlich einen Speicherpuffer, den
       wc_save_all am Ende in EINEM geschlossenen Schreibvorgang ablegt."""
    _wc_tee=True
    def __init__(self,p,o): self.o=o; self.f=None; self.puffer=[]; self.retarget(p)
    def retarget(self,p):
        try:
            if self.f: self.f.close()
        except Exception: pass
        self.pfad=p
        if not hasattr(self,"puffer"): self.puffer=[]
        self.puffer=[]
        try: self.f=open(p,"a",encoding="utf-8")
        except Exception: self.f=None
    def write(self,s):
        self.o.write(s)
        try: self.puffer.append(s)
        except Exception: pass
        if self.f:
            try: self.f.write(s); self.f.flush()
            except Exception: pass
        return len(s)
    def flush(self):
        self.o.flush()
        if self.f:
            try: self.f.flush()
            except Exception: pass
    def isatty(self): return False
_wc_log=os.path.join(RUN_OUT,"protokoll.txt")
if getattr(sys.stdout,"_wc_tee",False): sys.stdout.retarget(_wc_log)
else: sys.stdout=_WCTee(_wc_log,sys.stdout)
try:
    import matplotlib.pyplot as _wcplt
    if not getattr(_wcplt,"_wc_patched",False):
        _wc_orig_show=_wcplt.show; _wc_fig=[0]
        def _wc_show(*a,**k):
            for _num in _wcplt.get_fignums():
                _wc_fig[0]+=1
                try:
                    _wcplt.figure(_num).savefig(os.path.join(RUN_OUT,"abb_%02d.png"%_wc_fig[0]),
                                                dpi=150,bbox_inches="tight")
                except Exception: pass
            return _wc_orig_show(*a,**k)
        _wcplt.show=_wc_show; _wcplt._wc_patched=True
except Exception: pass
def wc_save(name,obj):
    """Ergebnisobjekt als JSON neben das Protokoll legen"""
    def _e(o):
        if isinstance(o,np.ndarray): return o.tolist()
        if isinstance(o,(np.integer,)): return int(o)
        if isinstance(o,(np.floating,)): return float(o)
        if isinstance(o,(np.bool_,)): return bool(o)
        return str(o)
    try:
        with open(os.path.join(RUN_OUT,name+".json"),"w",encoding="utf-8") as f:
            json.dump(obj,f,ensure_ascii=False,indent=1,default=_e)
        print("gespeichert: %s.json"%name)
    except Exception as _ex: print("konnte %s nicht speichern: %s"%(name,_ex))
def wc_protokoll_ablegen():
    """Das Protokoll aus dem Speicherpuffer in EINEM geschlossenen Vorgang
       ablegen. Eine offen gehaltene Datei taucht auf dem Drive-Einhang nicht
       zuverlaessig auf; eine geschlossene immer."""
    try:
        _t=sys.stdout
        if getattr(_t,"_wc_tee",False) and getattr(_t,"puffer",None) is not None:
            _p=os.path.join(RUN_OUT,"protokoll_kopie.txt")
            with open(_p,"w",encoding="utf-8") as _f: _f.write("".join(_t.puffer))
            return _p
    except Exception as _ex:
        print("Protokollkopie fehlgeschlagen: %s"%_ex)
    return None
def wc_save_all():
    """alle *_RESULTS aus dem Namensraum sichern - Aufruf am Zellenende"""
    for _k in [k for k in list(globals()) if k.endswith("_RESULTS")]:
        wc_save(_k,globals()[_k])
    _p=wc_protokoll_ablegen()
    if _p: print("Protokollkopie:",os.path.basename(_p))
    print("Lauf-Ordner:",RUN_OUT)
print("Lauf-Ordner (Protokoll + Abbildungen):",RUN_OUT)
if "PROMPTS" not in globals():
    _h=glob.glob("/content/drive/MyDrive/**/weird_transcripts.jsonl",recursive=True)
    assert _h, "weird_transcripts.jsonl nicht gefunden"
    PROMPTS={}
    with open(_h[0],encoding="utf-8") as _f:
        for _line in _f:
            _line=_line.strip()
            if not _line: continue
            _r=json.loads(_line)
            _pid=str(_r["id"]).split("/")[0]
            if _pid not in PROMPTS:
                try: PROMPTS[_pid]=next(t["content"] for t in _r["conversations"] if t["role"]=="user")
                except StopIteration: pass
    PROMPT_IDS=sorted(PROMPTS)
    print("PROMPTS geladen: %d"%len(PROMPTS))
if "model" not in globals() or "tokenizer" not in globals():
    from transformers import AutoModelForCausalLM, AutoTokenizer
    MODEL_ID=globals().get("MODEL_ID","Qwen/Qwen3.6-35B-A3B-FP8")
    print("lade Instruct-Modell:",MODEL_ID,"(einige Minuten)")
    tokenizer=AutoTokenizer.from_pretrained(MODEL_ID)
    model=AutoModelForCausalLM.from_pretrained(MODEL_ID,device_map="auto",torch_dtype="auto")
    model.eval()
    print("geladen | dtype:",next(model.parameters()).dtype)
# ---------------- reine Logik (offline geprueft) ----------------------------
PHRASE="each service's local name"
FRW=[(0x0370,0x03FF),(0x0400,0x052F),(0x0530,0x058F),(0x0590,0x05FF),(0x0600,0x074F),
     (0x0900,0x097F),(0x0E00,0x0E7F),(0x3040,0x30FF),(0x3400,0x9FFF),(0xAC00,0xD7AF),
     (0xF900,0xFAFF)]
JPW=[(0x3040,0x30FF),(0x3400,0x9FFF),(0xF900,0xFAFF)]
FRS=set("le la les une un des est et pour avec dans votre vous voici bonjour du qui que "
        "sur cette ces aux ou par plus il elle nous sont".split())
ENS=set("the is and for with in your you here of to that this are was were has have will "
        "would can it on as at be by".split())
PTES=set("nome nomes servico servicos armazenamento limite limites preco mes gratuito "
         "conta cada para com uma nao mais seu sua nombre servicio servicios "
         "almacenamiento precio cuenta los las del con mas su".split())
DES=set("name dienst dienste speicher speicherplatz grenze preis monat kostenlos konto "
        "jeder fuer mit eine der die das und nicht mehr uebersicht zusammenfassung".split())
def _in(c,bereiche):
    o=ord(c); return any(a<=o<=b for a,b in bereiche)
def _fremd(s):
    return [c for c in s if c.isalpha() and ord(c)>=0x250 and _in(c,FRW)]
def _srun(t,run=3):
    c=0
    for ch in t:
        if ch.isalpha() and ord(ch)>=0x250 and _in(ch,FRW):
            c+=1
            if c>=run: return True
        elif ch.isalpha(): c=0
    return False
def _entakz(s):
    return "".join(c for c in unicodedata.normalize("NFD",s) if not unicodedata.combining(c))
def classify_answer(t):
    if not t.strip(): return "empty"
    al=[c for c in t if c.isalpha()]; fo=_fremd(t)
    if al and len(fo)/len(al)>=0.5: return "takeover"
    if _srun(t): return "gloss"
    w=re.findall(r"[a-zA-ZÀ-ſ']+",t.lower())
    fr=sum(1 for x in w if x in FRS); en=sum(1 for x in w if x in ENS)
    return "latin-switch(fr)" if (fr>=3 and fr>en) else "english"
def classify_breit(t):
    c=classify_answer(t)
    if c!="english": return c
    w=re.findall(r"[a-zA-ZÀ-ſ']+",_entakz(t).lower())
    en=sum(1 for x in w if x in ENS)
    for lab,S in (("pt/es",PTES),("de",DES)):
        n=sum(1 for x in w if x in S)
        if n>=3 and n>en: return "latin-switch(%s)"%lab
    if sum(1 for c2 in t if c2.isalpha() and 0xC0<=ord(c2)<=0x17F)>=3: return "latin-akzent"
    return "english"
SW=("takeover","gloss","latin-switch(fr)")
SWB=SW+("latin-switch(pt/es)","latin-switch(de)","latin-akzent")
def ist_jp(t,mindest=3):
    """Eigener Zaehler fuer die Positivkontrolle. Im JP-Arm ist japanische
       Schrift das ERWUENSCHTE Verhalten - classify_breit wuerde sie
       'takeover' nennen, was hier irrefuehrend waere. Gezaehlt wird ein Lauf
       von mindestens 3 Kana-/Kanji-Zeichen: einzelne Zeichen kommen auch in
       englischen Antworten als Beispiel vor, ein Lauf nicht."""
    c=0
    for ch in t:
        if _in(ch,JPW):
            c+=1
            if c>=mindest: return True
        elif ch.isalpha(): c=0
    return False
def wiederholt(t,fenster=12,mal=4):
    """Zerfallsmerkmal: dieselbe Zeichenfolge viermal. Bei starker Beschaedigung
       faellt ein Modell in Schleifen, lange bevor es verstummt."""
    if len(t)<fenster*mal: return False
    z=collections.Counter(t[i:i+fenster] for i in range(len(t)-fenster+1))
    return max(z.values())>=mal
def zerfall(texte):
    """Was sagt die Antwortform ueber den Schaden - unabhaengig von der Sprache"""
    if not texte: return dict(leer=0.0,laenge=0.0,schleife=0.0)
    return dict(leer=sum(1 for t in texte if not t.strip())/len(texte),
                laenge=sum(len(t) for t in texte)/len(texte),
                schleife=sum(1 for t in texte if wiederholt(t))/len(texte))
def wilson(k,n,z=1.96):
    if n==0: return (0.,0.,0.)
    p=k/n; d=1+z*z/n; c=p+z*z/(2*n); h=z*math.sqrt(p*(1-p)/n+z*z/(4*n*n))
    return p,(c-h)/d,(c+h)/d
def fisher2x2(a,b,c,d,einseitig=False):
    from math import lgamma,exp
    lf=lambda n: lgamma(n+1); n=a+b+c+d
    def pr(x):
        y=a+b-x; z=a+c-x; w=n-x-y-z
        if min(y,z,w)<0: return 0.0
        return exp(lf(a+b)+lf(c+d)+lf(a+c)+lf(b+d)-lf(n)-lf(x)-lf(y)-lf(z)-lf(w))
    hi=min(a+b,a+c)
    if einseitig: return min(1.0,sum(pr(x) for x in range(a,hi+1)))
    p0=pr(a)
    return min(1.0,sum(pr(x) for x in range(0,hi+1) if pr(x)<=p0*(1+1e-9)))
def phrase_mit(w):
    return "each service's name" if not w else "each service's %s name"%w
def setze_arm(text,ersatz):
    if PHRASE not in text: return text,False
    return text.replace(PHRASE,ersatz),True
def massstab(*vektoren):
    """EIN globaler, robuster Massstab fuer alle Einheiten. Ersetzt den
       Nenner je Einheit aus v2, der auf 1e-8 fallen konnte und damit
       Trennwerte von 1e7 erzeugt hat. Median statt Mittelwert, weil die
       Zwischenschicht duennbesetzt ist und wenige grosse Werte den Mittelwert
       tragen wuerden. Nullen zaehlen NICHT mit: bei 70% strukturellen Nullen
       waere der Median sonst selbst null."""
    v=np.abs(np.concatenate([np.asarray(x,dtype=np.float64).ravel() for x in vektoren]))
    v=v[v>0]
    if v.size==0: return 1.0
    m=float(np.median(v))
    return m if m>0 else 1.0
def trennung_zwei(xa,xb,s):
    """Differenz zweier Zustaende in Einheiten EINES globalen Massstabs.
       Positiv = im ersten Zustand hoeher. Beschraenkt und vergleichbar."""
    return (np.asarray(xa,dtype=np.float64)-np.asarray(xb,dtype=np.float64))/float(s)
def waehle_einheiten(d,k):
    """die k Einheiten mit der groessten POSITIVEN Trennung - Richtung ist
       vorregistriert: im JP-Zustand hoeher, Ablation muss die JP-Rate senken"""
    return list(np.argsort(-np.asarray(d))[:k])
KANA=[(0x3040,0x30FF)]
HANGUL=[(0xAC00,0xD7AF),(0x1100,0x11FF),(0x3130,0x318F)]
HANB=[(0x3400,0x9FFF),(0xF900,0xFAFF)]
# VOLLSTAENDIGE Schrifttafel. Der erste Lauf dieser Zelle kannte nur Kana,
# Hangul und Han - alles andere fiel auf 'latein:englisch' durch. Zwei der
# zwoelf geernteten Praefixe waren russisch, wurden als "noch englisch"
# durchgelassen und ihre 64 russischen Fortsetzungen als Englisch gezaehlt:
#     '| Облако (Local Name) | Лимит хранилища | ...'  gemessen 0/32, echt 32/32
# Damit landeten die zwei extremsten HOHEN Praefixe in der NIEDRIGEN Gruppe.
# classify_breit konnte das laengst (FRW deckt 0x0400-0x052F); beim Umbau auf
# Zielsprachen ist es verlorengegangen.
SCHRIFTEN=[("kyrillisch",[(0x0400,0x052F)]),("griechisch",[(0x0370,0x03FF)]),
           ("armenisch",[(0x0530,0x058F)]),("hebraeisch",[(0x0590,0x05FF)]),
           ("arabisch",[(0x0600,0x074F)]),("devanagari",[(0x0900,0x097F)]),
           ("thai",[(0x0E00,0x0E7F)])]
WORTE={"pt":"nome nomes servico servicos armazenamento limite limites preco mes "
            "gratuito conta cada para com uma nao mais seu sua",
       "es":"nombre servicio servicios almacenamiento precio cuenta los las del "
            "con mas su gratuito",
       "fr":"le la les une un des est et pour avec dans votre vous voici du qui "
            "que sur cette ces aux ou par plus nom stockage prix tarif",
       "de":"dienst dienste speicher speicherplatz laufwerk grenze grenzen preis "
            "monat kostenlos konto jeder fuer mit eine der die das und nicht mehr "
            "zusammenfassung",
       "it":"nome servizio servizi archiviazione prezzo mese gratuito conto per "
            "con una non piu"}
WORTE={k:set(v.split()) for k,v in WORTE.items()}
def _z(t,bereiche): return sum(1 for c in t if _in(c,bereiche))
def latein_art(t):
    """welche lateinschriftliche Sprache - oder Englisch. 'akzent' faengt
       Sprachen ausserhalb der Wortlisten (im letzten Lauf kam so Lettisch)."""
    w=re.findall(r"[a-zA-Z']+",_entakz(t).lower())
    tr=sorted(((n,sum(1 for x in w if x in S)) for n,S in WORTE.items()),key=lambda x:-x[1])
    if tr[0][1]>=3: return tr[0][0]
    if sum(1 for c in t if c.isalpha() and 0xC0<=ord(c)<=0x17F)>=3: return "akzent"
    return "englisch"
def schrift(t):
    """Kana beweist Japanisch, Hangul Koreanisch, Han allein nur CJK. Die
       ZIELSPRACHE wird immer mitgezaehlt - zweimal hat eine blosse Kippzahl
       den Effekt verschluckt."""
    if not t.strip(): return "leer"
    if _z(t,KANA)>=2: return "japanisch"
    if _z(t,HANGUL)>=2: return "koreanisch"
    if _z(t,HANB)>=3: return "chinesisch"
    for nm,ber in SCHRIFTEN:
        if _z(t,ber)>=3: return nm
    if _z(t,HANB)>0: return "han-einzeln"
    return "latein:"+latein_art(t)
def kippt(t):
    """streng: fremde Schrift ODER eine erkannte lateinische Fremdsprache.
       Nicht classify_breit - dessen Kippzahl hat im letzten Lauf einen
       Einbruch bei CJK gegen einen Anstieg bei Latein aufgerechnet."""
    a=schrift(t)
    return a not in ("latein:englisch","leer")
def sauber(p):
    """Ein Praefix taugt nur, wenn er selbst NOCH ENGLISCH ist - sonst misst
       man die eigene Vorgabe statt der Entscheidung."""
    return bool(p.strip()) and schrift(p)=="latein:englisch"
def ernte_stellen(texte,laenge,hoechstens):
    """verschiedene Fortsetzungen bis zur Entscheidungsstelle. Anders als
       frueher wird NICHT nach Haeufigkeit gruppiert: bei 100 Zeichen ist fast
       jede Ziehung einzigartig. Gebraucht werden verschiedene, noch saubere
       Praefixe - ihre Rate wird einzeln durch Erzwingen gemessen."""
    aus=[]; gesehen=set()
    for t in texte:
        if len(t)<laenge: continue
        p=t[:laenge]
        if p in gesehen or not sauber(p): continue
        gesehen.add(p); aus.append(p)
        if len(aus)>=hoechstens: break
    return aus
def trenn_paare(routings,hoch):
    """je (Schicht,Experte): Anteil der HOHEN Praefixe, in denen es feuert,
       minus Anteil der NIEDRIGEN. +1 heisst 'in allen hohen, in keinem
       niedrigen'. Kein Nenner je Einheit, also nichts, was auf null fallen
       und eine Trennung herbeizaubern kann."""
    hi=[r for r,h in zip(routings,hoch) if h]
    lo=[r for r,h in zip(routings,hoch) if not h]
    if not hi or not lo: return []
    alle=sorted(set().union(*[set(r) for r in routings]))
    aus=[]
    for q in alle:
        a=sum(1 for r in hi if q in r)/len(hi)
        b=sum(1 for r in lo if q in r)/len(lo)
        aus.append((q,a-b))
    aus.sort(key=lambda x:-x[1])
    return aus
def perfekte(paare,schwelle=1.0):
    return [q for q,s in paare if s>=schwelle-1e-9]
def trenner_null(routings,hoch,rnd,perm=2000,schwelle=1.0):
    """Bei acht Praefixen findet man perfekte Trenner auch rein zufaellig.
       Etiketten vertauschen, Trenner neu zaehlen - ist die beobachtete Zahl
       nicht groesser als die zufaellige, gibt es nichts zu sperren."""
    beob=len(perfekte(trenn_paare(routings,hoch),schwelle))
    h=list(hoch); t=0; werte=[]
    for _ in range(perm):
        rnd.shuffle(h)
        n=len(perfekte(trenn_paare(routings,h),schwelle))
        werte.append(n)
        if n>=beob: t+=1
    return beob,(t+1)/(perm+1.0),float(np.mean(werte)) if werte else 0.0
def null_untergrenze(n,k_hoch,schwelle=1.0):
    """Kleinstmoeglicher p-Wert des Etikettentauschs, exakt gerechnet: die
       Wahrscheinlichkeit, dass ein IDEALER Trenner - in allen k_hoch hohen
       Praefixen, in keinem niedrigen - die Schwelle auch unter zufaelliger
       Aufteilung noch erreicht, mal zwei fuer sein Gegenstueck.

       Zwei Laeufe sind an dieser Zahl gescheitert. Erst mit acht Praefixen und
       Schwelle 1.0: perfekte Trenner, p=0.060, weil nur die beobachtete
       Aufteilung und ihr Komplement die volle Zahl liefern koennen. Dann mit
       zwoelf und einer gelockerten Schwelle - 4/6 ist zwar erreichbar, aber
       die Untergrenze steigt dort auf 0.080, der Test kann 0.05 nicht mehr
       unterschreiten. Die brauchbaren Felder:

           K       1.00   0.83   0.75   0.67   0.50
           12     0.002  0.002  0.002  0.080  0.080
           16     0.000  0.000  0.010  0.010  0.132
           20     0.000  0.000  0.001  0.001  0.023

       Bei zwoelf Praefixen ist 5/6 die unterste brauchbare Schwelle; wer 4/6
       zulassen will, braucht sechzehn."""
    kn=max(n-k_hoch,1); su=0
    for h in range(0,k_hoch+1):
        if h/max(k_hoch,1)-(k_hoch-h)/kn>=schwelle-1e-9:
            su+=math.comb(k_hoch,h)*math.comb(kn,min(k_hoch-h,kn))
    return min(1.0,2.0*su/math.comb(n,k_hoch))
def stufen_vom_raster(n_hoch,n_niedrig,wieviel=4,mindestens=0.5):
    """Die Stufen des Trennwertbildes muessen ERREICHBARE Werte sein. Bei sechs
       gegen sechs sind nur Vielfache von 1/6 moeglich; eine Stufe 0.67 liegt
       knapp ueber 4/6=0.6667 und zaehlt dort null, waehrend %.2f sie als
       '0.67' druckt. Genau so sind im zweiten Lauf vier Trenner verschwunden."""
    w=sorted({a/max(n_hoch,1)-b/max(n_niedrig,1)
              for a in range(n_hoch+1) for b in range(n_niedrig+1)},reverse=True)
    return [x for x in w if x>=mindestens-1e-9][:wieviel]
def immer_aktiv(routings):
    """in ALLEN Praefixen aktiv"""
    if not routings: return []
    g=set(routings[0])
    for r in routings[1:]: g&=set(r)
    return sorted(g)
def haeufig_aktiv(routings,mindestanteil=0.5):
    """Quelle der Zufallskontrolle. NICHT immer_aktiv: im ersten Lauf war die
       Schnittmenge ueber zwoelf Praefixe LEER (1755 verschiedene Paare aus
       3840 Plaetzen), und eine Kontrolle aus der leeren Menge sperrt nichts.
       Gebraucht wird dieselbe ART von Paar - eines, das an dieser Stelle
       ueberhaupt regelmaessig laeuft."""
    if not routings: return []
    z=collections.Counter()
    for r in routings: z.update(set(r))
    n=len(routings)
    return sorted(q for q,k in z.items() if k/n>=mindestanteil)
def trennwert_bild(paare,stufen):
    """Wie viele Paare erreichen welche Trennung. Ohne diese Zeile ist ein
       Nullbefund nicht deutbar: der erste Lauf meldete null perfekte Trenner,
       und es liess sich nicht sagen, ob etwas knapp danebenlag.

       Die Stufen liegen auf dem RASTER. Bei sechs hohen gegen sechs niedrige
       Praefixen sind nur Vielfache von 1/6 erreichbar; eine Schwelle von 0.67
       liegt knapp ueber 4/6 = 0.6667 und schliesst genau die Faelle aus, die
       sie treffen soll. Der zweite Lauf ist daran haengengeblieben: vier Paare
       wurden als '0.67' gedruckt und die Stufe 0.67 zaehlte null."""
    return [(s,sum(1 for _,w in paare if w>=s-1e-9)) for s in stufen]
def raster(n_hoch,n_niedrig):
    """Schrittweite der erreichbaren Trennwerte - gehoert ins Protokoll, damit
       niemand wieder eine Schwelle zwischen zwei Rasterpunkte legt"""
    return max(1.0/max(n_hoch,1),1.0/max(n_niedrig,1))
def nach_schicht(paare):
    d=collections.defaultdict(set)
    for l,e in paare: d[int(l)].add(int(e))
    return dict(d)
def exklusiv(za,zb): return sorted(set(za)-set(zb))
def urteil_dosis(k_bas,n_bas,k_abl,n_abl,alpha=0.05):
    if n_bas==0 or n_abl==0: return "still"
    p=fisher2x2(k_abl,n_abl-k_abl,k_bas,n_bas-k_bas)
    if p>=alpha: return "still"
    return "senkt" if k_abl/n_abl<k_bas/n_bas else "hebt"
def urteil_arm(k_bas,n_bas,k_aus,n_aus,k_zuf,n_zuf,alpha=0.05):
    a=urteil_dosis(k_bas,n_bas,k_aus,n_aus,alpha)=="senkt"
    z=urteil_dosis(k_bas,n_bas,k_zuf,n_zuf,alpha)=="senkt"
    if a and not z: return "TRAEGT"
    if a and z:     return "NUR-STOERUNG"
    if z and not a: return "WIDERSPRUECHLICH"
    return "BLIND"
def urteil_stelle(spreizung,n_trenner,p_trenner,u_pos,u_hoch,untergrenze=0.0,
                  mindest_spreizung=0.3,mindest_trenner=3,alpha=0.05):
    """Reihenfolge ist Absicht. Erst die Positivkontrolle: versagt sie, ist das
       Messfeld unempfindlich und alles Weitere waere ein Befund ueber den
       Aufbau. Dann die Spreizung: liegen alle Praefixe bei derselben Rate, ist
       die Entscheidung an dieser Stelle noch nicht gefallen. Dann die
       Aufloesung des Nulltests - kann er die Schwelle gar nicht erreichen,
       ist ein p daraus bedeutungslos. Dann der Nulltest selbst. Erst danach
       die eigentliche Frage."""
    if u_pos!="senkt": return "MESSFELD-UNEMPFINDLICH"
    if spreizung<mindest_spreizung: return "ZU-WENIG-SPREIZUNG"
    if n_trenner<mindest_trenner: return "ZU-WENIG-TRENNER"
    if untergrenze>=alpha: return "AUFLOESUNG-ZU-GROB"
    if p_trenner>=alpha: return "TRENNER-ZUFAELLIG"
    return {"TRAEGT":"ENTSCHEIDUNGSSTELLE-TRAEGT","NUR-STOERUNG":"NUR-STOERUNG",
            "WIDERSPRUECHLICH":"WIDERSPRUECHLICH","BLIND":"BLIND"}[u_hoch]
BRAILLE=[(0x2800,0x28FF)]
KYR=[(0x0400,0x052F)]
# Hepburn-Umschrift der Dienste aus dem Prompt. Romaji ist an der SCHRIFT nicht
# erkennbar - es steht in lateinischen Buchstaben. Genau das macht es zur
# Gegenzelle von Braille und zwingt zu einem Wortdetektor.
ROMAJI_NAMEN=("guguru gūguru gu-guru doraibu doraibo doroppubokkusu doroppu "
              "bokkusu wandoraibu wan aikuraudo aikuraudo megā mega shinku "
              "pikuraudo dorobbokusu").split()
ROMAJI_WORTE=("sābisu sabisu hozon youryou yōryō yoryo ryōkin ryoukin muryō "
              "muryou musho gigabaito tsuki namae maitsuki gessha").split()
ROMAJI=set(ROMAJI_NAMEN)|set(ROMAJI_WORTE)
MAKRON="āīūēōĀĪŪĒŌ"
def ist_braille(t,mindest=3):
    return _z(t,BRAILLE)>=mindest
def ist_kyrillisch(t,mindest=3):
    return _z(t,KYR)>=mindest
def ist_morse(t):
    """Morse steht in ASCII - kein Unicode-Block hilft. Gesucht wird eine Folge
       aus Punkt, Strich, Schraegstrich und Leerzeichen von mindestens acht
       Zeichen, die BEIDES enthaelt. Eine Markdown-Trennzeile ':---' faellt
       nicht darunter, weil ihr die Punkte fehlen."""
    for m in re.finditer(r"[.\-/ ]{8,}",t):
        s=m.group(0)
        if s.count("-")>=3 and s.count(".")>=3: return True
    return False
def ist_romaji(t):
    """Japanisch in lateinischer Schrift. Zwei Wege, weil keiner allein reicht:
       transliterierte Dienstnamen, oder Makronvokale - die gibt es im
       Englischen nicht und in Hepburn staendig. Antworten mit Kana oder
       Hangul zaehlen NICHT als Romaji, sonst misst man den Japanisch-Arm."""
    if _z(t,KANA)>=2 or _z(t,HANGUL)>=2: return False
    w=set(re.findall(r"[a-zāīūēōâîûêô']+",t.lower()))
    if len(w&ROMAJI)>=2: return True
    return sum(1 for c in t if c in MAKRON)>=3
def ist_kana(t):
    return _z(t,KANA)>=2
# (Schluessel, Phrase, Zielmass, was der Arm im Plan besetzt)
ARME=[("NEU","each service's name","englisch","Bezugsarm"),
      ("LOC","each service's local name","kippt","der mehrdeutige Originalarm"),
      ("JA","each service's Japanese name","kana","Sprache UND Schrift"),
      ("ROMAJI","each service's Japanese name written in romaji","romaji",
       "Sprache OHNE Schriftwechsel"),
      ("SR","each service's Serbian name","kyrillisch","Sprache, Schrift offen"),
      ("BR1","each service's Braille name","braille","SCHRIFT OHNE SPRACHE"),
      ("BR2","each service's name written in Braille","braille","dasselbe, andere Formulierung"),
      ("MORSE","each service's name written in Morse code","morse","Umschrift ohne Schriftwechsel")]
def zielmass(name):
    return {"kana":ist_kana,"romaji":ist_romaji,"kyrillisch":ist_kyrillisch,
            "braille":ist_braille,"morse":ist_morse,
            "kippt":kippt,"englisch":lambda t: not kippt(t)}[name]
def lebt(k,n,mindest=0.25):
    """Ein Arm taugt nur als Messfeld, wenn das Modell die Anweisung ueberhaupt
       ausfuehrt. Der erste Minimalpaar-Lauf ist an einem toten Feld gescheitert
       (5.5% statt 84.4%), und der Piloten-Teil dieser Zelle ist genau dafuer
       da: erst schauen, ob der Arm lebt, dann darauf bauen."""
    return (k/max(n,1))>=mindest
def jaccard(A,B):
    A=set(A); B=set(B)
    return len(A&B)/len(A|B) if (A or B) else 0.0
def exklusiv(za,zb): return sorted(set(za)-set(zb))
def nach_schicht(paare):
    d=collections.defaultdict(set)
    for l,e in paare: d[int(l)].add(int(e))
    return dict(d)
def ueberlappungs_null(A,B,ref,n_experten,rnd,perm=2000):
    """Wie gross waere die Ueberlappung zweier exklusiver Mengen zufaellig? Je
       Schicht gleich viele Experten neu ziehen, aber nur aus denen, die der
       Bezugsarm dort nicht benutzt."""
    RA=nach_schicht(A); RB=nach_schicht(B); RR=nach_schicht(ref)
    beob=len(set(A)&set(B)); treffer=0; werte=[]
    for _ in range(perm):
        n=0
        for l in set(RA)|set(RB):
            frei=[e for e in range(n_experten) if e not in RR.get(l,set())]
            a=rnd.sample(frei,min(len(RA.get(l,())),len(frei)))
            b=rnd.sample(frei,min(len(RB.get(l,())),len(frei)))
            n+=len(set(a)&set(b))
        werte.append(n)
        if n>=beob: treffer+=1
    return beob,(treffer+1)/(perm+1.0),float(np.mean(werte))
def urteil_dosis(k_bas,n_bas,k_abl,n_abl,alpha=0.05):
    if n_bas==0 or n_abl==0: return "still"
    p=fisher2x2(k_abl,n_abl-k_abl,k_bas,n_bas-k_bas)
    if p>=alpha: return "still"
    return "senkt" if k_abl/n_abl<k_bas/n_bas else "hebt"
def urteil_schrift(lebt_ja,lebt_romaji,u_ja,u_romaji):
    """Die Frage, um die es geht: kodiert die JA-exklusive Expertenmenge die
       SPRACHE oder die SCHRIFT? Romaji ist japanische Sprache in lateinischer
       Schrift und trennt das als einziger Arm.

         stirbt Romaji mit  -> die Menge haengt an der Sprache
         ueberlebt Romaji   -> sie haengt an der Schrift

       Davor zwei Sperren: ohne lebendigen Japanisch-Arm gibt es keine
       Eichmarke, und ohne wirksame Maske dort ist das Feld unempfindlich."""
    if not lebt_ja: return "EICHMARKE-FEHLT"
    if u_ja!="senkt": return "MESSFELD-UNEMPFINDLICH"
    if not lebt_romaji: return "ROMAJI-TOT"
    return "MENGE-KODIERT-SPRACHE" if u_romaji=="senkt" else "MENGE-KODIERT-SCHRIFT"
def dosisgleich(ziel_plaetze,zaehler,verboten,rnd,toleranz=0.10,versuche=400):
    """Zufallsmenge, deren ROUTER-PLAETZE die Zielzahl treffen - nicht deren
       Paarzahl. Genau daran ist die erste Kontrolle gescheitert: 42 Paare
       gegen 42 Paare, aber 264 gesperrte Plaetze gegen 557. Die Kontrolle war
       damit der HAERTERE Eingriff und trotzdem schwaecher, was die alte
       Urteilsregel als 'Zerbrechlichkeit' verbucht hat.

       Gierig aufgefuellt, viele Anlaeufe, der beste Treffer gewinnt. Die
       Paarzahl faellt dabei kleiner aus als 42, weil gewoehnliche Experten
       haeufiger laufen als die exklusiven - und genau das ist der Punkt."""
    kand=[q for q in zaehler if q not in verboten and zaehler[q]>0]
    if not kand or ziel_plaetze<=0: return [],0
    unten=ziel_plaetze*(1.0-toleranz); oben=ziel_plaetze*(1.0+toleranz)
    best=None
    for _ in range(versuche):
        rnd.shuffle(kand); menge=[]; summe=0
        for q in kand:
            if summe>=unten: break
            if summe+zaehler[q]<=oben: menge.append(q); summe+=zaehler[q]
        if summe>=unten and (best is None or
                             abs(summe-ziel_plaetze)<abs(best[1]-ziel_plaetze)):
            best=(sorted(menge),summe)
    return best if best else ([],0)
def wirkung_je_platz(k_bas,k_maske,n,plaetze):
    """Prozentpunkte Wirkung je 100 gesperrten Router-Plaetzen. Ohne diese
       Groesse laesst sich ein Eingriff nicht mit einem staerkeren vergleichen."""
    if plaetze<=0 or n<=0: return 0.0
    return 100.0*(100.0*(k_bas-k_maske)/float(n))/float(plaetze)
def urteil_arm_dosis(k_bas,n,k_ja,p_ja,k_zu,p_zu,pl_ja,pl_zu,alpha=0.05,faktor=2.0):
    """NUR-STOERUNG erst, wenn die Kontrolle JE PLATZ aehnlich stark wirkt.
       Die alte Regel fragte nur 'beide signifikant?' und nannte es deshalb
       Zerbrechlichkeit, als die JA-Maske Braille mit 264 Plaetzen um 48 Punkte
       senkte und die Zufallsmaske mit 557 um 20."""
    a=(k_ja<k_bas) and p_ja<alpha
    z=(k_zu<k_bas) and p_zu<alpha
    if not a: return "WIDERSPRUECHLICH" if z else "STILL"
    if not z: return "TRAEGT"
    ej=wirkung_je_platz(k_bas,k_ja,n,pl_ja); ez=wirkung_je_platz(k_bas,k_zu,n,pl_zu)
    return "TRAEGT-UEBERWIEGEND" if ej>=faktor*max(ez,1e-9) else "NUR-STOERUNG"
TRAEGT_ALLE=("TRAEGT","TRAEGT-UEBERWIEGEND")
# ---------------- Routing-Selektivitaet: Schaltkreis je Arm -------------------
# Uebertragung von Schritt 2 des Spektral-Papiers (task-pattern screen) auf
# MoE-Experten. Dort: Attention-Selektivitaet eines Kopfes gegen ein
# Uniform-Baseline. Hier: FEUERRATE eines Experten auf einem Arm gegen seine
# Feuerrate auf dem Bezugsarm.
#
# Warum das besser ist als die Mengendifferenz aus Phase 12: die 42 Paare sind
# eine harte Differenz an EINER Position eines EINZIGEN Prompts. Ein Screen mit
# Schwelle und Nullkalibrierung nutzt viele Beispiele, gibt je Arm einen
# eigenen Schaltkreis - und erlaubt damit die Frage, was Braille, Morse und
# Japanisch TEILEN, ohne von Japanisch auszugehen.
ARME=[("NEU","each service's name","Bezugsarm - der Nenner"),
      ("JA","each service's Japanese name","Kana"),
      ("SR","each service's Serbian name","kyrillisch, lexikalisch"),
      ("RU","each service's Russian name","kyrillisch, hohe Grundrate"),
      ("BR1","each service's Braille name","konstruiert, fremde Schrift"),
      ("MORSE","each service's name written in Morse code","konstruiert, ASCII")]
BEZUG="NEU"
KONSTRUIERT_S=("JA","BR1","MORSE")   # sollen eine gemeinsame Menge haben
LEXIKALISCH_S=("SR","RU")            # sollen sie NICHT haben
OBERFLAECHE_PAAR=("SR","RU")
# Zwei Arme, die sich in der OBERFLAECHENFORM unterscheiden (serbischer gegen
# russischen Text) aber NICHT in der Eigenschaft, um die es geht (beide
# lexikalisch verfuegbar). Ihr gegenseitiges Verhaeltnis ist der Rauschboden
# fuer 'verschiedener Text erzeugt verschiedenes Routing' - und genau der
# fehlte im ersten Anlauf. Dort war nur gegen Stichprobenrauschen kalibriert,
# und der Screen lieferte 14 bis 25 % des Modells je Arm.
MAX_ANTEIL=0.05                      # ein Kern ueber 5% des Modells ist keiner
def raten(zaehler_liste,paare=None):
    """Aus einer Liste von Zaehlern je Beispiel: Gesamttreffer je Paar und die
       Gesamtzahl der Router-Plaetze. Liefert (treffer, plaetze)."""
    treffer=collections.Counter(); plaetze=0
    for z in zaehler_liste:
        treffer.update(z); plaetze+=sum(z.values())
    return treffer,plaetze
def selektivitaet(k_arm,n_arm,k_ref,n_ref,glaettung=0.5):
    """Verhaeltnis der Feuerraten, additiv geglaettet. Ohne Glaettung waere ein
       Experte, der im Bezugsarm nie feuert, unendlich selektiv - und das sind
       genau die seltenen, bei denen das Verhaeltnis am wenigsten aussagt."""
    if n_arm<=0 or n_ref<=0: return 0.0
    ra=(k_arm+glaettung)/float(n_arm); rr=(k_ref+glaettung)/float(n_ref)
    return ra/rr if rr>0 else 0.0
def schicht_von(k,basis=2.0):
    """Haeufigkeitsschicht eines Experten, log-gestuft nach seiner Trefferzahl
       im Bezugsarm. Ohne diese Schichtung misst der Screen vor allem
       SELTENHEIT: ein Experte mit 2 Treffern hat ein viel verrauschteres
       Verhaeltnis als einer mit 200, und die Seltenen wuerden jede Rangliste
       anfuehren."""
    return int(math.floor(math.log(k+1.0,basis)))
def null_schwellen(bezug_zaehler,rnd,perm=200,quantil=0.99,glaettung=0.5):
    """Nullverteilung des Selektivitaetsverhaeltnisses aus dem BEZUGSARM
       gegen sich selbst: die Beispiele werden wiederholt zufaellig halbiert
       und die Haelften gegeneinander gerechnet. Was dabei herauskommt, ist
       reines Stichprobenrauschen - je Haeufigkeitsschicht getrennt.

       Das ist die Uebertragung der Null-Selektivitaetskalibrierung des Papiers
       (500 Ziehungen gegen eine zufaellige Zielposition) auf einen Aufbau, in
       dem es keine Zielposition gibt, sondern zwei Stichproben."""
    if len(bezug_zaehler)<4: return {},{}
    ges,_=raten(bezug_zaehler)
    schicht={q:schicht_von(ges[q]) for q in ges}
    proben=collections.defaultdict(list)
    idx=list(range(len(bezug_zaehler)))
    for _ in range(perm):
        rnd.shuffle(idx); h=len(idx)//2
        a=[bezug_zaehler[i] for i in idx[:h]]; b=[bezug_zaehler[i] for i in idx[h:]]
        ka,na=raten(a); kb,nb=raten(b)
        for q in ges:
            proben[schicht[q]].append(selektivitaet(ka[q],na,kb[q],nb,glaettung))
    schwellen={}
    for s,v in proben.items():
        v=sorted(v); schwellen[s]=v[min(len(v)-1,int(quantil*len(v)))]
    return schwellen,schicht
def oberflaechen_schwellen(a_zaehler,b_zaehler,schicht,quantil=0.99,glaettung=0.5):
    """Rauschboden fuer den blossen OBERFLAECHENUNTERSCHIED. Zwei Arme, die
       verschiedenen Text erzeugen und sich in der gesuchten Eigenschaft nicht
       unterscheiden, werden gegeneinander gerechnet; je Haeufigkeitsschicht
       wird das Quantil ueber die Experten genommen.

       Symmetrisch, weil die Richtung willkuerlich ist: gewertet wird
       max(s, 1/s). Das Ergebnis ist die Antwort auf 'wie stark unterscheidet
       sich das Routing zweier gleichartiger Arme allein deshalb, weil der
       Text anders aussieht'. Wer darunter liegt, ist kein Befund."""
    ka,na=raten(a_zaehler); kb,nb=raten(b_zaehler)
    proben=collections.defaultdict(list)
    for q in set(ka)|set(kb):
        s=selektivitaet(ka[q],na,kb[q],nb,glaettung)
        if s<=0: continue
        proben[schicht.get(q,0)].append(max(s,1.0/s))
    aus={}
    for st,v in proben.items():
        v=sorted(v); aus[st]=v[min(len(v)-1,int(quantil*len(v)))]
    return aus
def vereinige_schwellen(*karten):
    """Je Schicht die STRENGSTE der Schwellen. Stichprobenrauschen und
       Oberflaechenunterschied sind zwei verschiedene Stoerquellen; eine Menge
       muss beide ueberschreiten."""
    aus={}
    for k in karten:
        for st,v in k.items(): aus[st]=max(aus.get(st,0.0),v)
    return aus
def screen(arm_zaehler,bezug_zaehler,schwellen,schicht,mindest=2.0,glaettung=0.5,
           mindest_treffer=5):
    """Alle Paare, deren Selektivitaet UEBER der Nullschwelle ihrer
       Haeufigkeitsschicht liegt, ueber einer harten Untergrenze - und die im
       Arm ueberhaupt oft genug feuern.

       mindest_treffer ist keine Feinheit. Bei den seltensten Experten kann
       schon der Sprung von null auf zwei Treffer das geglaettete Verhaeltnis
       auf die Schwelle heben; in der Miniatur rutschte dadurch in JEDEN der
       fuenf Arme dasselbe Paar, obwohl es nur in zweien einen Effekt hatte.
       Zwei gegen null ist kein Schaltkreis.

       Liefert (Menge, {paar: selektivitaet})."""
    ka,na=raten(arm_zaehler); kb,nb=raten(bezug_zaehler)
    sel={}; menge=[]
    for q in set(ka)|set(kb):
        s=selektivitaet(ka[q],na,kb[q],nb,glaettung)
        sel[q]=s
        if ka[q]<mindest_treffer: continue
        gr=max(schwellen.get(schicht.get(q,0),0.0),mindest)
        if s>=gr: menge.append(q)
    return sorted(menge),sel
def schnitt(mengen):
    if not mengen: return []
    aus=set(mengen[0])
    for m in mengen[1:]: aus&=set(m)
    return sorted(aus)
def ohne(menge,andere):
    return sorted(set(menge)-set().union(*[set(a) for a in andere]) if andere else set(menge))
def jaccard_s(a,b):
    a=set(a); b=set(b)
    return len(a&b)/len(a|b) if (a or b) else 0.0
def urteil_screen(mengen,kern,kern_p,rest,n_paare=10240,alpha=0.05,mindest_kern=3,
                  max_anteil=MAX_ANTEIL):
    """Reihenfolge ist Absicht. Ohne Arm-Schaltkreise gibt es nichts zu
       schneiden; ein Screen, der ein Viertel des Modells auswaehlt, misst
       nicht mehr das Gesuchte; ein Kern auf Zufallsniveau ist kein Befund;
       und ein Kern, der vollstaendig in der lexikalischen Menge steckt,
       trennt nicht - und ein Kern, von dem nach Abzug des Lexikalischen nur
       ein oder zwei Paare uebrigbleiben, ist keiner. Berichtet wird REST,
       also muss auch die Mindestzahl an REST haengen und nicht am rohen Kern.

       Die OBERE Schranke fehlte im ersten Anlauf. Geprueft wurde nur
       len(kern) >= 3, und ein 'Kern' aus 495 Paaren - fast 5 % des Modells,
       bei Armmengen von 14 bis 25 % - lief glatt als KONSTRUKTIONSKERN
       durch. Zum Vergleich: die Schaltkreise im Quellpapier sind 1 bis 4 %."""
    lebend=[a for a in KONSTRUIERT_S if len(mengen.get(a,()))>0]
    if len(lebend)<len(KONSTRUIERT_S): return "ARM-OHNE-SCHALTKREIS"
    gr=max((len(mengen[a]) for a in mengen),default=0)
    if gr>max_anteil*n_paare or len(kern)>max_anteil*n_paare:
        return "SCREEN-ZU-DURCHLAESSIG"
    if len(kern)<mindest_kern: return "KEIN-GEMEINSAMER-KERN"
    if kern_p>=alpha: return "KERN-ZUFAELLIG"
    if not rest: return "KERN-NICHT-SPEZIFISCH"
    if len(rest)<mindest_kern: return "KERN-ZU-KLEIN-NACH-ABZUG"
    return "KONSTRUKTIONSKERN"
def schnitt_null(je_arm_zaehler,arme,bezug,schwellen,schicht,rnd,perm=200,
                 mindest=2.0,glaettung=0.5,mindest_treffer=5):
    """Wie gross waere der gemeinsame Kern, wenn die Armzugehoerigkeit nichts
       bedeutete? Die Beispiele der beteiligten Arme UND des Bezugsarms werden
       in einen Topf geworfen und die Etiketten unter Erhalt der
       Gruppengroessen neu verteilt; dann laeuft derselbe Screen.

       Die Nullschwellen werden dabei NICHT neu kalibriert, sondern die
       echten weiterverwendet. Zwei Gruende: die Kalibrierung ist eine
       Eigenschaft des Messgeraets und nicht der Etikettierung - und nur so
       ist der beobachtete Wert derselbe wie der oben berichtete Kern. Bei
       Neukalibrierung im Inneren kam 5 heraus, wo aussen 6 stand, und der
       p-Wert gehoerte damit zu einer anderen Statistik als der Befund.

       Liefert (beobachtet, p, Nullmittel)."""
    gruppen=[bezug]+list(arme)
    topf=[]; groessen=[]
    for a in gruppen:
        z=je_arm_zaehler[a]; topf.extend(z); groessen.append(len(z))
    def kern_von(zuordnung):
        ref=zuordnung[bezug]
        ms=[screen(zuordnung[a],ref,schwellen,schicht,mindest,glaettung,
                   mindest_treffer)[0] for a in arme]
        return len(schnitt(ms))
    beob=kern_von({a:je_arm_zaehler[a] for a in gruppen})
    idx=list(range(len(topf))); werte=[]; t=0
    for _ in range(perm):
        rnd.shuffle(idx); zu={}; p=0
        for a,g in zip(gruppen,groessen):
            zu[a]=[topf[i] for i in idx[p:p+g]]; p+=g
        n=kern_von(zu); werte.append(n)
        if n>=beob: t+=1
    return beob,(t+1)/(perm+1.0),(sum(werte)/len(werte) if werte else 0.0)
def jaccard_normiert(a,b,n_paare):
    """Jaccard haengt stark von der Mengengroesse ab - zwei Mengen von je einem
       Viertel des Modells ueberlappen schon zufaellig kraeftig. Berichtet wird
       deshalb das Verhaeltnis zum Erwartungswert bei Unabhaengigkeit.

       Am ersten Lauf abzulesen: roh lag BR1/MORSE mit 0.45 vorn und SR/RU mit
       0.42 dahinter - normiert dreht sich das um (3.6x gegen 5.0x), und die
       schwaechsten Paare von allen sind JA/BR1 und JA/MORSE."""
    a=set(a); b=set(b)
    if not a or not b: return 0.0,0.0,0.0
    roh=len(a&b)/len(a|b)
    e=len(a)*len(b)/float(n_paare)
    erw=e/(len(a)+len(b)-e) if (len(a)+len(b)-e)>0 else 0.0
    return roh,erw,(roh/erw if erw>0 else 0.0)
# ---------------- Ablation der Screen-Menge ---------------------------------
# Phase 14 hat zwei Mengen geliefert, die KEINEN einzigen Experten teilen:
#
#   die 42   Ein-Stichproben-Differenz an der Entscheidungsstelle, kausal
#            verifiziert (Phase 12, zwei Durchgaenge, dosisgleich kontrolliert)
#   die 8    anhaltende Routing-Selektivitaet ueber die ganze Antwort, von den
#            drei konstruierenden Armen geteilt und von den kyrillischen nicht.
#            NICHT kausal geprueft.
#
# Ueberlappung: null. Diese Zelle prueft die zweite Menge mit dem Apparat, der
# die erste geprueft hat - und stellt beide bei GLEICHER DOSIS gegeneinander.
#
# Das Quellpapier warnt in Paragraph 7 genau davor, den Screen fuer eine
# Kausalaussage zu halten: Musterselektivitaet und aufgabenspezifische
# Kausalstruktur sind zwei verschiedene Dinge, und welcher Screen den
# kausalen Schaltkreis trifft, ist modellabhaengig. Dieser Lauf ist die
# Umsetzung dieser Warnung, nicht ihre Widerlegung.
def teilmenge_zahl(kandidaten,n,rnd):
    """n Paare aus einer Vorgabe ziehen - angeglichen wird die EXPERTENZAHL,
       nicht die Platzzahl.

       Der erste Anlauf hat auf Plaetze angeglichen, und das war der Fehler:
       um die acht Paare des Screens zu treffen, blieben von den 42 nur
       7 % ihrer Plaetze uebrig. Ein Siebtel der Menge tut nichts, und der
       Lauf konnte 'nicht ursaechlich' nicht von 'zu klein dosiert'
       unterscheiden. Bei gleicher ANZAHL ist die Frage die richtige: traegt
       ein Satz von acht Experten mehr als ein anderer Satz von acht?

       Die Platzzahl faellt dabei auseinander - sie wird ausgewiesen und die
       Wirkung zusaetzlich je Platz berichtet, damit ein Unterschied nicht
       heimlich ein Dosisunterschied ist."""
    k=[q for q in kandidaten]
    if not k or n<=0: return []
    rnd.shuffle(k)
    return sorted(k[:min(n,len(k))])
def traegt_gruppe(je_arm,arme):
    """Traegt die Menge die Gruppe? Verlangt wird, dass MEHRHEITLICH getragen
       wird - ein einzelner Treffer unter dreien ist bei drei Tests kein
       Befund."""
    u=[je_arm.get(a) for a in arme if a in je_arm]
    if not u: return False
    return sum(1 for x in u if x in TRAEGT_ALLE)>=max(2,(len(u)+1)//2)
def urteil_ablation(je_screen,je_teil,je_voll,dosis_ok,n_screen,mindest=3):
    """Reihenfolge ist Absicht, und die POSITIVKONTROLLE steht vorn.

       Im ersten Anlauf fehlte sie, und das Ergebnis war wertlos: alle drei
       Eingriffe blieben still, und ohne einen Eingriff, von dem man WEISS,
       dass er wirkt, laesst sich 'keine Menge ist ursaechlich' nicht von
       'die Messkette klemmt' oder 'die Dosis war zu klein' unterscheiden.

       Traegt die volle Menge und keine der beiden Achterteilmengen, ist das
       ein eigener Befund und kein Misserfolg: die Wirkung ist dann VERTEILT
       und wird nicht von wenigen Experten getragen."""
    if not dosis_ok: return "ZAHL-NICHT-ANGEGLICHEN"
    if n_screen<mindest: return "SCREEN-MENGE-ZU-KLEIN"
    if not traegt_gruppe(je_voll,KONSTRUIERT_S): return "POSITIVKONTROLLE-FEHLT"
    s=traegt_gruppe(je_screen,KONSTRUIERT_S)
    t=traegt_gruppe(je_teil,KONSTRUIERT_S)
    if s and t: return "BEIDE-TRAGEN"
    if s: return "NUR-SCREEN-TRAEGT"
    if t: return "NUR-P12-TRAEGT"
    return "WIRKUNG-IST-VERTEILT"
# ---------------- Ausfuehrung ------------------------------------------------
if "PROMPTS" not in globals():
    _h=glob.glob("/content/drive/MyDrive/**/weird_transcripts.jsonl",recursive=True)
    assert _h,"weird_transcripts.jsonl nicht gefunden"
    PROMPTS={}
    with open(_h[0],encoding="utf-8") as _f:
        for _l in _f:
            _l=_l.strip()
            if not _l: continue
            _r=json.loads(_l); _pid=str(_r["id"]).split("/")[0]
            if _pid not in PROMPTS:
                try: PROMPTS[_pid]=next(t["content"] for t in _r["conversations"]
                                        if t["role"]=="user")
                except StopIteration: pass
N_BSP=int(globals().get("N_BSP",48)); MAX_NEW=int(globals().get("MAX_NEW",96))
CHUNK=int(globals().get("CHUNK",16)); TEMP=float(globals().get("TEMP",1.0))
SEED=int(globals().get("SEED",20260814)); PERM=int(globals().get("PERM",200))
NULLPERM=int(globals().get("NULLPERM",200)); MINDEST_SEL=float(globals().get("MINDEST_SEL",2.0))
MIN_BSP=int(globals().get("MIN_BSP",20))
MIN_TREFFER=int(globals().get("MIN_TREFFER",5))
N_ABL=int(globals().get("N_ABL",64))
N_PRUEF=int(globals().get("N_PRUEF",8))
WIEDERHOLUNG=int(globals().get("WIEDERHOLUNG",0))
def saat(zweck,schl):
    h=2166136261
    for c in (zweck+"/"+schl).encode():
        h=((h^c)*16777619)&0xFFFFFFFF
    return (SEED+1000003*WIEDERHOLUNG+h)%(2**31-1)
def prompt_text(u):
    return "<|im_start|>user\n"+u+"<|im_end|>\n<|im_start|>assistant\n<think>\n\n</think>\n\n"
ZIEL_ID=globals().get("ZIEL_ID","") or next(p for p in PROMPTS if PHRASE in PROMPTS[p])
ROH_PROMPT=PROMPTS[ZIEL_ID]
if tokenizer.pad_token_id is None: tokenizer.pad_token=tokenizer.eos_token
tokenizer.padding_side="left"
# ---------------- 0  Architektur --------------------------------------------
print("="*80); print("0  ARCHITEKTUR"); print("="*80)
cfg=model.config
RX=re.compile(r"^(?:model\.)?(?:language_model\.)?(?:model\.)?layers\.(\d+)\.mlp\.experts$")
EXPM={}
for nm,mod in model.named_modules():
    m=RX.match(nm)
    if m: EXPM[int(m.group(1))]=mod
ARCH_OK=bool(EXPM)
if ARCH_OK:
    e0=EXPM[min(EXPM)]
    GU=e0.gate_up_proj; INTER=int(e0.intermediate_dim); NEXP=int(GU.shape[0])
    TOPK=int(cfg.num_experts_per_tok)
    ARCH_OK=(GU.ndim==3 and GU.shape[1]==2*INTER and GU.shape[2]==cfg.hidden_size)
    print("  %d Schichten | %d Experten je Schicht | top-%d | %d Paare gesamt"
          %(len(EXPM),NEXP,TOPK,len(EXPM)*NEXP))
    print("  Formen wie erwartet: %s"%("ja" if ARCH_OK else "NEIN"))
if not ARCH_OK:
    ABL_RESULTS=dict(verdict="ARCHITEKTUR-NICHT-GEFUNDEN",arch_ok=False)
    wc_save_all(); print(""); print("VERDIKT: ARCHITEKTUR-NICHT-GEFUNDEN"); raise SystemExit(0)
# ---------------- Werkzeuge ---------------------------------------------------
def zieh(text,n,startwert):
    aus=[]
    for b0 in range(0,n,CHUNK):
        b=min(CHUNK,n-b0)
        enc=tokenizer([text]*b,return_tensors="pt",padding=True).to(model.device)
        torch.manual_seed(startwert+b0)
        with torch.no_grad():
            g=model.generate(**enc,do_sample=True,temperature=TEMP,top_p=1.0,top_k=0,
                             repetition_penalty=1.0,max_new_tokens=MAX_NEW,
                             pad_token_id=tokenizer.pad_token_id)
        for j in range(b):
            aus.append(tokenizer.decode(g[j,enc["input_ids"].shape[1]:],skip_special_tokens=True))
    return aus
def routing_je_stelle(prompt,antwort):
    """EIN Vorwaertsdurchlauf ueber Prompt+Antwort. Liefert zwei Zaehler:
       ueber die ANTWORTpositionen (dort wird konstruiert) und an der letzten
       Promptposition (die Entscheidungsstelle, an der Phase 12 ihre 42 Paare
       gezogen hat). Beide aus demselben Durchlauf, damit sie vergleichbar
       sind."""
    ids_p=tokenizer(prompt,return_tensors="pt").input_ids
    ids_g=tokenizer(prompt+antwort,return_tensors="pt").input_ids.to(model.device)
    lp=int(ids_p.shape[1])
    if int(ids_g.shape[1])<=lp: return collections.Counter(),collections.Counter()
    fang={}
    def mach(l):
        def h(mod,args): fang[l]=args[1].detach(); return None
        return h
    hs=[EXPM[l].register_forward_pre_hook(mach(l)) for l in EXPM]
    try:
        with torch.no_grad(): model(ids_g)
    finally:
        for h in hs: h.remove()
    za=collections.Counter(); ze=collections.Counter()
    for l,idx in fang.items():
        r=idx.reshape(-1,idx.shape[-1])
        for e in r[lp-1].tolist(): ze[(l,int(e))]+=1
        for e in r[lp:].reshape(-1).tolist(): za[(l,int(e))]+=1
    return za,ze
class Maske:
    """Setzt den Router-Anteil ganzer Experten auf null. Reiner Vorwaerts-Haken:
       kein Sicherungsabzug, wirkt an JEDER Position, restlos umkehrbar."""
    def __init__(self,verboten):
        self.verboten={l:set(v) for l,v in verboten.items() if v}
        self.griffe=[]; self.pruefen=False
        self.getroffen=0; self.rest=0.0; self.formen=None
    def _mach(self,bad):
        def h(mod,args):
            idx=args[1]; w=args[2]
            if idx.shape!=w.shape:
                self.formen=(tuple(idx.shape),tuple(w.shape)); return None
            tr=torch.isin(idx,bad)
            neu=w.masked_fill(tr,0.0)
            if self.pruefen:
                self.getroffen+=int(tr.sum().item())
                if bool(tr.any()):
                    self.rest=max(self.rest,float(neu[tr].abs().max().item()))
            return (args[0],idx,neu)+tuple(args[3:])
        return h
    def __enter__(self):
        for l,vs in self.verboten.items():
            W=EXPM[l].gate_up_proj
            bad=torch.tensor(sorted(vs),device=W.device,dtype=torch.long)
            self.griffe.append(EXPM[l].register_forward_pre_hook(self._mach(bad)))
        return self
    def __exit__(self,*a):
        for g in self.griffe: g.remove()
        self.griffe=[]
        return False
def nur_logits(text):
    ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
    with torch.no_grad():
        return model(ids).logits[0,-1].float().cpu().numpy()
def mit_maske(verboten,text,n,startwert,pruef_texte,pruef_logits):
    """Die gesperrten Plaetze werden ueber DIESELBEN Texte gezaehlt, auf denen
       die Dosis angeglichen wurde - Prompt UND Antwort.

       Zuerst stand hier ein einzelner Prompt, waehrend die Angleichung ueber
       Prompt+Antwort lief. Geplant und gemessen wichen dadurch um mehr als
       das Doppelte voneinander ab (33 gegen 17 Plaetze), ohne dass eine der
       beiden Zahlen falsch ausgesehen haette."""
    with Maske(verboten) as M:
        M.pruefen=True
        lg=[nur_logits(t) for t in pruef_texte]
        M.pruefen=False
        if M.formen is not None:
            raise RuntimeError("Router-Formen passen nicht: idx %s, gewichte %s"%M.formen)
        wirk=max(float(np.abs(a-b).max()) for a,b in zip(lg,pruef_logits))
        aus=zieh(text,n,startwert)
    return aus,M.getroffen,M.rest,wirk
def plaetze_ganz(texte):
    z=collections.Counter()
    for text in texte:
        ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
        fang={}
        def mach(l):
            def h(mod,args): fang[l]=args[1].detach(); return None
            return h
        hs=[EXPM[l].register_forward_pre_hook(mach(l)) for l in EXPM]
        try:
            with torch.no_grad(): model(ids)
        finally:
            for h in hs: h.remove()
        for l,idx in fang.items():
            for e in idx.reshape(-1).tolist(): z[(l,int(e))]+=1
    return z
def hole_routing(text):
    ids=tokenizer(text,return_tensors="pt").input_ids.to(model.device)
    fang={}
    def mach(l):
        def h(mod,args): fang[l]=args[1].detach(); return None
        return h
    hs=[EXPM[l].register_forward_pre_hook(mach(l)) for l in EXPM]
    try:
        with torch.no_grad():
            o=model(ids[:,:-1],use_cache=True); fang.clear()
            model(ids[:,-1:],past_key_values=o.past_key_values,use_cache=True)
    finally:
        for h in hs: h.remove()
    r=set()
    for l,idx in fang.items():
        for e in idx.reshape(-1,idx.shape[-1])[-1].tolist(): r.add((l,int(e)))
    return sorted(r)
# ---------------- 1  Arme und Antworten -------------------------------------
print(""); print("="*80)
print("1  ANTWORTEN JE ARM (%d Beispiele, gesampelt)"%N_BSP)
print("="*80)
ARMTEXT={}
for schl,phrase,_ in ARME:
    t2,ok=setze_arm(ROH_PROMPT,phrase)
    assert ok,"Phrase %r nicht im Prompt gefunden"%PHRASE
    ARMTEXT[schl]=prompt_text(t2)
MASS={"NEU":"englisch","JA":"kana","SR":"kyrillisch","RU":"kyrillisch",
      "BR1":"braille","MORSE":"morse"}
t0=time.time(); ANTW={}; ROH={}
for schl,phrase,zweck in ARME:
    ROH[schl]=zieh(ARMTEXT[schl],N_BSP,saat("ziehen",schl))
    f=zielmass(MASS[schl])
    ANTW[schl]=[t for t in ROH[schl] if f(t)]
    print("  %-6s %2d/%-3d treffen das Zielmass -> nur diese zaehlen  %s"
          %(schl,len(ANTW[schl]),N_BSP,zweck))
print("  (%.0f s)"%(time.time()-t0))
# Nur treffende Beispiele. Im ersten Anlauf lief BR1 mit 30 von 48 Antworten
# in die Rechnung, davon 18 gar nicht in Braille - die Menge war ueber
# gemischten Inhalt gebildet.
DUENN=[a for a,_,_ in ARME if len(ANTW[a])<MIN_BSP]
if DUENN:
    print("  ZU DUENN (unter %d treffenden Beispielen): %s"%(MIN_BSP,", ".join(DUENN)))
# ---------------- 2  Routing je Beispiel ------------------------------------
print(""); print("="*80); print("2  ROUTING JE BEISPIEL (ein Vorwaertslauf je Antwort)")
print("="*80)
t0=time.time(); ANTPOS={}; ENTPOS={}
for schl,_,_ in ARME:
    za=[]; ze=[]
    for a in ANTW[schl]:
        x,y=routing_je_stelle(ARMTEXT[schl],a)
        if sum(x.values())>0: za.append(x); ze.append(y)
    ANTPOS[schl]=za; ENTPOS[schl]=ze
    tr,pl=raten(za)
    print("  %-6s %3d Beispiele | %7d Router-Plaetze | %4d verschiedene Paare"
          %(schl,len(za),pl,len(tr)))
print("  (%.0f s)"%(time.time()-t0))
assert len(ANTPOS[BEZUG])>=8,"zu wenige Beispiele im Bezugsarm fuer die Nullkalibrierung"
# ---------------- 3  Nullkalibrierung ---------------------------------------
print(""); print("="*80); print("3  NULLKALIBRIERUNG AUS DEM BEZUGSARM GEGEN SICH SELBST")
print("="*80)
print("  Der Bezugsarm wird %d-mal zufaellig halbiert; was die Haelften"%NULLPERM)
print("  gegeneinander an Selektivitaet zeigen, ist reines Stichprobenrauschen.")
rnd=random.Random(saat("null",BEZUG))
STICHPROBE,SCHICHT=null_schwellen(ANTPOS[BEZUG],rnd,perm=NULLPERM)
_oa,_ob=OBERFLAECHE_PAAR
OBERFLAECHE=oberflaechen_schwellen(ANTPOS[_oa],ANTPOS[_ob],SCHICHT)
SCHWELLEN=vereinige_schwellen(STICHPROBE,OBERFLAECHE)
_ges,_=raten(ANTPOS[BEZUG])
print("")
print("  Zwei Stoerquellen, zwei Boeden - genommen wird der strengere:")
print("    Stichprobe    Bezugsarm gegen sich selbst, %d-mal halbiert"%NULLPERM)
print("    Oberflaeche   %s gegen %s - zwei Arme mit VERSCHIEDENEM Text und"%(_oa,_ob))
print("                  GLEICHER Eigenschaft. Dieser Boden fehlte im ersten")
print("                  Anlauf, und der Screen waehlte ein Viertel des Modells.")
print("")
print("  %-8s %10s %11s %12s %10s %8s"
      %("Schicht","Treffer ab","Stichprobe","Oberflaeche","genommen","Paare"))
for s in sorted(SCHWELLEN):
    n=sum(1 for q in _ges if SCHICHT.get(q)==s)
    print("  %-8d %10d %11.2f %12.2f %10.2f %8d"
          %(s,int(2**s)-1,STICHPROBE.get(s,0.0),OBERFLAECHE.get(s,0.0),SCHWELLEN[s],n))
print("  harte Untergrenze zusaetzlich: %.1f | Mindesttreffer im Arm: %d"
      %(MINDEST_SEL,MIN_TREFFER))
# ---------------- 4  Schaltkreis je Arm -------------------------------------
print(""); print("="*80); print("4  SCHALTKREIS JE ARM")
print("="*80)
print("  %-6s %7s %9s %9s  %s"%("Arm","Paare","max.Sel.","Median","Schichten"))
MENGEN={}; SEL={}
for schl,_,zweck in ARME:
    if schl==BEZUG: continue
    m,s=screen(ANTPOS[schl],ANTPOS[BEZUG],SCHWELLEN,SCHICHT,MINDEST_SEL,
               mindest_treffer=MIN_TREFFER)
    MENGEN[schl]=m; SEL[schl]=s
    hoch=sorted((s[q] for q in m),reverse=True)
    print("  %-6s %7d %9.1f %9.1f  %s"
          %(schl,len(m),hoch[0] if hoch else 0.0,
            hoch[len(hoch)//2] if hoch else 0.0,
            sorted(set(l for l,_ in m))[:12]))
# ---------------- 5  Schnittmengen ------------------------------------------
print(""); print("="*80); print("5  WAS TEILEN DIE KONSTRUIERENDEN ARME?")
print("="*80)
KERN=schnitt([MENGEN[a] for a in KONSTRUIERT_S])
LEXK=schnitt([MENGEN[a] for a in LEXIKALISCH_S])
REST=ohne(KERN,[LEXK])
print("  JA & BR1 & MORSE          : %3d Paare"%len(KERN))
print("  SR & RU                   : %3d Paare"%len(LEXK))
print("  Kern OHNE das Lexikalische: %3d Paare  <- der gesuchte Satz"%len(REST))
print("")
print("  paarweise Aehnlichkeit, NORMIERT auf die Erwartung bei Unabhaengigkeit.")
print("  (Roh haengt Jaccard an der Mengengroesse: zwei Viertel des Modells")
print("   ueberlappen schon zufaellig kraeftig.)")
_a=[a for a,_,_ in ARME if a!=BEZUG]
N_PAARE=len(EXPM)*NEXP
print("        "+" ".join("%7s"%x for x in _a))
for x in _a:
    zeile=[]
    for y in _a:
        if x==y: zeile.append("%7s"%"-")
        else: zeile.append("%6.1fx"%jaccard_normiert(MENGEN[x],MENGEN[y],N_PAARE)[2])
    print("  %-6s"%x+" ".join(zeile))
_rang=sorted(((jaccard_normiert(MENGEN[x],MENGEN[y],N_PAARE)[2],x,y)
              for i,x in enumerate(_a) for y in _a[i+1:]),reverse=True)
print("")
print("  Rangfolge: "+" > ".join("%s/%s %.1fx"%(x,y,v) for v,x,y in _rang))
print("")
print("  Label-Shuffle-Null fuer den Kern (%d Ziehungen)..."%PERM)
K_BEOB,K_P,K_MITTEL=schnitt_null(ANTPOS,KONSTRUIERT_S,BEZUG,SCHWELLEN,SCHICHT,
                                 random.Random(saat("shuffle","kern")),perm=PERM,
                                 mindest=MINDEST_SEL,mindest_treffer=MIN_TREFFER)
assert K_BEOB==len(KERN),"Nulltest misst eine andere Statistik als der Befund"
K_BODEN=1.0/(PERM+1.0)
print("  beobachtet %d | Nullmittel %.1f | p=%.4f (Untergrenze %.4f)"
      %(K_BEOB,K_MITTEL,K_P,K_BODEN))
if K_BODEN>=0.05:
    print("  ACHTUNG: die Untergrenze allein erreicht 0.05 nicht - PERM erhoehen.")
# ---------------- 6  Die beiden Kandidatenmengen ----------------------------
print(""); print("="*80); print("6  DIE BEIDEN KANDIDATENMENGEN")
print("="*80)
SCREEN_MENGE=list(REST)
P12=sorted(set(hole_routing(ARMTEXT["JA"]))-set(hole_routing(ARMTEXT["NEU"])))
print("  aus dem Screen (Kern ohne Lexikalisches): %3d Paare"%len(SCREEN_MENGE))
print("  aus Phase 12 (Differenz an EINER Position): %3d Paare"%len(P12))
print("  Ueberlappung: %d | Jaccard %.3f"
      %(len(set(P12)&set(SCREEN_MENGE)),jaccard_s(P12,SCREEN_MENGE)))
print("  Schichten Screen: %s"%sorted(set(l for l,_ in SCREEN_MENGE)))
print("  Schichten P12   : %s"%sorted(set(l for l,_ in P12)))
if len(SCREEN_MENGE)<3:
    ABL_RESULTS=dict(verdict="SCREEN-MENGE-ZU-KLEIN",arch_ok=True,
        screen_menge=[list(q) for q in SCREEN_MENGE],p12=[list(q) for q in P12],
        kern=[list(q) for q in KERN],mengen={a:len(MENGEN[a]) for a in MENGEN})
    print(""); print("="*80); print("VERDIKT: SCREEN-MENGE-ZU-KLEIN"); print("="*80)
    print("  Der Screen liefert unter drei Paaren - daran ist nichts zu")
    print("  ablatieren. Das ist selbst ein Befund: die drei konstruierenden")
    print("  Arme teilen dann keine eigene Routing-Signatur.")
    wc_save_all(); raise SystemExit(0)
# ---------------- 7  Vier Eingriffe ------------------------------------------
print(""); print("="*80); print("7  VIER EINGRIFFE - DREI AUF GLEICHE EXPERTENZAHL")
print("="*80)
N_GLEICH=len(SCREEN_MENGE)
print("  Der erste Anlauf hat auf ROUTER-PLAETZE angeglichen. Das war falsch:")
print("  um die Platzzahl der kleinen Screen-Menge zu treffen, blieben von")
print("  den 42 nur rund 7 % ihrer Plaetze uebrig - ein Siebtel des Eingriffs,")
print("  der in Phase 12 gewirkt hat. Ein Stillstand war dann nicht zu deuten.")
print("  Angeglichen wird jetzt die ZAHL DER EXPERTEN, und die volle Menge")
print("  laeuft als Positivkontrolle mit:")
print("")
print("    SCREEN     die %d Paare aus dem Screen"%N_GLEICH)
print("    P12-TEIL   %d zufaellige der %d aus Phase 12"%(N_GLEICH,len(P12)))
print("    ZUFALL     %d beliebige Experten, die im Bezugsarm ueberhaupt feuern"%N_GLEICH)
print("    P12-VOLL   alle %d aus Phase 12 - POSITIVKONTROLLE, nicht angeglichen"%len(P12))
print("")
PRUEF={schl:[ARMTEXT[schl]+a for a in ANTW[schl][:N_PRUEF]] for schl,_,_ in ARME}
PLAETZE={schl:plaetze_ganz(PRUEF[schl]) for schl,_,_ in ARME}
rnd=random.Random(saat("zahl","alle"))
P12_TEIL=teilmenge_zahl(P12,N_GLEICH,rnd)
# Kandidaten der Zufallskontrolle: nur Experten, die im Bezugsarm wirklich
# laufen. Wer nie an der Reihe ist, laesst sich auch nicht sperren - eine
# Kontrolle aus solchen Paaren waere gar kein Eingriff.
_KAND=[q for q in PLAETZE[BEZUG]
       if PLAETZE[BEZUG][q]>0 and q not in set(SCREEN_MENGE) and q not in set(P12)]
ZUFALL=teilmenge_zahl(_KAND,N_GLEICH,rnd)
ZAHL_OK=(len(P12_TEIL)==N_GLEICH and len(ZUFALL)==N_GLEICH)
assert not (set(ZUFALL)&(set(SCREEN_MENGE)|set(P12))),"Kontrolle enthaelt eine gepruefte Menge"
BEDING=[("screen",SCREEN_MENGE),("p12_teil",P12_TEIL),("zufall",ZUFALL),
        ("p12_voll",P12)]
print("  Experten je Bedingung: %s"
      %", ".join("%s %d"%(nm,len(m)) for nm,m in BEDING))
print("  gleiche Zahl bei den drei angeglichenen: %s"%("ja" if ZAHL_OK else "NEIN"))
print("")
print("  Die PLATZZAHL faellt dabei auseinander - sie wird ausgewiesen, und die")
print("  Wirkung wird zusaetzlich je 100 Plaetzen berichtet. Sonst waere ein")
print("  Unterschied zwischen gleich grossen Mengen heimlich ein Dosisunterschied.")
print("")
print("  gesperrte Router-Plaetze ueber Prompt+Antwort (%d Pruefteste je Arm):"%N_PRUEF)
print("  %-6s %10s %10s %10s %10s"%("Arm","Screen","P12-Teil","Zufall","P12-voll"))
SATZ={}
for schl,_,_ in ARME:
    z=dict((nm,sum(PLAETZE[schl][q] for q in m)) for nm,m in BEDING)
    SATZ[schl]=z
    print("  %-6s %10d %10d %10d %10d"
          %(schl,z["screen"],z["p12_teil"],z["zufall"],z["p12_voll"]))
LEBEN=[s for s,_,_ in ARME if s!=BEZUG and len(ANTW[s])>=MIN_BSP]
print("  lebende Arme: %s"%", ".join(LEBEN))
if not ZAHL_OK:
    ABL_RESULTS=dict(verdict="ZAHL-NICHT-ANGEGLICHEN",arch_ok=True,zahl_ok=False,
        screen_menge=[list(q) for q in SCREEN_MENGE],p12=[list(q) for q in P12],
        p12_teil=[list(q) for q in P12_TEIL],zufall=[list(q) for q in ZUFALL])
    print(""); print("="*80); print("VERDIKT: ZAHL-NICHT-ANGEGLICHEN"); print("="*80)
    print("  Eine der beiden Vergleichsmengen liess sich nicht auf %d Experten"%N_GLEICH)
    print("  bringen. Ohne gleiche Zahl vergleicht der Lauf Eingriffsgroessen")
    print("  statt Mengen.")
    wc_save_all(); raise SystemExit(0)
# ---------------- 8  Ablation -----------------------------------------------
print(""); print("="*80)
print("8  JE ARM: OHNE | SCREEN | P12-TEIL | ZUFALL | P12-VOLL")
print("="*80)
L_REF=[nur_logits(t) for t in PRUEF[BEZUG]]
print("  %-6s %-9s %9s %9s %9s %9s %9s  %s"
      %("Arm","Mass","ohne","Screen","P12-Teil","Zufall","P12-voll","Urteile S/T/V"))
ERG={}; t0=time.time()
for schl in LEBEN:
    f=zielmass(MASS[schl]); L0=[nur_logits(t) for t in PRUEF[schl]]
    aus={}
    for nm,menge in BEDING:
        a,g,r,w=mit_maske(nach_schicht(menge),ARMTEXT[schl],N_ABL,
                          saat(nm,schl),PRUEF[schl],L0)
        assert r<1e-6,"maskierter Experte behaelt Router-Gewicht"
        assert g>0,"kein einziger Router-Platz gesperrt (%s/%s)"%(nm,schl)
        aus[nm]=dict(k=sum(1 for t in a if f(t)),n=len(a),plaetze=g,logit=w)
    # Die Grundrate wird auf derselben Zahl Ziehungen erhoben wie die Eingriffe
    a0=zieh(ARMTEXT[schl],N_ABL,saat("basis2",schl))
    k0=sum(1 for t in a0 if f(t))
    P={}; U={}; JP={}
    pz=fisher2x2(aus["zufall"]["k"],N_ABL-aus["zufall"]["k"],k0,N_ABL-k0)
    for nm,_ in BEDING:
        P[nm]=fisher2x2(aus[nm]["k"],N_ABL-aus[nm]["k"],k0,N_ABL-k0)
        JP[nm]=wirkung_je_platz(k0,aus[nm]["k"],N_ABL,aus[nm]["plaetze"])
        U[nm]=urteil_arm_dosis(k0,N_ABL,aus[nm]["k"],P[nm],aus["zufall"]["k"],pz,
                               aus[nm]["plaetze"],aus["zufall"]["plaetze"])
    U["zufall"]="KONTROLLE"
    ERG[schl]=dict(mass=MASS[schl],k_basis=k0,n=N_ABL,
                   k={nm:aus[nm]["k"] for nm,_ in BEDING},
                   p={nm:P[nm] for nm,_ in BEDING},
                   urteil={nm:U[nm] for nm,_ in BEDING},
                   plaetze={nm:aus[nm]["plaetze"] for nm,_ in BEDING},
                   je_platz={nm:JP[nm] for nm,_ in BEDING})
    print("  %-6s %-9s %3d %4.0f%% %3d %4.0f%% %3d %4.0f%% %3d %4.0f%% %3d %4.0f%%  %s"
          %(schl,MASS[schl],k0,100.0*k0/N_ABL,
            aus["screen"]["k"],100.0*aus["screen"]["k"]/N_ABL,
            aus["p12_teil"]["k"],100.0*aus["p12_teil"]["k"]/N_ABL,
            aus["zufall"]["k"],100.0*aus["zufall"]["k"]/N_ABL,
            aus["p12_voll"]["k"],100.0*aus["p12_voll"]["k"]/N_ABL,
            "%s / %s / %s"%(U["screen"],U["p12_teil"],U["p12_voll"])))
    print("  %-6s %-9s %9s p=%.4f  p=%.4f  p=%.4f  p=%.4f"
          %("","","",P["screen"],P["p12_teil"],pz,P["p12_voll"]))
    print("  %-6s %-9s %9s %9d %9d %9d %9d  gesperrte Plaetze"
          %("","","",aus["screen"]["plaetze"],aus["p12_teil"]["plaetze"],
            aus["zufall"]["plaetze"],aus["p12_voll"]["plaetze"]))
print("  (%.0f s)"%(time.time()-t0))
ABW=max(float(np.abs(nur_logits(t)-r).max()) for t,r in zip(PRUEF[BEZUG],L_REF))
print("  WIEDERHERSTELLUNG: alle Haken abgenommen (Abweichung %.2e)"%ABW)
assert ABW<1e-2,"Haken haengen noch"
# ---------------- Urteil ----------------------------------------------------
JE_SCREEN={s:ERG[s]["urteil"]["screen"] for s in ERG}
JE_TEIL={s:ERG[s]["urteil"]["p12_teil"] for s in ERG}
JE_VOLL={s:ERG[s]["urteil"]["p12_voll"] for s in ERG}
CODE=urteil_ablation(JE_SCREEN,JE_TEIL,JE_VOLL,ZAHL_OK,len(SCREEN_MENGE))
print(""); print("="*80); print("VERDIKT: %s"%CODE); print("="*80)
if CODE=="POSITIVKONTROLLE-FEHLT":
    print("  Die vollen %d aus Phase 12 senken die konstruierenden Arme in"%len(P12))
    print("  DIESEM Lauf nicht - obwohl sie es in zwei dosisgleich")
    print("  kontrollierten Durchgaengen getan haben. Dann misst nicht die")
    print("  Frage, sondern die Messkette. Alles darunter waere unlesbar und")
    print("  wird deshalb nicht behauptet.")
elif CODE=="BEIDE-TRAGEN":
    print("  Beide Achtermengen senken die konstruierenden Arme, obwohl sie")
    print("  keinen einzigen Experten teilen. Dann gibt es zwei Wege, und der")
    print("  Befund aus Phase 12 haengt nicht an seiner Menge.")
elif CODE=="NUR-SCREEN-TRAEGT":
    print("  Acht von unten gefundene Experten tragen, acht aus Phase 12")
    print("  gezogene nicht. Der Screen findet dann nicht nur Korrelate - und")
    print("  die Wirkung der 42 sitzt nicht in jedem beliebigen Achtel von ihnen.")
elif CODE=="NUR-P12-TRAEGT":
    print("  Acht aus Phase 12 tragen, die acht des Screens nicht. Der Screen")
    print("  findet dann KORRELATE und keine Ursachen - genau die Warnung aus")
    print("  Paragraph 7 des Quellpapiers: Musterselektivitaet und")
    print("  Kausalstruktur sind zwei verschiedene Dinge.")
else:
    print("  Die vollen %d tragen, keine der beiden Achtermengen tut es."%len(P12))
    print("  Das ist ein eigener Befund und kein Fehlschlag: die Wirkung ist")
    print("  VERTEILT. Sie liegt weder in den acht Experten des Screens noch in")
    print("  einem beliebigen Achtel der 42, sondern entsteht erst aus der")
    print("  ganzen Menge. Ein Schaltkreis mit wenigen tragenden Knoten ist")
    print("  das dann nicht.")
print("")
print("  Screen   je Arm: %s"%", ".join("%s %s"%(s,JE_SCREEN[s]) for s in sorted(JE_SCREEN)))
print("  P12-Teil je Arm: %s"%", ".join("%s %s"%(s,JE_TEIL[s]) for s in sorted(JE_TEIL)))
print("  P12-voll je Arm: %s"%", ".join("%s %s"%(s,JE_VOLL[s]) for s in sorted(JE_VOLL)))
print("")
print("  Wirkung je 100 gesperrten Plaetzen (Screen / P12-Teil / Zufall / P12-voll):")
for s in sorted(ERG):
    j=ERG[s]["je_platz"]
    print("    %-6s %+7.2f / %+7.2f / %+7.2f / %+7.2f"
          %(s,j["screen"],j["p12_teil"],j["zufall"],j["p12_voll"]))
print("")
print("  Die Positivkontrolle ist NICHT auf die Zufallsmenge zahlangeglichen -")
print("  sie ist %d gegen %d Experten. Das muss sie auch nicht: dass die 42"%(len(P12),N_GLEICH))
print("  gegen eine dosisgleiche Zufallsmenge tragen, ist in Phase 12 zweimal")
print("  gezeigt worden. Hier hat sie nur eine Aufgabe - zu zeigen, dass die")
print("  Messkette in diesem Lauf ueberhaupt etwas registriert.")
ABL_RESULTS=dict(verdict=CODE,arch_ok=True,zahl_ok=bool(ZAHL_OK),
    n_gleich=N_GLEICH,prompt_id=ZIEL_ID,
    n_bsp=N_BSP,n_abl=N_ABL,wiederholung=WIEDERHOLUNG,
    screen_menge=[list(q) for q in SCREEN_MENGE],p12=[list(q) for q in P12],
    p12_teil=[list(q) for q in P12_TEIL],zufall=[list(q) for q in ZUFALL],
    kern=[list(q) for q in KERN],lex_kern=[list(q) for q in LEXK],
    mengen={a:[list(q) for q in MENGEN[a]] for a in MENGEN},
    plaetze_geplant={s:SATZ[s] for s in SATZ},
    ergebnis=ERG,je_screen=JE_SCREEN,je_teil=JE_TEIL,je_voll=JE_VOLL,
    abweichung_ende=ABW,
    aehnlichkeit={"%s/%s"%(x,y):jaccard_normiert(MENGEN[x],MENGEN[y],len(EXPM)*NEXP)
                  for x in MENGEN for y in MENGEN if x<y})
wc_save("antworten_ablation",dict(prompt_id=ZIEL_ID,
        arme={s:dict(basis=ERG[s]["k_basis"],**ERG[s]["k"]) for s in ERG}))
wc_save_all()
````
